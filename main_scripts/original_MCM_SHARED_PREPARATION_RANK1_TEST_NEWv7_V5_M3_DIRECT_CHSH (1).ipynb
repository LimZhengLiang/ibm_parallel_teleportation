```python
import copy
from importlib.metadata import PackageNotFoundError, version
from pathlib import Path
import json
import re
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t as student_t
from scipy.linalg import hadamard

from qiskit import (
    QuantumCircuit, QuantumRegister, ClassicalRegister, transpile,
    generate_preset_pass_manager,
)
from qiskit.circuit.equivalence_library import (
    EquivalenceLibrary, SessionEquivalenceLibrary as sel,
)
from qiskit.circuit.library import XGate, YGate
from qiskit.quantum_info import Statevector
from qiskit.primitives.containers import BitArray, SamplerPub
from qiskit.circuit.controlflow import ControlFlowOp
from qiskit.transpiler import InstructionProperties, PassManager
from qiskit.transpiler.passes import BasisTranslator
from qiskit_aer import AerSimulator
from qiskit_experiments.framework import ExperimentData
from qiskit_experiments.library import StateTomography
from contextlib import ExitStack
import atexit
from qiskit_ibm_runtime import (
    QiskitRuntimeService, SamplerV2 as Sampler, Session,
)
from qiskit_ibm_runtime.circuit import MidCircuitMeasure
from qiskit_ibm_runtime.transpiler.passes.scheduling import (
    ALAPScheduleAnalysis, PadDelay, PadDynamicalDecoupling,
)

# A hardware execution cell gets one dedicated Runtime Session.  ExitStack
# enters the Session context, and the IPython post-run hook exits it even if
# the cell raises.  All hardware samplers created in that cell share it.
_HARDWARE_SESSION_STACK = None
_HARDWARE_SESSIONS = {}
_HARDWARE_SESSION_CLEANUP_REGISTERED = False


def close_hardware_sessions(_result=None):
    """Close this cell's Runtime sessions and release dedicated access."""
    global _HARDWARE_SESSION_STACK
    global _HARDWARE_SESSIONS
    global _HARDWARE_SESSION_CLEANUP_REGISTERED
    stack = _HARDWARE_SESSION_STACK
    _HARDWARE_SESSION_STACK = None
    _HARDWARE_SESSIONS = {}
    close_error = None
    if stack is not None:
        try:
            stack.close()
        except Exception as exc:
            close_error = exc
    if _HARDWARE_SESSION_CLEANUP_REGISTERED:
        try:
            shell = get_ipython()
            if shell is not None:
                shell.events.unregister(
                    "post_run_cell", close_hardware_sessions
                )
        except (NameError, ValueError):
            pass
        _HARDWARE_SESSION_CLEANUP_REGISTERED = False
    if close_error is not None:
        raise close_error


def make_hardware_session_sampler(backend):
    """Return SamplerV2 in a shared, cell-scoped Runtime Session."""
    global _HARDWARE_SESSION_STACK
    global _HARDWARE_SESSION_CLEANUP_REGISTERED
    if _HARDWARE_SESSION_STACK is None:
        _HARDWARE_SESSION_STACK = ExitStack()
        try:
            shell = get_ipython()
        except NameError:
            shell = None
        if shell is not None:
            shell.events.register(
                "post_run_cell", close_hardware_sessions
            )
            _HARDWARE_SESSION_CLEANUP_REGISTERED = True
        else:
            atexit.register(close_hardware_sessions)
    backend_name = getattr(backend, "name", str(backend))
    if callable(backend_name):
        backend_name = backend_name()
    if backend_name not in _HARDWARE_SESSIONS:
        _HARDWARE_SESSIONS[backend_name] = (
            _HARDWARE_SESSION_STACK.enter_context(
                Session(backend=backend)
            )
        )
        print(
            f"Dedicated IBM Runtime Session opened on {backend_name}; "
            "it will close when this execution cell finishes.",
            flush=True,
        )
    session = _HARDWARE_SESSIONS[backend_name]
    return Sampler(mode=session)
import mthree
MTHREE_IMPLEMENTATION = "mthree.M3Mitigation"
import mapomatic as mm

print("Imports complete.")

# ---------------------------- Configuration ----------------------------
EXPECTED_RUNTIME_VERSION = "0.43.0"
EXPERIMENT_VARIANT = "MCM measure_2 three-round cascade DD sweep + state-fidelity heatmaps"
BACKEND_NAME = "ibm_pittsburgh"
DISENTANGLER_RANK = 1  # moved up from below so OUTPUT_DIR can be tagged by rank
N_TRIALS = 1  # Independent acquisitions per level/channel/round/DD policy.
# Balanced Mapomatic limits for every A/B state-analysis pair.
# Pair risk = 1 - (1-readout_error_left)*(1-readout_error_right).
LAYOUT_MAX_SOURCE_PAIR_READOUT_RISK = 0.12
LAYOUT_MAX_ANALYSIS_PAIR_READOUT_RISK = 0.18
LAYOUT_MAX_CHANNEL_READOUT_RISK_GAP = 0.06
LAYOUT_REPICK_ATTEMPTS = 8
LAYOUT_MEAN_READOUT_WEIGHT = 1.0
LAYOUT_WORST_READOUT_WEIGHT = 2.0
LAYOUT_CHANNEL_BALANCE_WEIGHT = 1.0
SHOTS = 5000
M3_CALIBRATION_SHOTS = 5000
OPTIMIZATION_LEVEL = 2  # MINIMAL VERIFICATION RUN: was 3 - testing whether a lower transpiler level changes round-2/3 XY4 behavior
SEED_TRANSPILER = 42

# All files are written here. Change this before execution if desired.
OUTPUT_ROOT = Path.cwd() / "epr_mcm"
OUTPUT_DIR = OUTPUT_ROOT / BACKEND_NAME / f'rank{DISENTANGLER_RANK}'
HARDWARE_CALIBRATION_HISTORY_PATH = Path.cwd() / 'hardware_calibration_history.txt'  # shared across all four notebooks; append-only

MCM_DD_AUDIT_CSV = OUTPUT_DIR / "mcm_dynamic_dd_schedule_audit.csv"

# Safety guard: set True only when intentionally submitting hardware jobs.
RUN_EXPERIMENT = True
SHOW_SCHEDULER_TIMING = False  # Scheduler plotting disabled by request.
MAX_TIMING_PLOTS = 12
STATUS_POLL_SECONDS = 15
MAINTENANCE_POLL_SECONDS = 600  # Recheck backend availability every 10 min.
RUNTIME_INSTANCE = None  # Search all accessible projects for BACKEND_NAME.
JOB_MAX_ATTEMPTS = 5
JOB_RETRY_BASE_SECONDS = 30
RETRYABLE_IBM_ERROR_CODES = (
    {
        # IBM's error registry explicitly recommends retrying these
        # availability, queue-publication, rate-limit, timeout, internal,
        # and job-state synchronization failures.
        1004, 1011, 1202, 1245, 1302, 1307, 1319, 1323,
        1808, 1809, 3202, 3203, 3208, 3212, 3213, 3214,
        3245, 3246, 3259, 3704, 3900, 3911,
        4300, 4301, 4303, 4305, 4308, 4314,
        5203, 6072, 9999,
    }
    | set(range(6051, 6069))
    | set(range(9700, 9711))
)
BACKEND_PENDING_WARNING_THRESHOLD = 500
ENABLE_DD = False
DD_SEQUENCE = "XpXm"
DD_SCHEDULING_METHOD = "alap"
DD_POLICIES = ["none", "XY4", "XX"]
TIMING_PLOTS_PER_POLICY = 4

CONDITION_ORDER = ["a_real", "b_real_m3", "c_sim", "d_sim_m3"]
CONDITION_LABELS = {
    "a_real": "(a) Real HW",
    "b_real_m3": "(b) Real HW + M3",
    "c_sim": "(c) Sim (standard-measure proxy)",
    "d_sim_m3": "(d) Sim proxy + M3",
}

OUTPUT_NAMES = {
    ("A", "fidelity"): "corrected_epr_entanglement_fidelity_channel_A.csv",
    ("B", "fidelity"): "corrected_epr_entanglement_fidelity_channel_B.csv",
    ("A", "chsh"): "corrected_epr_chsh_channel_A.csv",
    ("B", "chsh"): "corrected_epr_chsh_channel_B.csv",
    ("A", "fidelity_png"): "corrected_epr_entanglement_fidelity_channel_A.png",
    ("B", "fidelity_png"): "corrected_epr_entanglement_fidelity_channel_B.png",
    ("A", "chsh_png"): "corrected_epr_chsh_channel_A.png",
    ("B", "chsh_png"): "corrected_epr_chsh_channel_B.png",
}

print(f"RUN_EXPERIMENT={RUN_EXPERIMENT}")
print(f"Outputs: {OUTPUT_DIR.resolve()}")


def require_runtime_043():
    """Fail before submission unless the requested Runtime version is active."""
    try:
        installed = version("qiskit-ibm-runtime")
    except PackageNotFoundError as exc:
        raise RuntimeError("qiskit-ibm-runtime is not installed.") from exc
    if installed != EXPECTED_RUNTIME_VERSION:
        raise RuntimeError(
            f"This script requires qiskit-ibm-runtime=={EXPECTED_RUNTIME_VERSION}; "
            f"found {installed}."
        )

    # The Sampler's built-in DD option remains disabled for MCM. This sweep
    # uses Runtime 0.43's dynamic-circuit scheduling passes before submission.

from qiskit_ibm_runtime.circuit import MidCircuitMeasure
from qiskit.circuit import Measure
from qiskit import transpile as _qiskit_transpile
from qiskit_experiments.framework import BaseExperiment


def _append_measure_2(circuit, qubits, clbits):
    """Broadcast IBM readout to scalar bits, registers, or bit lists."""
    qubits = circuit._qbit_argument_conversion(qubits)
    clbits = circuit._cbit_argument_conversion(clbits)
    if len(qubits) != len(clbits):
        raise ValueError('Readout requires one classical bit per measured qubit.')
    for qubit, clbit in zip(qubits, clbits):
        circuit.append(MidCircuitMeasure(), [qubit], [clbit])
    return circuit


def _readout_is_simulator(backend):
    if backend is None:
        return False
    try:
        return bool(backend.configuration().simulator)
    except (AttributeError, TypeError):
        return 'aer' in type(backend).__module__.lower()


def _rewrite_readouts(circuit, simulator=False):
    """Preserve registers/layout/metadata and recursively use one readout kind."""
    result = circuit.copy_empty_like()
    for item in circuit.data:
        operation = item.operation
        if operation.name in ('measure', 'measure_2'):
            operation = Measure() if simulator else MidCircuitMeasure()
        elif getattr(operation, 'blocks', ()):
            operation = operation.replace_blocks([
                _rewrite_readouts(block, simulator) for block in operation.blocks
            ])
        result.append(operation, item.qubits, item.clbits)
    result._layout = circuit.layout
    return result


def _validate_physical_readouts(circuit, backend, root_indices=None):
    if backend is None or _readout_is_simulator(backend):
        return
    root_indices = list(range(circuit.num_qubits)) if root_indices is None else root_indices
    for item in circuit.data:
        physical = [root_indices[circuit.find_bit(q).index] for q in item.qubits]
        if item.operation.name in ('measure', 'measure_2'):
            if item.operation.name != 'measure_2':
                raise RuntimeError('Hardware circuit contains standard measure.')
            if not backend.target.instruction_supported('measure_2', tuple(physical)):
                raise RuntimeError(f'measure_2 is unsupported on physical qubits {physical}.')
        for block in getattr(item.operation, 'blocks', ()):
            _validate_physical_readouts(block, backend, physical)


def transpile(circuits, backend=None, *args, **kwargs):
    """Normalize before routing/scheduling; validate every measured physical qubit."""
    single = isinstance(circuits, QuantumCircuit)
    prepared = [_rewrite_readouts(c, _readout_is_simulator(backend))
                for c in ([circuits] if single else circuits)]
    result = _qiskit_transpile(prepared[0] if single else prepared, backend, *args, **kwargs)
    for circuit in ([result] if single else result):
        _validate_physical_readouts(circuit, backend)
    return result


# Qiskit Experiments generates its own terminal measurements (including
# LocalReadoutError/MST and RB). Normalize its routed circuits as well.
if not hasattr(BaseExperiment, '_original_readout_transpiled_circuits'):
    BaseExperiment._original_readout_transpiled_circuits = BaseExperiment._transpiled_circuits


def _experiment_measure_2_circuits(self):
    options = dict(self.transpile_options.__dict__)
    options['initial_layout'] = list(self.physical_qubits)
    return transpile(self.circuits(), self.backend, **options)


BaseExperiment._transpiled_circuits = _experiment_measure_2_circuits

if not hasattr(BaseExperiment, '_original_readout_run_jobs'):
    BaseExperiment._original_readout_run_jobs = BaseExperiment._run_jobs


def _experiment_measure_2_jobs(self, circuits, sampler=None, **run_options):
    # Covers experiments such as optimized RB with their own transpiler.
    circuits = [_rewrite_readouts(c, _readout_is_simulator(self.backend)) for c in circuits]
    for circuit in circuits:
        _validate_physical_readouts(circuit, self.backend)
    return self._original_readout_run_jobs(circuits, sampler=sampler, **run_options)


BaseExperiment._run_jobs = _experiment_measure_2_jobs


def _calibrate_all_measure_2(mitigator, qubits, shots=5000):
    """Local assignment matrices measured with the same hardware instruction."""
    backend = mitigator.system
    physical = list(dict.fromkeys(int(q) for q in qubits))
    circuits = []
    for prepared in (0, 1):
        circuit = QuantumCircuit(len(physical), len(physical))
        if prepared:
            circuit.x(range(len(physical)))
        _append_measure_2(circuit, range(len(physical)), range(len(physical)))
        circuits.append(circuit)
    compiled = transpile(circuits, backend, initial_layout=physical, optimization_level=0)
    sampler = (Sampler(mode=backend) if _readout_is_simulator(backend)
               else make_hardware_session_sampler(backend))
    result = sampler.run(compiled, shots=int(shots)).result()
    matrices = [None] * backend.num_qubits
    for logical, qubit in enumerate(physical):
        matrix = np.zeros((2, 2))
        for prepared, pub in enumerate(result):
            counts = pub.data.c.get_counts()
            for bits, count in counts.items():
                matrix[int(bits.replace(' ', '')[::-1][logical]), prepared] += count
        matrix /= matrix.sum(axis=0)
        matrices[qubit] = matrix
    mitigator.cals_from_matrices(matrices)
    return mitigator
```

```python
# Rank-1 decoder and its matching AME resource roles.
# The legacy V234 function names remain as compatibility entry points only.
RANK_ACTIVE_QUBITS = (0, 1, 4)
RANK_CHANNEL_RESOURCE_ROLES = (("A", 2, 0), ("B", 3, 1))
RANK_RECEIVERS = {"A": 0, "B": 1}

def _apply_ame52_prep(qc, q):
    """Append the AME(5,2) preparation used by v13."""
    qc.h(q[2]); qc.h(q[3])
    qc.cx(q[2], q[1])
    qc.cx(q[1], q[0])
    qc.h(q[2])
    qc.cx(q[2], q[4])
    qc.cx(q[3], q[2])
    qc.cx(q[2], q[0])
    return qc


def build_V234_dagger_current():
    """Exact synthesized rank-1 three-qubit decoder (local active-qubit order)."""
    q = QuantumRegister(3, "q")
    qc = QuantumCircuit(q, name="rank1_decoder")
    qc.h(q[0]); qc.s(q[0]); qc.h(q[1])
    qc.cx(q[0], q[1]); qc.sdg(q[0]); qc.h(q[0])
    qc.cx(q[1], q[2]); qc.cx(q[0], q[1])
    return qc


def apply_V234_dagger(qc, *ame_qubits):
    """Apply the selected decoder to its rank-specific AME qubits."""
    if len(ame_qubits) == 1 and hasattr(ame_qubits[0], "__len__"):
        ame_qubits = tuple(ame_qubits[0])
    if len(ame_qubits) == 5:
        targets = [ame_qubits[index] for index in RANK_ACTIVE_QUBITS]
    elif len(ame_qubits) == 3:
        targets = list(ame_qubits)
    else:
        raise ValueError("Pass either the five AME qubits or the three active decoder qubits.")
    qc.append(build_V234_dagger_current().to_instruction(), targets)
    return qc


def build_source_epr_pair_circuit():
    q = QuantumRegister(2, "pair")
    qc = QuantumCircuit(q, name="source_epr_pair")
    qc.h(q[0]); qc.cx(q[0], q[1])
    return qc


def build_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=True):
    """One complete rank-specific parallel teleportation round."""
    return build_cascade_epr_half_teleportation_circuit(1, with_corrections, use_measure_2)


def build_two_round_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=True):
    """Two complete rank-specific parallel teleportation rounds."""
    return build_cascade_epr_half_teleportation_circuit(2, with_corrections, use_measure_2)


# Bell target and the fixed CHSH operator used in Equation (9).
PHI_PLUS = np.array([1, 0, 0, 1], dtype=complex) / np.sqrt(2)
X = np.array([[0, 1], [1, 0]], dtype=complex)
Z = np.array([[1, 0], [0, -1]], dtype=complex)
CHSH_OPERATOR = np.sqrt(2) * (np.kron(X, X) + np.kron(Z, Z))


def metrics_from_density_matrix(rho):
    """Return Bell fidelity and CHSH S from exactly the same rho."""
    rho = np.asarray(rho, dtype=complex)
    rho = 0.5 * (rho + rho.conj().T)
    rho /= np.trace(rho)
    fidelity = float(np.real(np.vdot(PHI_PLUS, rho @ PHI_PLUS)))
    chsh_from_rho = float(np.real(np.trace(CHSH_OPERATOR @ rho)))
    lower = chsh_from_rho / (2 * np.sqrt(2))
    upper = chsh_from_rho / (4 * np.sqrt(2)) + 0.5
    if not (lower - 1e-8 <= fidelity <= upper + 1e-8):
        raise AssertionError(
            f"Internal same-rho Equation (9) failure: S={chsh_from_rho}, "
            f"F={fidelity}, bounds=({lower}, {upper})."
        )
    return fidelity, chsh_from_rho


print("Circuit and same-rho metric definitions loaded.")

CASCADE_SCHEME = 'MCM'

def build_cascade_epr_half_teleportation_circuit(rounds=3, with_corrections=True, use_measure_2=True, round_boundaries=False):
    """Rank-specific prefixes of the three-round A/B cascade; references stay unmeasured."""
    if rounds not in (1, 2, 3):
        raise ValueError('rounds must be 1, 2, or 3')
    refs = [QuantumRegister(1, f'ref_{channel}') for channel in ('A', 'B')]
    inputs = [QuantumRegister(1, f'in_{channel}') for channel in ('A', 'B')]
    resources = [QuantumRegister(5, f'r{r}') for r in range(1, rounds + 1)]
    syndromes = [(ClassicalRegister(2, f'r{r}_mA'), ClassicalRegister(2, f'r{r}_mB'))
                 for r in range(1, rounds + 1)]
    qc = QuantumCircuit(*refs, *inputs, *resources,
                        *(reg for pair in syndromes for reg in pair),
                        name=f'{rounds}_round_rank{DISENTANGLER_RANK}_{CASCADE_SCHEME.lower()}')
    for ref, inp in zip(refs, inputs):
        qc.h(ref[0]); qc.cx(ref[0], inp[0])
    current = [reg[0] for reg in inputs]
    def mark_round(round_number):
        if not round_boundaries:
            return
        # Guard keeps all role markers at the completed round, even under ALAP.
        qc.barrier(*qc.qubits, label=f'CUMULATIVE_GUARD_{round_number}')
        for channel_index, channel in enumerate(('A', 'B')):
            qc.barrier(refs[channel_index][0],
                       label=f'CUMULATIVE_ROLE_{round_number}_{channel}_ref')
            qc.barrier(current[channel_index],
                       label=f'CUMULATIVE_ROLE_{round_number}_{channel}_receiver')
        qc.barrier(*qc.qubits, label=f'CUMULATIVE_ROUND_{round_number}')

    mark_round(0)
    deferred = []
    for round_number, (resource, (bits_a, bits_b)) in enumerate(zip(resources, syndromes), 1):
        _apply_ame52_prep(qc, resource)
        apply_V234_dagger(qc, resource)
        for channel_index, (bits, (_, bell_index, receiver_index)) in enumerate(
                zip((bits_a, bits_b), RANK_CHANNEL_RESOURCE_ROLES)):
            inp, bell, receiver = current[channel_index], resource[bell_index], resource[receiver_index]
            qc.cx(inp, bell); qc.h(inp)
            if CASCADE_SCHEME == 'MCM':
                _append_measure_2(qc, inp, bits[0]); _append_measure_2(qc, bell, bits[1])
                if with_corrections:
                    with qc.if_test((bits[1], 1)): qc.x(receiver)
                    with qc.if_test((bits[0], 1)): qc.z(receiver)
            else:
                if with_corrections:
                    qc.cx(bell, receiver); qc.cz(inp, receiver)
                deferred.extend(((inp, bits[0]), (bell, bits[1])))
            current[channel_index] = receiver
        mark_round(round_number)
    for qubit, clbit in deferred:
        _append_measure_2(qc, qubit, clbit)
    qc.metadata = {'teleportation_rounds': rounds, 'disentangler_rank': DISENTANGLER_RANK,
                   'rank_active_ame_qubits': list(RANK_ACTIVE_QUBITS),
                   'rank_channel_roles': [list(x) for x in RANK_CHANNEL_RESOURCE_ROLES],
                   'round_1_output_is_round_2_input': rounds >= 2,
                   'round_2_output_is_round_3_input': rounds >= 3}
    return qc


def build_three_round_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=True):
    return build_cascade_epr_half_teleportation_circuit(3, with_corrections, use_measure_2)


def cascade_channel_specs(full_layout):
    if len(full_layout) != 19:
        raise ValueError('Three-round cascade requires one shared 19-qubit layout.')
    specs = {'A': {}, 'B': {}}
    syndromes = []
    current = [2, 3]
    for r in (1, 2, 3):
        start = 4 + 5 * (r - 1)
        source_a = start + RANK_CHANNEL_RESOURCE_ROLES[0][1]
        source_b = start + RANK_CHANNEL_RESOURCE_ROLES[1][1]
        syndromes.extend((current[0], source_a, current[1], source_b))
        current = [start + RANK_RECEIVERS['A'], start + RANK_RECEIVERS['B']]
        for channel, ref, receiver in (('A', 0, current[0]), ('B', 1, current[1])):
            specs[channel][r] = {'layout': list(full_layout[:4 + 5 * r]),
                                  'indices': [ref, receiver],
                                  'm3_qubits': [full_layout[i] for i in syndromes + [ref, receiver]]}
    return specs


TRAJECTORY_STATE_PAIRS = {
    channel: {r: (f'ref_{channel}', f'in_{channel}' if r == 0 else f'q{receiver}_R{r}',
                  (ref, source if r == 0 else 4 + 5 * (r - 1) + receiver),
                  f'rho(ref_{channel}, in_{channel})' if r == 0
                  else f'rho(ref_{channel}, q{receiver}_R{r})')
              for r in (0, 1, 2, 3)}
    for channel, ref, source, receiver in (('A', 0, 2, 0), ('B', 1, 3, 1))
}


def _state_role(round_number):
    return ('measured_input_state', 'round_1_output_and_round_2_input',
            'round_2_output_and_round_3_input', 'round_3_output')[int(round_number)]


def save_heatmap_outputs(records, scheme):
    """Export each hardware DD policy independently; simulation stays CSV-only."""
    frame = records if isinstance(records, pd.DataFrame) else pd.DataFrame(records)
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    prefix, outputs = f'{scheme.lower()}_rank{DISENTANGLER_RANK}', {}
    for policy in ('none', 'XY4', 'XX'):
        policies = {channel: {r: policy for r in range(4)} for channel in ('A', 'B')}
        png = OUTPUT_DIR / f'{prefix}_HW_R0123_{policy}_heatmaps.png'
        csv = png.with_suffix('.csv')
        plot_state_fidelity_trajectory_heatmaps(frame, scheme, policies, png, csv)
        outputs[policy] = {'png': png, 'csv': csv}
    if (frame['domain'] == 'Noisy simulation').any():
        policies = {channel: {r: 'reference' for r in range(3)} for channel in ('A', 'B')}
        table, _ = _state_fidelity_export_rows(frame, ('sim_raw', 'sim_m3'), policies)
        csv = OUTPUT_DIR / f'{prefix}_SIM_R012_state_fidelity.csv'
        table.to_csv(csv, index=False)
        outputs['simulation_state_fidelity_csv'] = csv
    return outputs


def plot_state_fidelity_trajectory_heatmaps(records, scheme, policy_by_round_by_channel, save_path, csv_path):
    """Plot only real hardware: raw and M3, both with input plus rounds 1-3."""
    conditions = ('hw_raw', 'hw_m3')
    table, panels = _state_fidelity_export_rows(records, conditions, policy_by_round_by_channel)
    fig, axes = plt.subplots(1, 2, figsize=(20, 9), layout='constrained')
    for axis, condition in zip(np.asarray(axes).flat, conditions):
        labels, means = panels[condition]
        image = axis.imshow(means, vmin=0, vmax=1, cmap='viridis')
        for i in range(len(labels)):
            for j in range(len(labels)):
                value = means[i, j]
                if np.isfinite(value):
                    axis.text(j, i, f'{value:.3f}', ha='center', va='center', fontsize=8,
                              color='white' if value < .72 else 'black')
        axis.set_xticks(range(len(labels)), labels, rotation=45, ha='right', fontsize=9)
        axis.set_yticks(range(len(labels)), labels, fontsize=9)
        axis.set_title(HEATMAP_CONDITION_LABELS[condition])
    policy_text = '; '.join(f"{channel}: " + ', '.join(
        f"round {r}={policy_by_round_by_channel[channel][r]}" for r in (0, 1, 2, 3)) for channel in ('A', 'B'))
    fig.suptitle(f'{scheme} hardware state fidelity: input and rounds 1-3\n{policy_text}\nCross-channel comparisons omitted', fontsize=13)
    fig.colorbar(image, ax=np.asarray(axes).ravel().tolist(), shrink=.65, label='Mean state fidelity')
    fig.savefig(save_path, dpi=180, bbox_inches='tight')
    plt.show(); plt.close(fig)
    table.to_csv(csv_path, index=False)
```

```python
def _save_m3_calibration_audit(payload, output_dir):
    """Write a uniquely named audit, including in long Windows directories."""
    import os
    from uuid import uuid4
    output_dir = Path(os.path.abspath(output_dir))
    audit_path = output_dir / f'm3_{uuid4().hex}.json'
    io_path = audit_path
    if os.name == 'nt':
        # Extended paths avoid dependence on the Windows LongPathsEnabled
        # setting. UNC paths require the extended UNC prefix.
        slash = chr(92)
        prefix = slash * 2 + '?' + slash
        text = str(audit_path)
        if not text.startswith(prefix):
            text = (prefix + 'UNC' + slash + text[2:]
                    if text.startswith(slash * 2) else prefix + text)
        io_path = Path(text)
    io_path.parent.mkdir(parents=True, exist_ok=True)
    with io_path.open('x', encoding='utf-8') as stream:
        json.dump(payload, stream, indent=2, allow_nan=False)
    return audit_path


class _LayoutUncalibrated(Exception):
    pass


def _symmetric_pair_gate_error_cost(circuit, layouts, backend):
    """Mapomatic cost robust to symmetric CZ ordering/missing couplers."""
    from qiskit_ibm_runtime.models.backend_properties import BackendPropertyError

    def two_qubit_error(properties, gate_name, p0, p1):
        try:
            return properties.gate_error(gate_name, [p0, p1])
        except BackendPropertyError:
            try:
                return properties.gate_error(gate_name, [p1, p0])
            except BackendPropertyError as exc:
                raise _LayoutUncalibrated(
                    f"No {gate_name} calibration for ({p0},{p1}) "
                    "in either ordering."
                ) from exc

    properties = backend.properties()
    scored = []
    skipped = 0
    for layout in layouts:
        fidelity_product = 1.0
        try:
            for item in circuit.data:
                operation, qargs = item.operation, item.qubits
                if operation.num_qubits == 2 and operation.name != "barrier":
                    q0 = circuit.find_bit(qargs[0]).index
                    q1 = circuit.find_bit(qargs[1]).index
                    fidelity_product *= 1.0 - two_qubit_error(
                        properties, operation.name, layout[q0], layout[q1]
                    )
                elif operation.name in ("sx", "x"):
                    q0 = circuit.find_bit(qargs[0]).index
                    fidelity_product *= 1.0 - properties.gate_error(
                        operation.name, layout[q0]
                    )
                elif operation.name in ("measure", "reset"):
                    q0 = circuit.find_bit(qargs[0]).index
                    fidelity_product *= 1.0 - properties.readout_error(
                        layout[q0]
                    )
        except _LayoutUncalibrated:
            skipped += 1
            continue
        scored.append((list(layout), 1.0 - fidelity_product))
    return scored, skipped


def select_physical_layout(circuit, backend, seed=SEED_TRANSPILER):
    """Select the full logical layout using v13's Mapomatic workflow.

    Measurements and dynamic single-qubit correction blocks are removed only
    from the temporary layout proxy because they add no coupling-map edges.
    The actual executed circuit remains unchanged. Candidate embeddings are
    ranked using the backend's current gate/readout calibration snapshot.
    """
    layout_proxy = QuantumCircuit(
        *circuit.qregs, name=f"{circuit.name}_mapomatic_proxy"
    )
    skipped_control = skipped_measurements = 0
    for item in circuit.data:
        operation, qargs, cargs = item.operation, item.qubits, item.clbits
        if isinstance(operation, ControlFlowOp) or operation.name in {
            "if_else", "while_loop", "for_loop", "switch_case"
        }:
            skipped_control += 1
            continue
        if operation.name == "measure":
            skipped_measurements += 1
            continue
        if cargs:
            continue
        layout_proxy.append(operation, qargs, [])

    routed = transpile(
        layout_proxy,
        backend=backend,
        optimization_level=OPTIMIZATION_LEVEL,
        seed_transpiler=int(seed),
    )
    compact = mm.deflate_circuit(routed)
    candidates = mm.matching_layouts(compact, backend)
    if not candidates:
        raise RuntimeError(
            f"Mapomatic found no {circuit.num_qubits}-qubit placement on "
            f"{backend.name}."
        )

    scored, calibration_skips = _symmetric_pair_gate_error_cost(
        compact, candidates, backend
    )
    scored.sort(key=lambda item: item[1])
    if not scored:
        raise RuntimeError(
            "Mapomatic found topological candidates, but none had complete "
            "calibration data. Retry after the next backend calibration."
        )
    best_layout, best_score = scored[0]

    # Routing can temporarily activate bridge qubits. Mapomatic ranks the
    # complete routed region, then this block retains only the original nine
    # logical qubits required by Qiskit's initial_layout argument.
    if len(best_layout) != circuit.num_qubits:
        active_qubits, _ = mm.circuits.active_bits(routed)
        active_sorted = sorted(
            active_qubits, key=lambda q: routed.find_bit(q).index
        )
        physical_to_compact = {
            routed.find_bit(q).index: index
            for index, q in enumerate(active_sorted)
        }
        original_physical = routed.layout.initial_index_layout(
            filter_ancillas=True
        )
        compact_indices = [
            physical_to_compact[physical] for physical in original_physical
        ]
        best_layout = [best_layout[index] for index in compact_indices]

    if len(best_layout) != circuit.num_qubits:
        raise RuntimeError(
            f"Mapomatic returned {len(best_layout)} entries for a "
            f"{circuit.num_qubits}-qubit circuit."
        )
    if len(set(best_layout)) != len(best_layout):
        raise RuntimeError(f"Mapomatic returned duplicate qubits: {best_layout}")

    print(
        f"Mapomatic selected {best_layout} on {backend.name}; "
        f"calibration score={best_score:.6f}; "
        f"candidates={len(scored)}, skipped={calibration_skips}; "
        f"proxy omitted {skipped_control} control-flow blocks and "
        f"{skipped_measurements} measurements."
    )
    return list(best_layout)


def select_balanced_physical_layout(circuit, backend, seed=SEED_TRANSPILER):
    """Pick one layout that is readout-safe for both channels and rounds.

    Repicking uses different transpiler seeds and is completed before any
    M3 calibration or experiment job is submitted.  A cascaded state cannot
    change layout between rounds, so all A/B R0/R1/R2 pairs are constrained
    simultaneously.
    """
    layout_proxy = QuantumCircuit(
        *circuit.qregs, name=f"{circuit.name}_balanced_mapomatic_proxy"
    )
    skipped_control = skipped_measurements = 0
    for item in circuit.data:
        operation, qargs, cargs = item.operation, item.qubits, item.clbits
        if isinstance(operation, ControlFlowOp) or operation.name in {
            "if_else", "while_loop", "for_loop", "switch_case"
        }:
            skipped_control += 1
            continue
        if operation.name == "measure":
            skipped_measurements += 1
            continue
        if cargs:
            continue
        layout_proxy.append(operation, qargs, [])

    # Rank-aware source/receiver pairs for one-, two- and three-round prefixes.
    n_rounds = min(3, (circuit.num_qubits - 4) // 5)
    pair_specs = {
        f'{channel}_R{r}': (indices[0], indices[1], 'source' if r == 0 else 'analysis')
        for channel, pairs in TRAJECTORY_STATE_PAIRS.items()
        for r, (_, _, indices, _) in pairs.items()
        if n_rounds >= 1 and r <= n_rounds
    }

    properties = backend.properties()
    excluded_readout = set()
    if properties is not None:
        excluded_readout.update(properties.faulty_qubits())
    print(f'Excluded readout qubits on {backend.name}: {sorted(excluded_readout)}')
    if properties is None:
        raise RuntimeError("Backend properties are unavailable for balanced layout selection.")

    accepted = {}
    best_rejected = None
    for attempt in range(int(LAYOUT_REPICK_ATTEMPTS)):
        attempt_seed = int(seed) + attempt
        routed = transpile(
            layout_proxy, backend=backend,
            optimization_level=OPTIMIZATION_LEVEL,
            seed_transpiler=attempt_seed,
        )
        compact = mm.deflate_circuit(routed)
        candidates = mm.matching_layouts(compact, backend)
        if not candidates:
            print(f"Balanced layout attempt {attempt + 1}/{LAYOUT_REPICK_ATTEMPTS}: no embeddings.")
            continue
        scored, calibration_skips = _symmetric_pair_gate_error_cost(
            compact, candidates, backend
        )
        active_qubits, _ = mm.circuits.active_bits(routed)
        active_sorted = sorted(
            active_qubits, key=lambda q: routed.find_bit(q).index
        )
        routed_to_compact = {
            routed.find_bit(q).index: index
            for index, q in enumerate(active_sorted)
        }
        original_routed = list(
            routed.layout.initial_index_layout(filter_ancillas=True)
        )
        if len(original_routed) != circuit.num_qubits:
            raise RuntimeError(
                "Could not map Mapomatic candidates back to every original logical qubit."
            )

        passed_this_attempt = 0
        for candidate, gate_score in scored:
            try:
                logical_layout = [
                    int(candidate[routed_to_compact[int(routed_index)]])
                    for routed_index in original_routed
                ]
                if excluded_readout.intersection(logical_layout):
                    continue
                readout = {}
                for physical in set(logical_layout):
                    instruction = backend.target['measure_2'].get((physical,))
                    error = getattr(instruction, 'error', None)
                    # Ordinary readout calibration is only a ranking proxy
                    # when IBM exposes no measure_2-specific error estimate.
                    if error is None:
                        error = properties.readout_error(physical)
                    readout[physical] = float(error)
            except Exception:
                continue
            if not all(
                    np.isfinite(value) and 0.0 <= value < 1.0
                    for value in readout.values()):
                continue
            if len(set(logical_layout)) != len(logical_layout):
                continue

            measured_indices = range(circuit.num_qubits)
            if any(not backend.target.instruction_supported('measure_2', (logical_layout[index],))
                   for index in measured_indices):
                continue
            pair_risks = {}
            for label, (left, right, _) in pair_specs.items():
                e_left = readout[logical_layout[left]]
                e_right = readout[logical_layout[right]]
                pair_risks[label] = 1.0 - (1.0 - e_left) * (1.0 - e_right)
            source_risks = [
                risk for label, risk in pair_risks.items() if label.endswith("R0")
            ] or [0.0]
            analysis_risks = [
                risk for label, risk in pair_risks.items() if not label.endswith("R0")
            ] or [0.0]
            a_risk = max(
                [risk for label, risk in pair_risks.items() if label.startswith("A_")] or [0.0]
            )
            b_risk = max(
                [risk for label, risk in pair_risks.items() if label.startswith("B_")] or [0.0]
            )
            worst_source = max(source_risks)
            worst_analysis = max(analysis_risks)
            channel_gap = abs(a_risk - b_risk)
            risks = list(pair_risks.values()) or [0.0]
            objective = (
                float(gate_score)
                + LAYOUT_MEAN_READOUT_WEIGHT * float(np.mean(risks))
                + LAYOUT_WORST_READOUT_WEIGHT * max(risks)
                + LAYOUT_CHANNEL_BALANCE_WEIGHT * channel_gap
            )
            passes = (
                worst_source <= LAYOUT_MAX_SOURCE_PAIR_READOUT_RISK
                and worst_analysis <= LAYOUT_MAX_ANALYSIS_PAIR_READOUT_RISK
                and channel_gap <= LAYOUT_MAX_CHANNEL_READOUT_RISK_GAP
            )
            candidate_info = {
                "layout": logical_layout, "gate_score": float(gate_score),
                "objective": float(objective), "pair_risks": pair_risks,
                "worst_source": worst_source,
                "worst_analysis": worst_analysis,
                "channel_gap": channel_gap, "seed": attempt_seed,
            }
            violation = max(
                worst_source / LAYOUT_MAX_SOURCE_PAIR_READOUT_RISK,
                worst_analysis / LAYOUT_MAX_ANALYSIS_PAIR_READOUT_RISK,
                channel_gap / LAYOUT_MAX_CHANNEL_READOUT_RISK_GAP,
            )
            if best_rejected is None or (violation, objective) < best_rejected[0]:
                best_rejected = ((violation, objective), candidate_info)
            if passes:
                passed_this_attempt += 1
                key = tuple(logical_layout)
                if key not in accepted or objective < accepted[key]["objective"]:
                    accepted[key] = candidate_info
        print(
            f"Balanced layout attempt {attempt + 1}/{LAYOUT_REPICK_ATTEMPTS}: "
            f"seed={attempt_seed}, candidates={len(scored)}, "
            f"passed={passed_this_attempt}, calibration-skipped={calibration_skips}."
        )

    if not accepted:
        detail = "no candidate had complete readout calibration"
        if best_rejected is not None:
            info = best_rejected[1]
            detail = (
                f"best rejected layout={info['layout']}, "
                f"source risk={info['worst_source']:.3f}, "
                f"analysis risk={info['worst_analysis']:.3f}, "
                f"A/B gap={info['channel_gap']:.3f}, "
                f"pair risks={info['pair_risks']}"
            )
        raise RuntimeError(
            f"No balanced Mapomatic layout passed after {LAYOUT_REPICK_ATTEMPTS} "
            f"automatic repicks; {detail}. Limits: source<="
            f"{LAYOUT_MAX_SOURCE_PAIR_READOUT_RISK:.3f}, analysis<="
            f"{LAYOUT_MAX_ANALYSIS_PAIR_READOUT_RISK:.3f}, A/B gap<="
            f"{LAYOUT_MAX_CHANNEL_READOUT_RISK_GAP:.3f}. No calibration or "
            "experiment job has been submitted."
        )

    best = min(accepted.values(), key=lambda item: item["objective"])
    risk_text = ", ".join(
        f"{label}={risk:.3f}" for label, risk in best["pair_risks"].items()
    )
    print(
        f"Balanced Mapomatic selected {best['layout']} on {backend.name}; "
        f"seed={best['seed']}, objective={best['objective']:.6f}, "
        f"gate score={best['gate_score']:.6f}; {risk_text}; "
        f"proxy omitted {skipped_control} control-flow blocks and "
        f"{skipped_measurements} measurements."
    )
    return list(best["layout"])


def _job_identifier(job):
    try:
        return str(job.job_id())
    except Exception:
        return "local/no-runtime-id"


def _job_status(job):
    try:
        status = job.status()
        return getattr(status, "name", str(status))
    except Exception:
        return "status unavailable"


_TIMING_PLOTS_CREATED = 0
_SCHEDULER_TIMING_SIGNATURES = set()

def _scheduler_timing_signature(timing):
    """Canonicalize a schedule for structural duplicate detection.

    Line order, whitespace, and a uniform shift of every start time are
    ignored. Operations, channels, relative timing, and durations remain
    significant, so genuinely different schedules still get plotted.
    """
    if isinstance(timing, str):
        lines = timing.splitlines()
    else:
        lines = [str(item) for item in timing]

    rows = []
    for line in lines:
        fields = tuple(field.strip() for field in line.split(","))
        if not any(fields):
            continue
        start = None
        if len(fields) > 3:
            try:
                start = float(fields[3])
            except ValueError:
                pass
        rows.append((fields, start))

    starts = [start for _, start in rows if start is not None]
    origin = min(starts) if starts else 0.0
    canonical = []
    for fields, start in rows:
        normalized = list(fields)
        if start is not None:
            normalized[3] = f"{start - origin:.12g}"
        canonical.append(tuple(normalized))
    return tuple(sorted(canonical))



def capture_scheduler_timing(result, label):
    """Display the first returned hardware schedule inside the notebook.

    IBM supplies scheduler timing only after a hardware PUB has completed.
    Status messages remain the live progress indicator while the job runs.
    """
    global _TIMING_PLOTS_CREATED
    if not SHOW_SCHEDULER_TIMING:
        return

    for pub_index, pub_result in enumerate(result):
        metadata = getattr(pub_result, "metadata", {}) or {}
        timing = (
            metadata.get("compilation", {})
            .get("scheduler_timing", {})
            .get("timing")
        )
        if not timing:
            continue
        signature = _scheduler_timing_signature(timing)
        if signature in _SCHEDULER_TIMING_SIGNATURES:
            continue
        if _TIMING_PLOTS_CREATED >= MAX_TIMING_PLOTS:
            return

        _SCHEDULER_TIMING_SIGNATURES.add(signature)
        _TIMING_PLOTS_CREATED += 1
        try:
            from qiskit_ibm_runtime.visualization import (
                draw_circuit_schedule_timing,
            )
            fig = draw_circuit_schedule_timing(
                circuit_schedule=timing,
                included_channels=None,
                filter_readout_channels=False,
                filter_barriers=False,
                width=1200,
            )
            fig.update_layout(title=f"Scheduler timing: {label}, PUB {pub_index}")
            try:
                fig.show(renderer="notebook")
            except ValueError as exc:
                # Plotly's notebook renderer requires nbformat.  Keep the
                # hardware workflow alive and display an HTML fallback.
                from IPython.display import HTML, display
                display(HTML(fig.to_html(
                    full_html=False, include_plotlyjs="cdn"
                )))
                print(
                    f"Used HTML timing fallback ({exc}). "
                    "Install nbformat>=4.2.0 for Plotly's notebook renderer.",
                    flush=True,
                )
        except (ImportError, ModuleNotFoundError) as exc:
            print(
                f"Scheduler timing visualization is unavailable: {exc}",
                flush=True,
            )
        return


def submit_sampler_job(sampler, circuits, shots, label):
    """Submit visibly, applying terminal-readout restrictions only to hardware."""
    backend = _sampler_runtime_backend(sampler)
    # Aer represents mid-circuit readout with ordinary measure instructions.
    # Gates after those measurements are valid in a dynamic simulation.
    if not _readout_is_simulator(backend):
        assert_terminal_measurements_are_final(circuits, label)
    print(
        f"[{label}] submitting {len(circuits)} circuit(s), "
        f"{int(shots)} shots each...",
        flush=True,
    )
    job = sampler.run(circuits, shots=int(shots))
    print(
        f"[{label}] accepted: job_id={_job_identifier(job)}",
        flush=True,
    )
    return job


def wait_for_sampler_job(job, label):
    """Poll quietly and print only when the job status changes."""
    terminal = {
        "DONE", "COMPLETED", "ERROR", "CANCELLED", "CANCELED",
        "STATUS UNAVAILABLE",
    }
    last_status = None
    while True:
        status = _job_status(job)
        if status != last_status:
            print(f"[{label}] status={status}", flush=True)
            last_status = status
        if status.upper() in terminal:
            break
        time.sleep(STATUS_POLL_SECONDS)
    result = job.result()
    print(f"[{label}] result received", flush=True)
    capture_scheduler_timing(result, label)
    return result



def _job_error_code_and_message(job, exc):
    """Return IBM's numeric error code and a compact failure message."""
    try:
        message = job.error_message() or str(exc)
    except Exception:
        message = str(exc)
    message = str(message).strip()
    match = re.search(r"Error code\s+(\d+)", message, flags=re.IGNORECASE)
    return (int(match.group(1)) if match else None), message



def _is_retryable_ibm_failure(code, message):
    """Retry bounded IBM availability, queue, service, and state failures."""
    text = str(message).lower()
    transient_markers = (
        "internal compiler error",
        "temporary internal error",
        "internal server error",
        "error queueing job",
        "queue is unavailable",
        "queue is disabled",
        "failed to run job",
        "out of sync",
        "unexpected state",
        "internal error syncing job",
        "job transition not valid",
        "job status not valid",
        "temporarily unavailable",
        "service unavailable",
        "error publishing job",
        "error creating job",
    )
    return (
        code in RETRYABLE_IBM_ERROR_CODES
        or any(marker in text for marker in transient_markers)
    )


def run_sampler_job_with_retries(sampler, circuits, shots, label):
    """Retry every caught submission/result error up to the attempt limit."""
    last_error = None
    retry_backend = _sampler_runtime_backend(sampler)
    for attempt in range(1, JOB_MAX_ATTEMPTS + 1):
        attempt_label = (
            label if JOB_MAX_ATTEMPTS == 1
            else f"{label}, attempt {attempt}/{JOB_MAX_ATTEMPTS}"
        )
        job = None
        try:
            # Keep preflight inside the retry boundary: status-query errors
            # must not bypass the all-errors retry policy.
            if attempt > 1 and retry_backend is not None:
                validate_runtime_backend(
                    retry_backend, wait_for_active=True
                )
            # Submission belongs inside the try block: queueing can fail before
            # Runtime returns a usable job object.
            job = submit_sampler_job(
                sampler, circuits, shots, attempt_label
            )
            return wait_for_sampler_job(job, attempt_label)
        except Exception as exc:
            last_error = exc
            code, message = _job_error_code_and_message(job, exc)
            if attempt >= JOB_MAX_ATTEMPTS:
                print(
                    f"[{label}] failed: job_id={_job_identifier(job)}, "
                    f"error_code={code}, retries_exhausted=True",
                    flush=True,
                )
                raise
            delay = JOB_RETRY_BASE_SECONDS * attempt
            print(
                f"[{label}] caught error {code}; retry_policy=all_errors; "
                f"job_id={_job_identifier(job)}. Retrying in {delay}s. "
                f"Server message: {message}",
                flush=True,
            )
            time.sleep(delay)
    raise last_error


def run_experiment_data_with_retries(
        submit_experiment, backend, label):
    """Retry a complete qiskit-experiments submission as a fresh workload."""
    last_error = None
    for attempt in range(1, JOB_MAX_ATTEMPTS + 1):
        attempt_label = f"{label}, attempt {attempt}/{JOB_MAX_ATTEMPTS}"
        active_job = None
        try:
            # Keep preflight inside the retry boundary so every ordinary
            # exception remains eligible for the next bounded attempt.
            if attempt > 1:
                validate_runtime_backend(backend, wait_for_active=True)
            print(
                f"[{attempt_label}] preparing and submitting experiment jobs...",
                flush=True,
            )
            experiment_data = submit_experiment()
            submitted_jobs = list(experiment_data.jobs())
            if submitted_jobs:
                print(
                    f"[{attempt_label}] accepted job(s): "
                    + ", ".join(
                        f"{_job_identifier(item)} ({_job_status(item)})"
                        for item in submitted_jobs
                    ),
                    flush=True,
                )
            for job_index, active_job in enumerate(submitted_jobs):
                wait_for_sampler_job(
                    active_job,
                    f"{attempt_label}, job {job_index}",
                )
            experiment_data.block_for_results()
            print(
                f"[{attempt_label}] experiment result received.",
                flush=True,
            )
            return experiment_data
        except Exception as exc:
            last_error = exc
            code, message = _job_error_code_and_message(active_job, exc)
            if attempt >= JOB_MAX_ATTEMPTS:
                print(
                    f"[{label}] experiment failed: "
                    f"job_id={_job_identifier(active_job)}, "
                    f"error_code={code}, retries_exhausted=True",
                    flush=True,
                )
                raise
            delay = JOB_RETRY_BASE_SECONDS * attempt
            print(
                f"[{label}] caught experiment error {code}; "
                "retry_policy=all_errors; "
                f"recreating the complete experiment in {delay}s. "
                f"Server message: {message}",
                flush=True,
            )
            time.sleep(delay)
    raise last_error


def append_hardware_error_history(backend, history_path):
    """Append one timestamped backend-calibration summary to a short text log."""
    from datetime import datetime
    from pathlib import Path
    from statistics import median

    def valid_probability(value):
        try:
            number = float(value)
        except (TypeError, ValueError):
            return None
        return number if np.isfinite(number) and 0.0 <= number <= 1.0 else None

    target = getattr(backend, "target", None)
    two_qubit_errors = []
    two_qubit_gate_names = set()
    seen_two_qubit_instructions = set()
    if target is not None:
        for operation_name in target.operation_names:
            try:
                instruction_entries = target[operation_name]
                entry_items = instruction_entries.items()
            except (KeyError, AttributeError, TypeError):
                continue
            for qargs, instruction_properties in entry_items:
                if qargs is None or len(qargs) != 2 or instruction_properties is None:
                    continue
                error = valid_probability(getattr(instruction_properties, "error", None))
                if error is None:
                    continue
                key = (str(operation_name), tuple(int(qubit) for qubit in qargs))
                if key in seen_two_qubit_instructions:
                    continue
                seen_two_qubit_instructions.add(key)
                two_qubit_errors.append(error)
                two_qubit_gate_names.add(str(operation_name))

    readout_errors = []
    if target is not None:
        try:
            measurement_items = target["measure"].items()
        except (KeyError, AttributeError, TypeError):
            measurement_items = ()
        for qargs, instruction_properties in measurement_items:
            if qargs is None or len(qargs) != 1 or instruction_properties is None:
                continue
            error = valid_probability(getattr(instruction_properties, "error", None))
            if error is not None:
                readout_errors.append(error)

    try:
        backend_properties = backend.properties()
    except Exception:
        backend_properties = None
    if not readout_errors and backend_properties is not None:
        for qubit in range(int(getattr(backend, "num_qubits", 0))):
            try:
                error = valid_probability(backend_properties.readout_error(qubit))
            except Exception:
                error = None
            if error is not None:
                readout_errors.append(error)

    backend_name = getattr(backend, "name", "unknown_backend")
    if callable(backend_name):
        backend_name = backend_name()
    queried_at = datetime.now().astimezone().isoformat(timespec="seconds")
    calibration_time = getattr(backend_properties, "last_update_date", None)
    if hasattr(calibration_time, "isoformat"):
        calibration_time = calibration_time.isoformat()
    calibration_time = str(calibration_time) if calibration_time is not None else "unavailable"
    median_qerror = median(two_qubit_errors) if two_qubit_errors else None
    median_readout = median(readout_errors) if readout_errors else None

    def format_value(value):
        return "unavailable" if value is None else f"{value:.10g}"

    history_path = Path(history_path)
    history_path.parent.mkdir(parents=True, exist_ok=True)
    lines = [
        f"[{queried_at}]",
        f"backend={backend_name}",
        f"calibration_last_update={calibration_time}",
        f"two_qubit_gate_names={','.join(sorted(two_qubit_gate_names)) or 'unavailable'}",
        f"median_two_qubit_qerror={format_value(median_qerror)}",
        f"two_qubit_qerror_samples={len(two_qubit_errors)}",
        f"median_readout_error={format_value(median_readout)}",
        f"readout_error_samples={len(readout_errors)}",
        "",
    ]
    with history_path.open("a", encoding="utf-8") as handle:
        handle.write("\n".join(lines) + "\n")
    print(f"Hardware calibration summary appended to {history_path}", flush=True)
    return {
        "backend": str(backend_name),
        "queried_at": queried_at,
        "calibration_last_update": calibration_time,
        "median_two_qubit_qerror": median_qerror,
        "median_readout_error": median_readout,
    }


def make_runtime_service():
    """Use a named IBM instance when configured to avoid ambiguous routing."""
    if RUNTIME_INSTANCE:
        print(f"Using IBM Quantum service instance: {RUNTIME_INSTANCE}")
        return QiskitRuntimeService(instance=RUNTIME_INSTANCE)
    print(
        "RUNTIME_INSTANCE is not set; IBM will choose among accessible "
        "service instances. Set it when your account has multiple projects."
    )
    return QiskitRuntimeService()


def validate_runtime_backend(
        backend, require_measure_2=False, wait_for_active=False):
    """Require IBM's authoritative ``operational + active`` state.

    The portal can display Online while the Runtime status endpoint still
    reports maintenance.  On a retry/resume path, wait and recheck every ten
    minutes instead of converting that transient state into a fatal error.
    Local simulators do not have a Runtime hardware status endpoint.
    """
    if _readout_is_simulator(backend):
        return backend
    while True:
        status = backend.status()
        operational = bool(getattr(status, "operational", False))
        pending_jobs = int(getattr(status, "pending_jobs", 0) or 0)
        status_message = str(
            getattr(status, "status_msg", "") or ""
        ).strip()
        active = operational and status_message.lower() == "active"
        print(
            f"Backend preflight: name={backend.name}, "
            f"operational={operational}, pending_jobs={pending_jobs}, "
            f"status={status_message or 'unknown'}",
            flush=True,
        )
        if active:
            break
        if not wait_for_active:
            raise RuntimeError(
                f"{backend.name} is not ready for a new workload "
                f"(operational={operational}, status={status_message!r}). "
                "Choose an active backend or wait for maintenance to finish."
            )
        print(
            f"{backend.name} is not active; checking again in "
            f"{MAINTENANCE_POLL_SECONDS // 60} minutes.",
            flush=True,
        )
        time.sleep(MAINTENANCE_POLL_SECONDS)
    if require_measure_2 and "measure_2" not in set(
            getattr(backend, "supported_instructions", ())):
        raise RuntimeError(
            f"{backend.name} does not advertise measure_2. Choose a backend "
            "that supports the MidCircuitMeasure instruction."
        )
    if pending_jobs >= BACKEND_PENDING_WARNING_THRESHOLD:
        print(
            f"WARNING: {backend.name} currently reports {pending_jobs} "
            "pending jobs. IBM error 1245 is more likely during heavy queue "
            "or service pressure; consider a less busy active backend.",
            flush=True,
        )
    return backend



def assert_terminal_measurements_are_final(circuits, label="workload"):
    """Reject active quantum work after an ordinary measurement.

    Runtime's dynamic scheduler can carry a measured qubit through a later
    control-flow container solely to align that container's branches. Such a
    formal qarg is not evidence that a gate acts on the measured qubit. Walk
    every branch and inspect its real instructions instead. Scheduler-only
    barriers and delays are harmless; gates, resets, control-flow work, and a
    second measurement on the same qubit are rejected.
    """
    circuit_items = [circuits] if hasattr(circuits, "data") else list(circuits)

    def walk_block(block, root_qubits, measured_at, pub_index, path):
        measured_at = dict(measured_at)
        for instruction_index, instruction in enumerate(block.data):
            operation = instruction.operation
            operation_name = operation.name
            nested_blocks = tuple(getattr(operation, "blocks", ()) or ())

            if nested_blocks:
                # A ControlFlowOp's formal qargs may include qubits that only
                # receive alignment delays. Map each block's local qubits to
                # the actual outer/root qubits and validate branch contents.
                outer_roots = [
                    root_qubits[block.find_bit(qubit).index]
                    for qubit in instruction.qubits
                ]
                branch_states = []
                for branch_index, nested_block in enumerate(nested_blocks):
                    if nested_block.num_qubits > len(outer_roots):
                        raise RuntimeError(
                            f"{label}: PUB {pub_index} has malformed "
                            f"{operation_name!r} qubit mapping at {path}."
                        )
                    branch_states.append(walk_block(
                        nested_block,
                        outer_roots[:nested_block.num_qubits],
                        measured_at,
                        pub_index,
                        f"{path}/{operation_name}[{branch_index}]",
                    ))

                # If an ordinary measurement occurs in any possible branch,
                # conservatively treat it as measured for subsequent work.
                for branch_state in branch_states:
                    for qubit, location in branch_state.items():
                        measured_at.setdefault(qubit, location)
                continue

            touched = {
                root_qubits[block.find_bit(qubit).index]
                for qubit in instruction.qubits
            }
            harmless_scheduler_ops = {"barrier", "delay"}
            violating = (
                touched.intersection(measured_at)
                if operation_name not in harmless_scheduler_ops
                else set()
            )
            if violating:
                details = ", ".join(
                    f"q{qubit} measured at {measured_at[qubit]}"
                    for qubit in sorted(violating)
                )
                raise RuntimeError(
                    f"{label}: PUB {pub_index} is unsafe: "
                    f"{operation_name!r} at {path}/instruction "
                    f"{instruction_index} acts after an ordinary terminal "
                    f"measurement ({details}). Submission stopped."
                )
            if operation_name == "measure":
                for qubit in touched:
                    measured_at[qubit] = (
                        f"{path}/instruction {instruction_index}"
                    )
        return measured_at

    for pub_index, item in enumerate(circuit_items):
        circuit = getattr(item, "circuit", None)
        if circuit is None and isinstance(item, (tuple, list)) and item:
            circuit = item[0]
        if circuit is None:
            circuit = item
        if not hasattr(circuit, "data"):
            raise TypeError(
                f"{label}: PUB {pub_index} does not contain a QuantumCircuit."
            )
        walk_block(
            circuit,
            list(range(circuit.num_qubits)),
            {},
            pub_index,
            "top-level",
        )
    return True


def _sampler_runtime_backend(sampler):
    """Return the IBM backend from a raw Sampler or notebook sampler proxy."""
    backend = getattr(sampler, "_backend", None)
    if backend is None:
        backend = getattr(getattr(sampler, "_sampler", None), "_backend", None)
    return backend
def _m3_calibration_sampler(backend, sampler):
    """Use a no-DD proxy for M3; preserve the experiment sampler and session.

    ManualDynamicDDSampler is defined later in the notebook. Plain hardware
    and simulator samplers used before that definition pass through unchanged.
    """
    dynamic_sampler_type = globals().get("ManualDynamicDDSampler")
    if (dynamic_sampler_type is None
            or not isinstance(sampler, dynamic_sampler_type)
            or sampler.policy == "none"):
        return sampler
    print(
        f"[M3 calibration] experiment policy={sampler.policy}; "
        "using policy=none for calibration; server built-in DD=False",
        flush=True,
    )
    # configure_mcm_policy_sampler already disables server DD. Reuse its raw
    # session sampler through a separate no-DD proxy, retaining ISA checks.
    # Do not change sampler.policy or its options: the sweep still needs DD.
    return dynamic_sampler_type(
        sampler._sampler, backend, "none", sampler.physical_qubits
    )


def calibrate_m3(
        backend, sampler, qubits, shots=M3_CALIBRATION_SHOTS,
        mid_circuit_qubits=(), label_suffix=None):
    """Build local M3 matrices without mixing measurement types in one job.

    IBM tunes ``measure_2`` separately from terminal ``measure``.  Each group
    is therefore calibrated in its own homogeneous Sampler job and the local
    assignment matrices are combined afterward.  This also avoids the mixed
    16-PUB payload that can fail at the IBM queue with error 1245.
    """
    # Prep-and-measure M3 circuits need no DD idle window. Route every MCM
    # calibration through no-DD, including legacy sweep entry points.
    sampler = _m3_calibration_sampler(backend, sampler)
    physical = list(dict.fromkeys(int(q) for q in qubits))
    mid_circuit_qubits = {int(q) for q in mid_circuit_qubits}
    unknown_mid = mid_circuit_qubits.difference(physical)
    if unknown_mid:
        raise ValueError(
            "measure_2 calibration qubits are outside the calibration set: "
            f"{sorted(unknown_mid)}"
        )

    calibration_groups = [("measure_2", physical, True)]

    matrices = [None] * int(backend.num_qubits)
    for calibration_kind, group_qubits, use_measure_2 in calibration_groups:
        width = len(group_qubits)
        order = 1
        while order < width + 1:
            order *= 2
        patterns = ((hadamard(order)[:, 1:width + 1] + 1) // 2).astype(int)
        shots_per_pattern = int(np.ceil(2 * int(shots) / len(patterns)))

        q = QuantumRegister(width, f"m3q_{calibration_kind}")
        c = ClassicalRegister(width, "m3cal")
        calibration_circuits = []
        for row_index, pattern in enumerate(patterns):
            qc = QuantumCircuit(q, c, name=f"m3_{calibration_kind}_{row_index:02d}")
            for logical, prepared_one in enumerate(pattern):
                if prepared_one:
                    qc.x(q[logical])
            for logical in range(width):
                if use_measure_2:
                    qc.append(MidCircuitMeasure(), [q[logical]], [c[logical]])
                else:
                    _append_measure_2(qc, q[logical], c[logical])
            calibration_circuits.append(qc)

        transpiled = transpile(
            calibration_circuits,
            backend=backend,
            initial_layout=group_qubits,
            optimization_level=0,
            seed_transpiler=SEED_TRANSPILER,
        )
        label = f"M3 {calibration_kind} calibration on {backend.name}"
        if label_suffix:
            label += f", {label_suffix}"
        result = run_sampler_job_with_retries(
            sampler, transpiled, shots_per_pattern, label
        )

        if len(result) != len(patterns):
            raise RuntimeError('M3 calibration result count does not match preparation patterns.')
        accumulators = [np.zeros((2, 2), dtype=float) for _ in group_qubits]
        for pub_result, pattern in zip(result, patterns):
            counts = pub_result.data.m3cal.get_counts()
            for key, count in counts.items():
                bits = normalize_bit_key(key, width)[::-1]
                for logical in range(width):
                    observed = int(bits[logical])
                    prepared = int(pattern[logical])
                    accumulators[logical][observed, prepared] += float(count)

        for logical, physical_qubit in enumerate(group_qubits):
            matrix = accumulators[logical]
            column_sums = matrix.sum(axis=0)
            if np.any(column_sums <= 0):
                raise RuntimeError(
                    f"Empty M3 calibration column for physical qubit "
                    f"{physical_qubit}: {matrix}."
                )
            matrices[physical_qubit] = matrix / column_sums
        print(
            f"[{label}] matrices ready for physical qubits {group_qubits}.",
            flush=True,
        )

    mitigator = mthree.M3Mitigation(backend)
    mitigator.cals_from_matrices(matrices)
    # Use M3's own calibration diagnostics and warning behavior.
    faulty = [int(qubit) for qubit in mitigator.faulty_qubits]
    fidelities = mitigator.readout_fidelity(physical)
    audit_path = _save_m3_calibration_audit({
        "backend": backend.name,
        "diagnostics_source": "mthree.M3Mitigation",
        "faulty_qubits": faulty,
        "calibrations": [
            {"physical_qubit": int(qubit),
             "matrix": np.asarray(matrix, dtype=float).tolist(),
             "readout_fidelity": float(fidelity)}
            for qubit, matrix, fidelity in zip(
                physical, [matrices[qubit] for qubit in physical], fidelities)
        ],
        "rows_are_observed_bits_columns_are_prepared_bits": True,
    }, OUTPUT_DIR)
    print(f"M3 built-in faulty qubits: {faulty}", flush=True)
    print(f"M3 readout fidelities: {dict(zip(physical, fidelities))}", flush=True)
    print(f"M3 diagnostic audit saved: {audit_path}", flush=True)
    print(
        f"M3 calibration complete for physical qubits {physical}.",
        flush=True,
    )
    return mitigator


def normalize_bit_key(key, width):
    if isinstance(key, (int, np.integer)):
        return format(int(key), f"0{width}b")
    text = str(key).replace(" ", "")
    if text.startswith(("0x", "0X")):
        text = format(int(text, 16), f"0{width}b")
    return text.zfill(width)


def probabilities_to_integer_counts(probabilities, shots):
    """Largest-remainder conversion required by tomography analysis."""
    keys = list(probabilities)
    probs = np.array(
        [max(0.0, float(probabilities[k])) for k in keys], dtype=float
    )
    if not keys or probs.sum() <= 0:
        raise ValueError("Empty corrected probability distribution.")
    probs /= probs.sum()
    scaled = probs * int(shots)
    counts = np.floor(scaled).astype(int)
    remainder = int(shots) - int(counts.sum())
    if remainder:
        order = np.argsort(-(scaled - counts))
        counts[order[:remainder]] += 1
    return {k: int(v) for k, v in zip(keys, counts) if v}


def _apply_m3_quasi(counts, measured_physical, mitigator):
    """Apply M3 with classical-bit 0 mapped to the rightmost count bit."""
    width = len(measured_physical)
    cleaned = {
        normalize_bit_key(key, width): int(value)
        for key, value in counts.items()
    }
    lengths = {len(key) for key in cleaned}
    if lengths != {width}:
        raise ValueError(
            f"Expected {width}-bit counts, received lengths {lengths}."
        )
    return mitigator.apply_correction(cleaned, list(measured_physical), distance=-1)


def m3_quasi_distribution(counts, measured_physical, mitigator):
    """Return M3 quasiprobabilities without a positivity projection.

    Direct expectation values, including CHSH correlators, should be evaluated
    from the quasiprobabilities.  Projecting each setting independently to its
    nearest probability distribution introduces an avoidable estimator bias.
    """
    width = len(measured_physical)
    quasi = _apply_m3_quasi(counts, measured_physical, mitigator)
    out = {}
    for key, value in quasi.items():
        bitstring = normalize_bit_key(key, width)
        out[bitstring] = out.get(bitstring, 0.0) + float(value)
    return out


def m3_probability_distribution(counts, measured_physical, mitigator):
    """Return a physical M3 distribution for tomography reconstruction."""
    width = len(measured_physical)
    quasi = _apply_m3_quasi(counts, measured_physical, mitigator)
    nearest = quasi.nearest_probability_distribution()
    out = {}
    for key, value in nearest.items():
        bitstring = normalize_bit_key(key, width)
        out[bitstring] = out.get(bitstring, 0.0) + float(value)
    return out



def sampler_counts_for_register(result, register_name):
    counts = []
    for pub_result in result:
        register_data = getattr(pub_result.data, register_name)
        counts.append(register_data.get_counts())
    return counts


def sampler_joint_dynamic_counts(result, register_names):
    """Join Sampler registers into Qiskit's global classical-bit order.

    ``BitArray.concatenate_bits([low_register, ..., high_register])`` emits
    the high register on the left, exactly like a Qiskit count string.  The
    circuit's register-creation order must therefore be passed directly and
    must not be reversed.  Reversing it maps M3 calibrations to the wrong bits
    and, in these circuits, made Round-1 CHSH use syndrome bits.
    """
    names = list(register_names)
    counts = []
    for pub_result in result:
        data = pub_result.data
        joined = BitArray.concatenate_bits([
            getattr(data, name) for name in names
        ])
        counts.append(joined.get_counts())
    return counts


def _validate_sampler_register_order_contract():
    """Fail early if the installed BitArray ordering contract changes."""
    low = BitArray.from_samples(["01"], num_bits=2)
    middle = BitArray.from_samples(["10"], num_bits=2)
    high = BitArray.from_samples(["11"], num_bits=2)
    observed = BitArray.concatenate_bits([low, middle, high]).get_bitstrings()[0]
    expected = "111001"  # high + middle + low
    if observed != expected:
        raise RuntimeError(
            "Unexpected BitArray.concatenate_bits ordering: "
            f"expected {expected}, observed {observed}."
        )


_validate_sampler_register_order_contract()




print("Backend and mitigation helpers loaded.")
```

## Matched tomography and direct CHSH runners


```python
def run_raw_and_m3_tomography(
        base_circuit, backend, sampler, physical_layout,
        measurement_indices, measured_classical_physical,
        mitigator, shots, run_label=None):
    """Run tomography once; reconstruct raw and M3 states from same shots."""
    qst = StateTomography(
        base_circuit,
        physical_qubits=list(physical_layout),
        measurement_indices=list(measurement_indices),
    )
    qst.set_transpile_options(optimization_level=OPTIMIZATION_LEVEL, seed_transpiler=SEED_TRANSPILER)
    label = run_label or (
        f"tomography on {backend.name}, logical indices "
        f"{list(measurement_indices)}"
    )
    raw_data = run_experiment_data_with_retries(
        lambda: qst.run(
            backend=backend, sampler=sampler, shots=int(shots)
        ),
        backend,
        label,
    )
    raw_df = raw_data.analysis_results(dataframe=True)
    raw_df = raw_df[raw_df["name"] == "state"].copy()
    if len(raw_df) != 1:
        raise RuntimeError(
            f"Expected one unconditional tomography state, got {len(raw_df)}."
        )

    width = len(measured_classical_physical)
    mitigated_data = ExperimentData(experiment=qst)
    for datum in raw_data.data():
        raw_counts = {
            str(key).replace(" ", ""): int(value)
            for key, value in datum["counts"].items()
        }
        if {len(key) for key in raw_counts} != {width}:
            raise ValueError(
                f"Tomography expected {width} measured bits; got "
                f"{sorted({len(key) for key in raw_counts})}."
            )
        probabilities = m3_probability_distribution(
            raw_counts, measured_classical_physical, mitigator
        )
        corrected = dict(datum)
        corrected["counts"] = probabilities_to_integer_counts(
            probabilities, sum(raw_counts.values())
        )
        mitigated_data.add_data(corrected)

    qst.analysis.run(
        mitigated_data, replace_results=True
    ).block_for_results()
    m3_df = mitigated_data.analysis_results(dataframe=True)
    m3_df = m3_df[m3_df["name"] == "state"].copy()
    if len(m3_df) != 1:
        raise RuntimeError(
            f"Expected one M3 tomography state, got {len(m3_df)}."
        )

    return {
        "raw": np.asarray(raw_df.iloc[0]["value"].data, dtype=complex),
        "m3": np.asarray(m3_df.iloc[0]["value"].data, dtype=complex),
    }


CHSH_ANGLES = {
    "a0": 0.0,
    "a1": np.pi / 2,
    "b0": np.pi / 4,
    "b1": -np.pi / 4,
}
CHSH_SETTINGS = [
    ("a0", "b0"), ("a0", "b1"),
    ("a1", "b0"), ("a1", "b1"),
]


def build_direct_chsh_circuits(base_circuit, measured_logical_indices):
    circuits = {}
    left, right = [int(i) for i in measured_logical_indices]
    for setting_a, setting_b in CHSH_SETTINGS:
        qc = base_circuit.copy()
        chsh = ClassicalRegister(2, "chsh")
        qc.add_register(chsh)
        # Keep final CHSH analysis and measurements after all feedback.
        qc.barrier(*qc.qubits)
        qc.ry(-CHSH_ANGLES[setting_a], qc.qubits[left])
        qc.ry(-CHSH_ANGLES[setting_b], qc.qubits[right])
        _append_measure_2(qc, qc.qubits[left], chsh[0])
        _append_measure_2(qc, qc.qubits[right], chsh[1])
        circuits[(setting_a, setting_b)] = qc
    return circuits


def correlator(distribution, left_cbit, right_cbit, total_width):
    numerator = denominator = 0.0
    for key, value in distribution.items():
        bits = normalize_bit_key(key, total_width)[::-1]
        eigenvalue = 1.0 if bits[left_cbit] == bits[right_cbit] else -1.0
        numerator += float(value) * eigenvalue
        denominator += float(value)
    if denominator <= 0:
        raise ValueError("CHSH distribution has zero weight.")
    return numerator / denominator


def compute_chsh(distributions, left_cbit, right_cbit, total_width):
    expectations = {
        setting: correlator(
            dist, left_cbit, right_cbit, total_width
        )
        for setting, dist in distributions.items()
    }
    return float(
        expectations[("a0", "b0")]
        + expectations[("a0", "b1")]
        + expectations[("a1", "b0")]
        - expectations[("a1", "b1")]
    )


def run_raw_and_m3_chsh(
        base_circuit, backend, sampler, physical_layout,
        measured_logical_indices, measured_classical_physical,
        mitigator, shots, run_label=None):
    """Measure CHSH on the same logical/physical pair used by tomography."""
    measured_logical_indices = [int(i) for i in measured_logical_indices]
    circuits_by_setting = build_direct_chsh_circuits(
        base_circuit, measured_logical_indices
    )
    settings = list(circuits_by_setting)
    transpiled = transpile(
        [circuits_by_setting[s] for s in settings],
        backend=backend,
        initial_layout=list(physical_layout),
        optimization_level=OPTIMIZATION_LEVEL,
        seed_transpiler=SEED_TRANSPILER,
    )
    label = run_label or (
        f"direct CHSH on {backend.name}, logical indices "
        f"{measured_logical_indices}"
    )
    result = run_sampler_job_with_retries(
        sampler, transpiled, int(shots), label
    )
    expected_pair = [
        int(physical_layout[measured_logical_indices[0]]),
        int(physical_layout[measured_logical_indices[1]]),
    ]
    if base_circuit.num_clbits:
        register_names = [reg.name for reg in base_circuit.cregs] + ["chsh"]
        raw_list = sampler_joint_dynamic_counts(result, register_names)
        total_width = base_circuit.num_clbits + 2
        left_cbit = base_circuit.num_clbits
        right_cbit = left_cbit + 1
        correction_qubits = [int(q) for q in measured_classical_physical]
        if len(correction_qubits) != total_width:
            raise RuntimeError(
                f"CHSH/M3 width mismatch: circuit has {total_width} measured "
                f"bits but the correction map has {len(correction_qubits)}."
            )
        if correction_qubits[-2:] != expected_pair:
            raise RuntimeError(
                "Tomography/CHSH target-pair mismatch: final M3 qubits are "
                f"{correction_qubits[-2:]}, expected {expected_pair}."
            )
    else:
        raw_list = sampler_counts_for_register(result, "chsh")
        total_width = 2
        left_cbit, right_cbit = 0, 1
        correction_qubits = expected_pair

    if len(raw_list) != len(settings):
        raise RuntimeError(
            f"Expected {len(settings)} CHSH result distributions, got "
            f"{len(raw_list)}."
        )
    raw_distributions = {}
    m3_distributions = {}
    for setting, counts in zip(settings, raw_list):
        cleaned = {
            normalize_bit_key(key, total_width): int(value)
            for key, value in counts.items()
        }
        raw_distributions[setting] = cleaned
        # Use the same physical-probability projection as the validated
        # standalone CHSH implementation.  Unprojected M3 quasiprobabilities
        # can amplify an ill-conditioned readout calibration and produce
        # algebraic-but-unphysical CHSH values (observed for channel B).
        m3_distributions[setting] = m3_probability_distribution(
            cleaned, correction_qubits, mitigator
        )

    return {
        "raw": compute_chsh(
            raw_distributions, left_cbit, right_cbit, total_width
        ),
        "m3": compute_chsh(
            m3_distributions, left_cbit, right_cbit, total_width
        ),
    }



print("Matched tomography and direct-CHSH runners loaded.")
```

```python
# Run this definition-only cell AFTER the older CHSH definitions and BEFORE the sweep.
# It restores the complete CHSH result processing and return value.
# This cell does not submit jobs. Calling the sweep afterward does.
def find_barriers(circuit):
    """Report root-circuit indices (logical before transpilation, physical after)."""
    found = []
    def walk(block, roots, path):
        for index, item in enumerate(block.data):
            qubits = [roots[block.find_bit(q).index] for q in item.qubits]
            location = f"{path}/{index}"
            if item.operation.name == "barrier":
                found.append((location, qubits))
            for branch, nested in enumerate(getattr(item.operation, "blocks", ())):
                walk(nested, qubits, f"{location}/branch{branch}")
    walk(circuit, list(range(circuit.num_qubits)), "root")
    print(f"{len(found)} barrier(s) in {circuit.name!r}: {found}")
    return found


def run_raw_and_m3_chsh(
        base_circuit, backend, sampler, physical_layout,
        measured_logical_indices, measured_classical_physical,
        mitigator, shots, run_label=None):
    """Measure CHSH on the same logical/physical pair used by tomography."""
    measured_logical_indices = [int(i) for i in measured_logical_indices]
    circuits_by_setting = build_direct_chsh_circuits(
        base_circuit, measured_logical_indices
    )
    settings = list(circuits_by_setting)
    print(f"[barrier check | pre-transpile | {run_label}]")
    for setting in settings:
        print(f"Setting {setting}:")
        find_barriers(circuits_by_setting[setting])
    transpiled = transpile(
        [circuits_by_setting[s] for s in settings],
        backend=backend,
        initial_layout=list(physical_layout),
        optimization_level=OPTIMIZATION_LEVEL,
        seed_transpiler=SEED_TRANSPILER,
    )
    print(f"[barrier check | post-transpile | level={OPTIMIZATION_LEVEL} | {run_label}]")
    for setting, circuit in zip(settings, transpiled):
        print(f"Setting {setting}:")
        find_barriers(circuit)
    label = run_label or (
        f"direct CHSH on {backend.name}, logical indices "
        f"{measured_logical_indices}"
    )
    result = run_sampler_job_with_retries(
        sampler, transpiled, int(shots), label
    )
    expected_pair = [
        int(physical_layout[measured_logical_indices[0]]),
        int(physical_layout[measured_logical_indices[1]]),
    ]
    if base_circuit.num_clbits:
        register_names = [reg.name for reg in base_circuit.cregs] + ["chsh"]
        raw_list = sampler_joint_dynamic_counts(result, register_names)
        total_width = base_circuit.num_clbits + 2
        left_cbit = base_circuit.num_clbits
        right_cbit = left_cbit + 1
        correction_qubits = [int(q) for q in measured_classical_physical]
        if len(correction_qubits) != total_width:
            raise RuntimeError(
                f"CHSH/M3 width mismatch: circuit has {total_width} measured "
                f"bits but the correction map has {len(correction_qubits)}."
            )
        if correction_qubits[-2:] != expected_pair:
            raise RuntimeError(
                "Tomography/CHSH target-pair mismatch: final M3 qubits are "
                f"{correction_qubits[-2:]}, expected {expected_pair}."
            )
    else:
        raw_list = sampler_counts_for_register(result, "chsh")
        total_width = 2
        left_cbit, right_cbit = 0, 1
        correction_qubits = expected_pair

    if len(raw_list) != len(settings):
        raise RuntimeError(
            f"Expected {len(settings)} CHSH result distributions, got "
            f"{len(raw_list)}."
        )
    raw_distributions = {}
    m3_distributions = {}
    for setting, counts in zip(settings, raw_list):
        cleaned = {
            normalize_bit_key(key, total_width): int(value)
            for key, value in counts.items()
        }
        raw_distributions[setting] = cleaned
        # Use the same physical-probability projection as the validated
        # standalone CHSH implementation.  Unprojected M3 quasiprobabilities
        # can amplify an ill-conditioned readout calibration and produce
        # algebraic-but-unphysical CHSH values (observed for channel B).
        m3_distributions[setting] = m3_probability_distribution(
            cleaned, correction_qubits, mitigator
        )

    return {
        "raw": compute_chsh(
            raw_distributions, left_cbit, right_cbit, total_width
        ),
        "m3": compute_chsh(
            m3_distributions, left_cbit, right_cbit, total_width
        ),
    }

print("CHSH fix loaded: run_raw_and_m3_chsh now returns raw and m3 scores.")
```

```python
def configure_hardware_sampler(sampler):
    if ENABLE_DD:
        raise RuntimeError(
            "Built-in Runtime DD cannot be combined with this dynamic MCM "
            "workload. Use the DCM notebook for the matched-DD experiment."
        )
    if SHOW_SCHEDULER_TIMING:
        sampler.options.experimental = {
            "execution": {"scheduler_timing": True}
        }
    return sampler


def initialize_experiment():
    service = QiskitRuntimeService()
    backend_real = service.backend(BACKEND_NAME)
    supported = set(getattr(backend_real, "supported_instructions", ()))
    if "measure_2" not in supported:
        raise RuntimeError(
            f"{BACKEND_NAME} does not advertise measure_2. Choose a backend "
            "returned by service.backends(filters=lambda b: 'measure_2' in "
            "b.supported_instructions)."
        )
    backend_sim = AerSimulator.from_backend(backend_real)
    sampler_real = configure_hardware_sampler(
        make_hardware_session_sampler(backend_real)
    )
    sampler_sim = Sampler(mode=backend_sim)

    # Layout ranking does not depend on which measurement instruction is used.
    layout_proxy = build_epr_half_teleportation_circuit(
        with_corrections=True, use_measure_2=False
    )
    full_layout = select_balanced_physical_layout(layout_proxy, backend_real)
    print(f"Selected nine-qubit layout: {full_layout}")

    measured_logical = (2, 4, 3, 5)  # in_A, q0, in_B, q1
    measured_physical = [(full_layout[index],) for index in measured_logical]
    try:
        supported_qargs = set(
            backend_real.target.qargs_for_operation_name("measure_2")
        )
    except AttributeError:
        supported_qargs = set(backend_real.target["measure_2"])
    missing = [qargs for qargs in measured_physical if qargs not in supported_qargs]
    if missing:
        raise RuntimeError(
            f"{BACKEND_NAME} does not support measure_2 on selected physical "
            f"qubits {[q[0] for q in missing]}."
        )

    syndrome_physical = [qargs[0] for qargs in measured_physical]
    m3_real_terminal = calibrate_m3(
        backend_real, sampler_real, full_layout,
        shots=M3_CALIBRATION_SHOTS
    )
    m3_real_mixed = calibrate_m3(
        backend_real, sampler_real, full_layout,
        shots=M3_CALIBRATION_SHOTS,
        mid_circuit_qubits=syndrome_physical,
    )
    m3_sim = calibrate_m3(
        backend_sim, sampler_sim, full_layout,
        shots=M3_CALIBRATION_SHOTS
    )
    if getattr(m3_sim, "system", backend_sim) is not backend_sim:
        raise AssertionError("Simulator M3 instance mismatch.")

    return {
        "backend_real": backend_real,
        "backend_sim": backend_sim,
        "sampler_real": sampler_real,
        "sampler_sim": sampler_sim,
        "full_layout": full_layout,
        "m3_real_terminal": m3_real_terminal,
        "m3_real_mixed": m3_real_mixed,
        "m3_sim": m3_sim,
    }


def append_record(records, trial, domain, method, channel, phase, rho, direct_s):
    fidelity, chsh_from_rho = metrics_from_density_matrix(rho)
    direct_lower = direct_s / (2 * np.sqrt(2))
    direct_upper = direct_s / (4 * np.sqrt(2)) + 0.5
    records.append({
        "measurement_variant": EXPERIMENT_VARIANT,
        "dd_enabled": bool(ENABLE_DD and domain == "Hardware"),
        "dd_sequence": (
            DD_SEQUENCE if ENABLE_DD and domain == "Hardware" else "none"
        ),
        "trial": int(trial),
        "domain": domain,
        "method": method,
        "condition_key": (
            "a_real" if (domain, method) == ("Hardware", "raw") else
            "b_real_m3" if (domain, method) == ("Hardware", "m3") else
            "c_sim" if (domain, method) == ("Noisy simulation", "raw") else
            "d_sim_m3"
        ),
        "channel": channel,
        "phase": phase,
        "fidelity": fidelity,
        "chsh_S_direct": float(direct_s),
        "chsh_S_from_rho": chsh_from_rho,
        "equation9_lower_from_direct_S": direct_lower,
        "equation9_upper_from_direct_S": direct_upper,
        "equation9_direct_point_pass": bool(
            direct_lower <= fidelity <= direct_upper
        ),
    })


def run_corrected_experiment():
    env = initialize_experiment()
    full_layout = env["full_layout"]
    source_base = build_source_epr_pair_circuit()
    transported_hardware = build_epr_half_teleportation_circuit(
        with_corrections=True, use_measure_2=True
    )
    transported_simulator = build_epr_half_teleportation_circuit(
        with_corrections=True, use_measure_2=False
    )

    channel_specs = {
        "A": {
            "source_layout": [full_layout[0], full_layout[2]],
            "after_indices": [0, 8],
            # QST cbit order: mA0,mA1,mB0,mB1,tomo(refA),tomo(q4)
            "after_m3_qubits": [
                full_layout[2], full_layout[4],
                full_layout[3], full_layout[5],
                full_layout[0], full_layout[8],
            ],
        },
        "B": {
            "source_layout": [full_layout[1], full_layout[3]],
            "after_indices": [1, 7],
            # QST cbit order: mA0,mA1,mB0,mB1,tomo(refB),tomo(q3)
            "after_m3_qubits": [
                full_layout[2], full_layout[4],
                full_layout[3], full_layout[5],
                full_layout[1], full_layout[7],
            ],
        },
    }

    records = []
    for trial in range(1, N_TRIALS + 1):
        print("\n" + "=" * 76)
        print(f"Corrected same-state trial {trial}/{N_TRIALS}")
        print("=" * 76)

        for (
            domain, backend, sampler, before_mitigator, after_mitigator,
            transported_base,
        ) in [
            (
                "Hardware", env["backend_real"], env["sampler_real"],
                env["m3_real_terminal"], env["m3_real_mixed"],
                transported_hardware,
            ),
            (
                "Noisy simulation", env["backend_sim"], env["sampler_sim"],
                env["m3_sim"], env["m3_sim"], transported_simulator,
            ),
        ]:
            for channel, spec in channel_specs.items():
                print(f"{domain}, channel {channel}, before tomography/CHSH...")
                before_rhos = run_raw_and_m3_tomography(
                    source_base, backend, sampler,
                    spec["source_layout"], [0, 1],
                    spec["source_layout"], before_mitigator, SHOTS,
                )
                before_s = run_raw_and_m3_chsh(
                    source_base, backend, sampler,
                    spec["source_layout"], [0, 1],
                    spec["source_layout"], before_mitigator, SHOTS,
                )
                for method in ("raw", "m3"):
                    append_record(
                        records, trial, domain, method, channel, "before",
                        before_rhos[method], before_s[method],
                    )

                print(f"{domain}, channel {channel}, after tomography/CHSH...")
                after_rhos = run_raw_and_m3_tomography(
                    transported_base, backend, sampler,
                    full_layout, spec["after_indices"],
                    spec["after_m3_qubits"], after_mitigator, SHOTS,
                )
                after_s = run_raw_and_m3_chsh(
                    transported_base, backend, sampler,
                    full_layout, spec["after_indices"],
                    spec["after_m3_qubits"], after_mitigator, SHOTS,
                )
                for method in ("raw", "m3"):
                    append_record(
                        records, trial, domain, method, channel, "after",
                        after_rhos[method], after_s[method],
                    )

    return pd.DataFrame(records)


print("Trial driver loaded.")
```

```python
def mean_ci99(values):
    values = np.asarray(values, dtype=float)
    mean = float(values.mean())
    if len(values) < 2:
        return mean, 0.0
    critical = float(student_t.ppf(0.995, df=len(values) - 1))
    halfwidth = critical * float(values.std(ddof=1)) / np.sqrt(len(values))
    return mean, halfwidth


def summarize_metric(records, channel, metric):
    subset = records[records["channel"] == channel].copy()
    rows = []
    for (condition, phase), group in subset.groupby(
            ["condition_key", "phase"], sort=False):
        mean, ci = mean_ci99(group[metric])
        row = {
            "measurement_variant": EXPERIMENT_VARIANT,
            "dd_enabled": bool(group["dd_enabled"].iloc[0]),
            "dd_sequence": str(group["dd_sequence"].iloc[0]),
            "condition_key": condition,
            "condition_label": CONDITION_LABELS[condition],
            "phase": phase,
            "metric_key": metric,
            "mean": mean,
            "ci99_halfwidth": ci,
            "n_trials": len(group),
        }
        if metric == "fidelity":
            direct_s_mean = float(group["chsh_S_direct"].mean())
            row.update({
                "paired_direct_chsh_mean": direct_s_mean,
                "equation9_lower_from_direct_S_mean": (
                    direct_s_mean / (2 * np.sqrt(2))
                ),
                "equation9_upper_from_direct_S_mean": (
                    direct_s_mean / (4 * np.sqrt(2)) + 0.5
                ),
                "equation9_mean_point_pass": bool(
                    direct_s_mean / (2 * np.sqrt(2)) <= mean
                    <= direct_s_mean / (4 * np.sqrt(2)) + 0.5
                ),
                "tomography_derived_chsh_mean": float(
                    group["chsh_S_from_rho"].mean()
                ),
            })
        elif metric == "chsh_S_direct":
            row["tomography_derived_chsh_mean"] = float(
                group["chsh_S_from_rho"].mean()
            )
        rows.append(row)
    result = pd.DataFrame(rows)
    result["condition_order"] = result["condition_key"].map(
        {key: i for i, key in enumerate(CONDITION_ORDER)}
    )
    result["phase_order"] = result["phase"].map({"before": 0, "after": 1})
    return (
        result.sort_values(["condition_order", "phase_order"])
        .drop(columns=["condition_order", "phase_order"])
        .reset_index(drop=True)
    )


def plot_before_after(summary, channel, metric, save_path):
    x = np.arange(len(CONDITION_ORDER), dtype=float)
    width = 0.34
    fig, ax = plt.subplots(figsize=(10.5, 6.2))
    colors = {"before": "#A6CEE3", "after": "#1F78B4"}
    labels = {"before": "Before (source pair)", "after": "After (teleported)"}

    for offset, phase in [(-width / 2, "before"), (width / 2, "after")]:
        phase_rows = (
            summary[summary["phase"] == phase]
            .set_index("condition_key").loc[CONDITION_ORDER]
        )
        ax.bar(
            x + offset,
            phase_rows["mean"].to_numpy(float),
            width,
            yerr=phase_rows["ci99_halfwidth"].to_numpy(float),
            capsize=4,
            color=colors[phase],
            edgecolor="black",
            linewidth=0.7,
            label=labels[phase],
        )

    ax.set_xticks(x)
    ax.set_xticklabels(
        [CONDITION_LABELS[key] for key in CONDITION_ORDER],
        rotation=22, ha="right",
    )
    ax.set_xlabel("Condition")
    ax.grid(axis="y", alpha=0.25)

    pair = "(ref_A, q4)" if channel == "A" else "(ref_B, q3)"
    if metric == "fidelity":
        ax.axhline(1.0, color="gray", linestyle=":", label="Ideal")
        ax.set_ylim(0, 1.10)
        ax.set_ylabel("Entanglement fidelity to |Phi+>")
        ax.set_title(
            f"{EXPERIMENT_VARIANT} (DD={'on' if ENABLE_DD else 'off'}): "
            f"same-state EPR-half fidelity, "
            f"channel {channel} {pair}\n5 trials; Student's-t 99% CI"
        )
    else:
        ax.axhline(2.0, color="#555555", linestyle="--", label="Classical bound")
        ax.axhline(2 * np.sqrt(2), color="gray", linestyle=":", label="Tsirelson bound")
        ax.set_ylim(0, 3.05)
        ax.set_ylabel("Direct CHSH S")
        ax.set_title(
            f"{EXPERIMENT_VARIANT} (DD={'on' if ENABLE_DD else 'off'}): "
            f"matched EPR-half CHSH, channel {channel} "
            f"{pair}\n5 trials; Student's-t 99% CI"
        )

    ax.legend(loc="best", fontsize=9)
    fig.tight_layout()
    fig.savefig(save_path, dpi=180, bbox_inches="tight")
    plt.show()


def save_four_outputs(records):
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    outputs = {}
    for channel in ("A", "B"):
        fidelity_summary = summarize_metric(records, channel, "fidelity")
        chsh_summary = summarize_metric(records, channel, "chsh_S_direct")

        fidelity_csv = OUTPUT_DIR / OUTPUT_NAMES[(channel, "fidelity")]
        chsh_csv = OUTPUT_DIR / OUTPUT_NAMES[(channel, "chsh")]
        fidelity_png = OUTPUT_DIR / OUTPUT_NAMES[(channel, "fidelity_png")]
        chsh_png = OUTPUT_DIR / OUTPUT_NAMES[(channel, "chsh_png")]

        fidelity_summary.to_csv(fidelity_csv, index=False)
        chsh_summary.to_csv(chsh_csv, index=False)
        plot_before_after(fidelity_summary, channel, "fidelity", fidelity_png)
        plot_before_after(chsh_summary, channel, "chsh", chsh_png)
        outputs[channel] = {
            "fidelity_csv": fidelity_csv,
            "chsh_csv": chsh_csv,
            "fidelity_png": fidelity_png,
            "chsh_png": chsh_png,
        }

    print("\nGenerated files:")
    for channel_outputs in outputs.values():
        for path in channel_outputs.values():
            print(f"  {path.resolve()}")
    return outputs


print("Summary, CSV and plotting helpers loaded.")


SWEEP_GROUP_ORDER = [
    "none_raw", "none_m3",
    "XY4_raw", "XY4_m3",
    "XX_raw", "XX_m3",
]
SWEEP_GROUP_LABELS = {
    "none_raw": "No DD\nHW raw",
    "none_m3": "No DD\nHW + M3",
    "XY4_raw": "XY4 DD\nHW raw",
    "XY4_m3": "XY4 DD\nHW + M3",
    "XX_raw": "XX DD\nHW raw",
    "XX_m3": "XX DD\nHW + M3",
}


def _recursive_count_ops(circuit):
    try:
        return dict(circuit.count_ops(recurse=True))
    except TypeError:
        return dict(circuit.count_ops())


_MCM_DD_SCHEDULE_AUDIT = []


def _backend_pulse_alignment(backend, target):
    """Resolve pulse alignment from the IBM backend, then the target."""
    try:
        constraints = backend.configuration().timing_constraints or {}
    except Exception:
        constraints = {}
    return int(
        constraints.get("pulse_alignment")
        or getattr(target, "pulse_alignment", None)
        or 16
    )


def build_mcm_scheduling_pipeline(backend, policy, physical_qubits):
    """Build the safe local scheduling stage used by MCM DD policies.

    The complete dynamic circuit is first scheduled with ALAP and PadDelay,
    with end-of-circuit padding disabled.  DD gates are inserted afterward by
    ``_insert_control_flow_safe_dd`` only into top-level delays that precede
    every ordinary terminal measurement.  This avoids PadDynamicalDecoupling's
    block-end padding, which previously placed XY4 gates after tomography
    measurements and caused IBM 9699 failures.
    """
    if policy not in DD_POLICIES:
        raise ValueError(f"Unknown DD policy {policy!r}; choose {DD_POLICIES}.")

    target = copy.deepcopy(backend.target)
    native_basis = list(dict.fromkeys([
        *backend.operation_names,
        "if_else", "while_loop", "for_loop", "switch_case",
    ]))
    pulse_alignment = _backend_pulse_alignment(backend, target)
    translate_pm = None

    # Y is used only as a temporary logical marker.  Scheduling is based on
    # the calibrated X duration, and the marker is translated after insertion
    # to X followed by the zero-duration virtual frame change RZ(pi).
    if policy == "XY4" and "y" not in target.operation_names:
        if "x" not in target.operation_names:
            raise RuntimeError("Backend target has no calibrated X gate.")
        xy4_equivalences = EquivalenceLibrary(base=sel)
        y_native = QuantumCircuit(1, name="native_y_for_xy4")
        y_native.x(0)
        y_native.rz(np.pi, 0)
        xy4_equivalences.add_equivalence(YGate(), y_native)
        translate_pm = PassManager([
            BasisTranslator(xy4_equivalences, native_basis)
        ])

    schedule_pm = PassManager([
        ALAPScheduleAnalysis(target=target),
        PadDelay(
            target=target,
            fill_very_end=False,
            schedule_idle_qubits=False,
        ),
    ])
    return schedule_pm, translate_pm, pulse_alignment


def _assert_dynamic_structure_preserved(before, after, policy, pub_index):
    """Fail if scheduling alters measure_2 or feedback control flow."""
    for operation in ("measure_2", "if_else"):
        if before.get(operation, 0) != after.get(operation, 0):
            raise RuntimeError(
                f"MCM {policy} PUB {pub_index}: scheduling changed "
                f"{operation} count from {before.get(operation, 0)} to "
                f"{after.get(operation, 0)}. Submission stopped."
            )


def _append_mcm_schedule_audit(rows):
    _MCM_DD_SCHEDULE_AUDIT.extend(rows)
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    pd.DataFrame(_MCM_DD_SCHEDULE_AUDIT).to_csv(
        MCM_DD_AUDIT_CSV, index=False
    )


def _feedback_target_qubits(circuits):
    """Physical qubits directly addressed by control-flow bodies."""
    protected = set()
    for circuit in circuits:
        for item in circuit.data:
            if isinstance(item.operation, ControlFlowOp):
                protected.update(
                    circuit.find_bit(qubit).index
                    for qubit in item.qubits
                )
    return protected


def _nested_operation_count(circuit, operation_name):
    """Count an operation only inside control-flow block bodies."""
    total = 0

    def count_block(block):
        nonlocal total
        for item in block.data:
            if item.operation.name == operation_name:
                total += 1
            if isinstance(item.operation, ControlFlowOp):
                for nested in item.operation.blocks:
                    count_block(nested)

    for item in circuit.data:
        if isinstance(item.operation, ControlFlowOp):
            for block in item.operation.blocks:
                count_block(block)
    return total


def _assert_global_terminal_measurement_boundary(circuits, label):
    """Reject active work anywhere after an ordinary terminal measurement.

    IBM ``measure`` is reserved for the final acquisition block in these MCM
    workloads.  Mid-circuit acquisition is represented by ``measure_2``.
    Barriers and delays are harmless, but no gate, reset, or control-flow
    operation may follow the first ordinary measure in a block.
    """
    circuit_items = [circuits] if hasattr(circuits, "data") else list(circuits)

    def check_block(block, pub_index, path):
        terminal_started = False
        for instruction_index, item in enumerate(block.data):
            operation = item.operation
            name = operation.name
            if terminal_started and name not in {"measure", "barrier", "delay"}:
                raise RuntimeError(
                    f"{label}: PUB {pub_index} has active instruction {name!r} "
                    f"after the ordinary terminal-measurement block began at "
                    f"{path}/instruction {instruction_index}. Submission stopped."
                )
            if name == "measure":
                terminal_started = True
            for branch_index, nested in enumerate(
                    tuple(getattr(operation, "blocks", ()) or ())):
                check_block(
                    nested, pub_index,
                    f"{path}/{name}[{branch_index}]",
                )

    for pub_index, item in enumerate(circuit_items):
        circuit = getattr(item, "circuit", None)
        if circuit is None and isinstance(item, (tuple, list)) and item:
            circuit = item[0]
        if circuit is None:
            circuit = item
        check_block(circuit, pub_index, "top-level")
    return True


def _insert_control_flow_safe_dd(
        circuit, backend, allowed_qubits, pulse_alignment, policy):
    """Replace eligible top-level delays with an XX or XY4 identity cycle.

    Control-flow bodies are copied without modification.  DD is never inserted
    after ``measure_2`` on the measured wire, after an ordinary terminal
    measurement anywhere, or into end-of-circuit padding.  Every replacement
    has exactly the same duration as the delay it replaces.
    """
    if policy == "XX":
        sequence = ("x", "x")
        spacing = np.asarray([0.25, 0.50, 0.25], dtype=float)
    elif policy == "XY4":
        sequence = ("x", "y", "x", "y")
        spacing = np.asarray([0.125, 0.25, 0.25, 0.25, 0.125], dtype=float)
    else:
        raise ValueError(f"Manual DD insertion is unavailable for {policy!r}.")

    allowed = set(map(int, allowed_qubits))
    alignment = max(1, int(pulse_alignment))
    durations = backend.target.durations()
    rewritten = circuit.copy_empty_like()
    measured_qubits = set()
    terminal_started = False
    cycles = 0
    boundary_cycles = {}
    skipped_after_measure_2 = 0
    skipped_after_terminal_measure = 0

    for item in circuit.data:
        qindices = [circuit.find_bit(q).index for q in item.qubits]
        cindices = [circuit.find_bit(c).index for c in item.clbits]
        new_qargs = [rewritten.qubits[index] for index in qindices]
        new_cargs = [rewritten.clbits[index] for index in cindices]
        operation = item.operation
        name = operation.name

        if name == "measure":
            terminal_started = True

        eligible_delay = (
            name == "delay"
            and len(qindices) == 1
            and qindices[0] in allowed
            and operation.unit == "dt"
        )
        if eligible_delay and terminal_started:
            skipped_after_terminal_measure += 1
        if eligible_delay and qindices[0] in measured_qubits:
            skipped_after_measure_2 += 1

        if (
            not eligible_delay
            or terminal_started
            or qindices[0] in measured_qubits
        ):
            rewritten.append(operation, new_qargs, new_cargs)
        else:
            qubit = qindices[0]
            interval = int(operation.duration)
            try:
                x_duration = int(round(
                    durations.get("x", qubit, unit="dt")
                ))
            except Exception:
                rewritten.append(operation, new_qargs, new_cargs)
                x_duration = 0

            sequence_duration = len(sequence) * x_duration
            if (
                x_duration <= 0
                or x_duration % alignment != 0
                or interval / sequence_duration <= 2.0
            ):
                if x_duration > 0:
                    rewritten.append(operation, new_qargs, new_cargs)
            else:
                slack = interval - sequence_duration
                taus = (
                    alignment * np.floor(slack * spacing / alignment)
                ).astype(int)
                extra = int(slack - int(taus.sum()))
                middle = (len(taus) - 1) // 2
                to_middle = alignment * (extra // alignment)
                taus[middle] += to_middle
                taus[-1] += extra - to_middle
                if int(taus.sum()) + sequence_duration != interval:
                    raise RuntimeError(
                        f"Control-flow-safe {policy} changed delay "
                        f"duration {interval}."
                    )

                for gate_index, gate_name in enumerate(sequence):
                    if taus[gate_index] > 0:
                        rewritten.delay(
                            int(taus[gate_index]), new_qargs[0], unit="dt"
                        )
                    if gate_name == "x":
                        rewritten.x(new_qargs[0])
                    else:
                        rewritten.y(new_qargs[0])
                if taus[-1] > 0:
                    rewritten.delay(int(taus[-1]), new_qargs[0], unit="dt")
                cycles += 1

        if name in {"measure", "measure_2"}:
            measured_qubits.update(qindices)
        elif name == "reset":
            measured_qubits.difference_update(qindices)
        label = getattr(operation, 'label', None)
        if name == 'barrier' and label in {f'CUMULATIVE_ROUND_{r}' for r in range(4)}:
            boundary_cycles[int(label.rsplit('_', 1)[1])] = cycles

    # XYXY = -I.  Restore exact circuit equivalence, including global phase.
    if policy == "XY4" and cycles % 2:
        rewritten.global_phase += np.pi

    return rewritten, {
        "dd_cycles": cycles,
        "dd_cycles_by_round": boundary_cycles,
        "skipped_delays_after_measure_2": skipped_after_measure_2,
        "skipped_delays_after_terminal_measure": skipped_after_terminal_measure,
    }


class ManualDynamicDDSampler:
    """SamplerV2 proxy with conservative local DD for MCM circuits."""

    def __init__(self, sampler, backend, policy, physical_qubits):
        self._sampler = sampler
        self._backend = backend
        self.options = sampler.options
        self.policy = policy
        self.physical_qubits = list(physical_qubits)
        self._calls = 0
        (
            self._schedule_pm,
            self._translate_pm,
            self._pulse_alignment,
        ) = build_mcm_scheduling_pipeline(
            backend, policy, self.physical_qubits
        )

    def run(self, pubs, **run_options):
        default_shots = run_options.pop("shots", None)
        coerced_pubs = [
            SamplerPub.coerce(pub, shots=default_shots) for pub in pubs
        ]
        circuits = [pub.circuit for pub in coerced_pubs]
        if not circuits:
            raise ValueError("ManualDynamicDDSampler received no PUBs.")

        for index, circuit in enumerate(circuits):
            if self.physical_qubits and max(self.physical_qubits) >= circuit.num_qubits:
                raise RuntimeError(
                    f"MCM {self.policy} PUB {index} is not a full-width "
                    "physical circuit: selected physical qubits "
                    f"{self.physical_qubits}, circuit width={circuit.num_qubits}. "
                    "Route to the IBM backend before dynamic scheduling."
                )

        _assert_global_terminal_measurement_boundary(
            circuits, f"MCM {self.policy} input workload"
        )
        if self.policy == "none":
            self._calls += 1
            assert_terminal_measurements_are_final(
                circuits, "MCM no-DD plain ISA workload"
            )
            print(
                f"[Plain ISA dynamic submission; policy=none; "
                f"call={self._calls}] PUBs={len(circuits)}, "
                "validation=passed; custom scheduling bypassed",
                flush=True,
            )
            return self._sampler.run(coerced_pubs, **run_options)

        before_ops = [_recursive_count_ops(circuit) for circuit in circuits]
        feedback_qubits = _feedback_target_qubits(circuits)
        delayed = self._schedule_pm.run(circuits)
        delayed = list(delayed) if isinstance(delayed, list) else [delayed]

        padded = []
        insertion_stats = []
        for circuit in delayed:
            rewritten, stats = _insert_control_flow_safe_dd(
                circuit,
                self._backend,
                self.physical_qubits,
                self._pulse_alignment,
                self.policy,
            )
            padded.append(rewritten)
            insertion_stats.append(stats)
        padded_ops = [_recursive_count_ops(circuit) for circuit in padded]

        if self._translate_pm is not None:
            translated = self._translate_pm.run(padded)
            scheduled = (
                list(translated) if isinstance(translated, list)
                else [translated]
            )
        else:
            scheduled = padded
        final_ops = [_recursive_count_ops(circuit) for circuit in scheduled]

        # These checks run before any job is submitted.
        _assert_global_terminal_measurement_boundary(
            scheduled, f"MCM {self.policy} scheduled workload"
        )
        assert_terminal_measurements_are_final(
            scheduled, f"MCM {self.policy} scheduled workload"
        )

        self._calls += 1
        audit_rows = []
        for index, (before, padded_count, final, stats) in enumerate(
                zip(before_ops, padded_ops, final_ops, insertion_stats)):
            _assert_dynamic_structure_preserved(
                before, final, self.policy, index
            )
            inserted_x = padded_count.get("x", 0) - before.get("x", 0)
            inserted_y = padded_count.get("y", 0) - before.get("y", 0)
            has_feedback = before.get("if_else", 0) > 0
            nested_x_before = _nested_operation_count(circuits[index], "x")
            nested_x_after = _nested_operation_count(scheduled[index], "x")
            nested_y_before = _nested_operation_count(circuits[index], "y")
            nested_y_after = _nested_operation_count(scheduled[index], "y")

            if nested_x_after != nested_x_before or nested_y_after != nested_y_before:
                raise RuntimeError(
                    f"MCM {self.policy} PUB {index} modified a control-flow "
                    "body. Submission stopped."
                )
            if self.policy != "none" and has_feedback and stats["dd_cycles"] <= 0:
                raise RuntimeError(
                    f"MCM {self.policy} PUB {index} contains feedback but "
                    "received no safe DD cycle. Submission stopped."
                )
            if self.policy == "XY4" and (
                    inserted_x <= 0 or inserted_x != inserted_y):
                raise RuntimeError(
                    f"MCM XY4 PUB {index} inserted X={inserted_x}, "
                    f"Y={inserted_y}; expected matched X/Y pairs."
                )
            if self.policy == "XX" and (
                    inserted_y != 0 or inserted_x <= 0 or inserted_x % 2 != 0):
                raise RuntimeError(
                    f"MCM XX PUB {index} inserted X={inserted_x}, "
                    f"Y={inserted_y}; expected a positive even X-only sequence."
                )
            if (
                self.policy == "XY4"
                and "y" not in self._backend.operation_names
                and final.get("y", 0) != before.get("y", 0)
            ):
                raise RuntimeError(
                    f"MCM XY4 PUB {index} retained a non-native Y gate after "
                    "basis translation. Submission stopped."
                )

            audit_rows.append({
                "backend": self._backend.name,
                "policy": self.policy,
                "sampler_call": self._calls,
                "pub_index": index,
                "shots": coerced_pubs[index].shots,
                "circuit_width": circuits[index].num_qubits,
                "pulse_alignment_dt": self._pulse_alignment,
                "input_measure_2": before.get("measure_2", 0),
                "final_measure_2": final.get("measure_2", 0),
                "input_if_else": before.get("if_else", 0),
                "final_if_else": final.get("if_else", 0),
                "dd_cycles": stats["dd_cycles"],
                "inserted_x_before_translation": inserted_x,
                "inserted_y_before_translation": inserted_y,
                "final_x_delta": final.get("x", 0) - before.get("x", 0),
                "final_y_delta": final.get("y", 0) - before.get("y", 0),
                "final_rz_delta": final.get("rz", 0) - before.get("rz", 0),
                "scheduled_delay_count": final.get("delay", 0),
                "feedback_qubits_with_bodies_untouched": json.dumps(
                    sorted(feedback_qubits)
                ),
                "skipped_delays_after_measure_2": stats[
                    "skipped_delays_after_measure_2"
                ],
                "skipped_delays_after_terminal_measure": stats[
                    "skipped_delays_after_terminal_measure"
                ],
                "nested_x_before": nested_x_before,
                "nested_x_after": nested_x_after,
                "nested_y_before": nested_y_before,
                "nested_y_after": nested_y_after,
                "control_flow_bodies_untouched": True,
                "global_terminal_measurement_boundary": "passed",
                "validation": "passed",
            })

        _append_mcm_schedule_audit(audit_rows)
        scheduled_pubs = [
            SamplerPub(
                circuit,
                parameter_values=pub.parameter_values,
                shots=pub.shots,
            )
            for circuit, pub in zip(scheduled, coerced_pubs)
        ]
        print(
            f"[Safe local MCM DD; policy={self.policy}; call={self._calls}] "
            f"PUBs={len(circuits)}, validation=passed",
            flush=True,
        )
        return self._sampler.run(scheduled_pubs, **run_options)

    def __getattr__(self, name):
        return getattr(self._sampler, name)


def configure_mcm_policy_sampler(backend, policy, physical_qubits):
    sampler = make_hardware_session_sampler(backend)
    # Server-side DD is incompatible with dynamic circuits. All padding has
    # already been inserted locally by the Runtime 0.43 dynamic passes.
    sampler.options.dynamical_decoupling.enable = False
    if SHOW_SCHEDULER_TIMING:
        sampler.options.experimental = {
            "execution": {"scheduler_timing": True}
        }
    if policy == "none":
        policy_description = (
            "plain ISA submission; custom ALAP/PadDelay bypassed"
        )
    else:
        policy_description = (
            "safe top-level pre-terminal DD insertion; "
            "control-flow bodies untouched"
        )
    print(
        f"MCM policy {policy}: {policy_description}; "
        "server built-in DD=False; scheduler timing="
        f"{SHOW_SCHEDULER_TIMING}",
        flush=True,
    )
    if policy == "XX":
        print(
            "XX safety mode: X-X is inserted only into eligible "
            "top-level delays before terminal measurement.",
            flush=True,
        )
    return ManualDynamicDDSampler(
        sampler, backend, policy, physical_qubits
    )


def sweep_channel_specs(full_layout):
    return {
        "A": {
            "source_layout": [full_layout[0], full_layout[2]],
            "after_indices": [0, 8],
            "after_m3_qubits": [
                full_layout[2], full_layout[4],
                full_layout[3], full_layout[5],
                full_layout[0], full_layout[8],
            ],
        },
        "B": {
            "source_layout": [full_layout[1], full_layout[3]],
            "after_indices": [1, 7],
            "after_m3_qubits": [
                full_layout[2], full_layout[4],
                full_layout[3], full_layout[5],
                full_layout[1], full_layout[7],
            ],
        },
    }


def append_mcm_sweep_record(
        records, trial, method, channel, phase, rho, direct_s, policy):
    fidelity, chsh_from_rho = metrics_from_density_matrix(rho)
    direct_lower = direct_s / (2 * np.sqrt(2))
    direct_upper = direct_s / (4 * np.sqrt(2)) + 0.5
    records.append({
        "measurement_variant": EXPERIMENT_VARIANT,
        "trial": int(trial),
        "domain": "Hardware",
        "method": method,
        "series_key": f"{policy}_{method}",
        "dd_policy": policy,
        "dd_enabled": policy != "none",
        "dd_sequence": policy if policy != "none" else "none",
        "dd_implementation": "manual_runtime_dynamic_pass",
        "dd_scheduling_method": "alap",
        "server_builtin_dd_enabled": False,
        "channel": channel,
        "phase": phase,
        "fidelity": fidelity,
        "chsh_S_direct": float(direct_s),
        "chsh_S_from_rho": chsh_from_rho,
        "equation9_lower_from_direct_S": direct_lower,
        "equation9_upper_from_direct_S": direct_upper,
        "equation9_direct_point_pass": bool(
            direct_lower <= fidelity <= direct_upper
        ),
    })


def run_mcm_policy(
        records, backend, sampler, terminal_mitigator, mixed_mitigator,
        full_layout, policy):
    source_base = build_source_epr_pair_circuit()
    transported_base = build_epr_half_teleportation_circuit(
        with_corrections=True, use_measure_2=True
    )
    specs = sweep_channel_specs(full_layout)

    for trial in range(1, N_TRIALS + 1):
        print(f"\n[MCM; {policy}] trial {trial}/{N_TRIALS}", flush=True)
        for channel, spec in specs.items():
            print(
                f"MCM, policy={policy}, channel={channel}, before",
                flush=True,
            )
            before_rhos = run_raw_and_m3_tomography(
                source_base, backend, sampler,
                spec["source_layout"], [0, 1], spec["source_layout"],
                terminal_mitigator, SHOTS,
                run_label=(
                    f"MCM policy={policy}, trial={trial}, "
                    f"channel={channel}, before tomography"
                ),
            )
            before_s = run_raw_and_m3_chsh(
                source_base, backend, sampler,
                spec["source_layout"], [0, 1], spec["source_layout"],
                terminal_mitigator, SHOTS,
                run_label=(
                    f"MCM policy={policy}, trial={trial}, "
                    f"channel={channel}, before CHSH"
                ),
            )
            for method in ("raw", "m3"):
                append_mcm_sweep_record(
                    records, trial, method, channel, "before",
                    before_rhos[method], before_s[method], policy,
                )

            print(
                f"MCM, policy={policy}, channel={channel}, after",
                flush=True,
            )
            after_rhos = run_raw_and_m3_tomography(
                transported_base, backend, sampler,
                full_layout, spec["after_indices"],
                spec["after_m3_qubits"], mixed_mitigator, SHOTS,
                run_label=(
                    f"MCM policy={policy}, trial={trial}, "
                    f"channel={channel}, after tomography"
                ),
            )
            after_s = run_raw_and_m3_chsh(
                transported_base, backend, sampler,
                full_layout, spec["after_indices"],
                spec["after_m3_qubits"], mixed_mitigator, SHOTS,
                run_label=(
                    f"MCM policy={policy}, trial={trial}, "
                    f"channel={channel}, after CHSH"
                ),
            )
            for method in ("raw", "m3"):
                append_mcm_sweep_record(
                    records, trial, method, channel, "after",
                    after_rhos[method], after_s[method], policy,
                )


def run_mcm_dd_sweep():
    """Run no-DD, XY4, and XX MCM policies on the same real backend."""
    global _TIMING_PLOTS_CREATED, MAX_TIMING_PLOTS
    service = make_runtime_service()
    backend = service.backend(BACKEND_NAME)
    validate_runtime_backend(
        backend, require_measure_2=True, wait_for_active=True
    )
    supported = set(getattr(backend, "supported_instructions", ()))
    if "measure_2" not in supported:
        raise RuntimeError(
            f"{BACKEND_NAME} does not advertise measure_2. Choose a backend "
            "with dynamic measure_2 support."
        )

    layout_proxy = build_epr_half_teleportation_circuit(
        with_corrections=True, use_measure_2=False
    )
    full_layout = select_balanced_physical_layout(layout_proxy, backend)
    print(f"Selected nine-qubit layout: {full_layout}")

    measured_logical = (2, 4, 3, 5)
    measured_qargs = {(full_layout[index],) for index in measured_logical}
    try:
        supported_qargs = set(
            backend.target.qargs_for_operation_name("measure_2")
        )
    except AttributeError:
        supported_qargs = set(backend.target["measure_2"])
    missing = measured_qargs.difference(supported_qargs)
    if missing:
        raise RuntimeError(
            "The selected layout lacks measure_2 on physical qubits "
            f"{sorted(qargs[0] for qargs in missing)}."
        )
    syndrome_physical = [qargs[0] for qargs in sorted(measured_qargs)]

    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    # Every invocation starts a fresh acquisition; prior JSON is ignored.
    records = []
    for policy in DD_POLICIES:
        _TIMING_PLOTS_CREATED = 0
        MAX_TIMING_PLOTS = TIMING_PLOTS_PER_POLICY
        sampler = configure_mcm_policy_sampler(
            backend, policy, full_layout
        )
        terminal_mitigator = calibrate_m3(
            backend, sampler, full_layout,
            shots=M3_CALIBRATION_SHOTS,
            label_suffix=f"MCM policy={policy}, terminal",
        )
        mixed_mitigator = calibrate_m3(
            backend, sampler, full_layout,
            shots=M3_CALIBRATION_SHOTS,
            mid_circuit_qubits=syndrome_physical,
            label_suffix=f"MCM policy={policy}, mixed measure_2",
        )
        run_mcm_policy(
            records, backend, sampler, terminal_mitigator,
            mixed_mitigator, full_layout, policy,
        )
    return pd.DataFrame(records)


def summarize_mcm_sweep(records, channel, metric):
    subset = records[records["channel"] == channel].copy()
    paired = subset.pivot_table(
        index=["series_key", "trial"], columns="phase",
        values=metric, aggfunc="first",
    ).dropna(subset=["before", "after"])
    delta_stats = {}
    for series_key, group in paired.groupby(level="series_key"):
        delta_stats[series_key] = mean_ci99(
            group["after"].to_numpy(float)
            - group["before"].to_numpy(float)
        )

    rows = []
    for (series_key, phase), group in subset.groupby(
            ["series_key", "phase"], sort=False):
        mean, ci = mean_ci99(group[metric])
        delta_mean, delta_ci = delta_stats[series_key]
        rows.append({
            "series_key": series_key,
            "series_label": SWEEP_GROUP_LABELS[series_key],
            "domain": "Hardware",
            "method": str(group["method"].iloc[0]),
            "dd_policy": str(group["dd_policy"].iloc[0]),
            "dd_enabled": bool(group["dd_enabled"].iloc[0]),
            "dd_implementation": "manual_runtime_dynamic_pass",
            "server_builtin_dd_enabled": False,
            "channel": channel,
            "phase": phase,
            "metric_key": metric,
            "mean": mean,
            "ci99_halfwidth": ci,
            "paired_after_minus_before_mean": delta_mean,
            "paired_after_minus_before_ci99_halfwidth": delta_ci,
            "n_trials": len(group),
        })
    summary = pd.DataFrame(rows)
    summary["series_order"] = summary["series_key"].map(
        {key: index for index, key in enumerate(SWEEP_GROUP_ORDER)}
    )
    summary["phase_order"] = summary["phase"].map({"before": 0, "after": 1})
    return (
        summary.sort_values(["series_order", "phase_order"])
        .drop(columns=["series_order", "phase_order"])
        .reset_index(drop=True)
    )


def plot_mcm_dd_sweep(summary, channel, metric, save_path):
    order = [
        key for key in SWEEP_GROUP_ORDER
        if key in set(summary["series_key"])
    ]
    x = np.arange(len(order), dtype=float)
    width = 0.36
    fig, ax = plt.subplots(figsize=(13.5, 6.8))
    colors = {"before": "#A6CEE3", "after": "#1F78B4"}
    labels = {"before": "Before (source pair)", "after": "After (MCM teleported)"}

    for offset, phase in [(-width / 2, "before"), (width / 2, "after")]:
        phase_rows = (
            summary[summary["phase"] == phase]
            .set_index("series_key").loc[order]
        )
        ax.bar(
            x + offset, phase_rows["mean"].to_numpy(float), width,
            yerr=phase_rows["ci99_halfwidth"].to_numpy(float), capsize=4,
            color=colors[phase], edgecolor="black", linewidth=0.7,
            label=labels[phase],
        )

    ax.set_xticks(x)
    ax.set_xticklabels([SWEEP_GROUP_LABELS[key] for key in order])
    ax.grid(axis="y", alpha=0.25)
    ax.set_xlabel("Manual dynamic-DD policy and M3 treatment")
    pair = "(ref_A, q4)" if channel == "A" else "(ref_B, q3)"
    if metric == "fidelity":
        ax.axhline(1.0, color="gray", linestyle=":", label="Ideal")
        ax.set_ylim(0, 1.12)
        ax.set_ylabel("Entanglement fidelity to |Phi+>")
        title_metric = "entanglement fidelity"
    else:
        ax.axhline(2.0, color="#555555", linestyle="--", label="Classical bound")
        ax.axhline(
            2 * np.sqrt(2), color="gray", linestyle=":",
            label="Tsirelson bound",
        )
        ax.set_ylim(0, 3.08)
        ax.set_ylabel("Direct CHSH S")
        title_metric = "direct CHSH"
    ax.set_title(
        f"MCM measure_2 DD sweep: {title_metric}, channel {channel} {pair}\n"
        f"No DD vs XY4 vs XX; real hardware; {N_TRIALS} trials; 99% CI"
    )
    ax.legend(loc="best", fontsize=9)
    fig.tight_layout()
    fig.savefig(save_path, dpi=180, bbox_inches="tight")
    plt.show()


def save_mcm_dd_sweep_outputs(records):
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    outputs = {}
    for channel in ("A", "B"):
        for metric, short in [
            ("fidelity", "fidelity"),
            ("chsh_S_direct", "chsh"),
        ]:
            summary = summarize_mcm_sweep(records, channel, metric)
            csv_path = OUTPUT_DIR / f"mcm_dd_sweep_{short}_channel_{channel}.csv"
            png_path = OUTPUT_DIR / f"mcm_dd_sweep_{short}_channel_{channel}.png"
            summary.to_csv(csv_path, index=False)
            plot_mcm_dd_sweep(summary, channel, metric, png_path)
            outputs[(channel, short)] = {"csv": csv_path, "png": png_path}
    return outputs










def summarize_mcm_cascade(records, channel, metric):
    subset = records[records["channel"] == channel].copy()
    paired = subset.pivot_table(
        index=["series_key", "trial"], columns="round",
        values=metric, aggfunc="first",
    ).dropna(subset=[1, 2])
    delta_stats = {
        key: mean_ci99(group[2].to_numpy(float) - group[1].to_numpy(float))
        for key, group in paired.groupby(level="series_key")
    }
    round_stats = {}
    for (series_key, round_number), group in subset.groupby(
            ["series_key", "round"], sort=False):
        round_stats[(series_key, int(round_number))] = mean_ci99(group[metric])

    rows = []
    for (series_key, round_number), group in subset.groupby(
            ["series_key", "round"], sort=False):
        round_number = int(round_number)
        mean, ci = round_stats[(series_key, round_number)]
        r1_mean, r1_ci = round_stats[(series_key, 1)]
        r2_mean, r2_ci = round_stats[(series_key, 2)]
        delta_mean, delta_ci = delta_stats[series_key]
        rows.append({
            "series_key": series_key,
            "series_label": SWEEP_GROUP_LABELS[series_key],
            "domain": "Hardware",
            "method": str(group["method"].iloc[0]),
            "dd_policy": str(group["dd_policy"].iloc[0]),
            "dd_enabled": bool(group["dd_enabled"].iloc[0]),
            "channel": channel,
            "round": round_number,
            "state_role": (
                "round_1_output_and_round_2_input"
                if round_number == 1 else "round_2_output"
            ),
            "round_1_output_is_round_2_input": True,
        "round_2_output_is_round_3_input": True,
            "metric_key": metric,
            "mean": mean,
            "ci99_halfwidth": ci,
            "round_2_input_mean_from_round_1_output": r1_mean,
            "round_2_input_ci99_from_round_1_output": r1_ci,
            "round_2_output_mean": r2_mean,
            "round_2_output_ci99_halfwidth": r2_ci,
            "round_2_output_minus_input_mean": delta_mean,
            "round_2_output_minus_input_ci99_halfwidth": delta_ci,
            "n_trials": len(group),
        })
    summary = pd.DataFrame(rows)
    summary["series_order"] = summary["series_key"].map(
        {key: index for index, key in enumerate(SWEEP_GROUP_ORDER)}
    )
    return (
        summary.sort_values(["series_order", "round"])
        .drop(columns=["series_order"])
        .reset_index(drop=True)
    )


def plot_mcm_cascade(summary, channel, metric, save_path):
    order = [key for key in SWEEP_GROUP_ORDER if key in set(summary["series_key"])]
    x = np.arange(len(order), dtype=float)
    width = 0.36
    fig, ax = plt.subplots(figsize=(13.8, 6.9))
    colors = {1: "#80B1D3", 2: "#08519C"}
    labels = {
        1: "Round 1 output (= Round 2 input)",
        2: "Round 2 output",
        3: "Round 3 output",
    }
    for offset, round_number in [(-width / 2, 1), (width / 2, 2)]:
        rows = summary[summary["round"] == round_number].set_index("series_key").loc[order]
        ax.bar(
            x + offset, rows["mean"].to_numpy(float), width,
            yerr=rows["ci99_halfwidth"].to_numpy(float), capsize=4,
            color=colors[round_number], edgecolor="black", linewidth=0.7,
            label=labels[round_number],
        )
    ax.set_xticks(x)
    ax.set_xticklabels([SWEEP_GROUP_LABELS[key] for key in order])
    ax.grid(axis="y", alpha=0.25)
    ax.set_xlabel("Manual dynamic-DD policy and M3 treatment")
    pair = "channel A" if channel == "A" else "channel B"
    if metric == "fidelity":
        ax.axhline(1.0, color="gray", linestyle=":", label="Ideal")
        ax.set_ylim(0, 1.12)
        ax.set_ylabel("Entanglement fidelity to |Phi+>")
        title_metric = "entanglement fidelity"
    else:
        ax.axhline(2.0, color="#555555", linestyle="--", label="Classical bound")
        ax.axhline(2 * np.sqrt(2), color="gray", linestyle=":", label="Tsirelson bound")
        ax.set_ylim(0, 3.08)
        ax.set_ylabel("Direct CHSH S")
        title_metric = "direct CHSH"
    ax.set_title(
        f"MCM two-round parallel teleportation: {title_metric}, {pair}\n"
        f"Round 1 output feeds round 2; {N_TRIALS} trials; 99% CI"
    )
    ax.legend(loc="best", fontsize=9)
    fig.tight_layout()
    fig.savefig(save_path, dpi=180, bbox_inches="tight")
    plt.show()


```

## Corrected CHSH register order, Round 0 acquisition, and Round 0/1/2/3 hardware trajectory plots


```python
# Round-0 DD is locally scheduled and audited in both schemes.
# No extra idle time is added solely to force a DD cycle.
import copy
from uuid import uuid4
from qiskit import qpy
from qiskit.primitives.containers import SamplerPub
from qiskit.circuit.library import YGate
from qiskit.circuit.equivalence_library import EquivalenceLibrary, SessionEquivalenceLibrary as sel
from qiskit.transpiler import PassManager
from qiskit.transpiler.passes import BasisTranslator
from qiskit_ibm_runtime.transpiler.passes.scheduling import ALAPScheduleAnalysis, PadDelay

def _round0_pulse_alignment(backend, target):
    """Resolve pulse alignment from the IBM backend, then the target."""
    try:
        constraints = backend.configuration().timing_constraints or {}
    except Exception:
        constraints = {}
    return int(
        constraints.get("pulse_alignment")
        or getattr(target, "pulse_alignment", None)
        or 16
    )

def _round0_scheduling_pipeline(backend, policy, physical_qubits):
    """Build the safe local scheduling stage used by MCM DD policies.

    The complete dynamic circuit is first scheduled with ALAP and PadDelay,
    with end-of-circuit padding disabled.  DD gates are inserted afterward by
    ``_round0_insert_dd`` only into top-level delays that precede
    every ordinary terminal measurement.  This avoids PadDynamicalDecoupling's
    block-end padding, which previously placed XY4 gates after tomography
    measurements and caused IBM 9699 failures.
    """
    if policy not in ("none", "XY4", "XX"):
        raise ValueError(f"Unknown DD policy {policy!r}; choose none, XY4 or XX.")

    target = copy.deepcopy(backend.target)
    native_basis = list(dict.fromkeys([
        *backend.operation_names,
        "if_else", "while_loop", "for_loop", "switch_case",
    ]))
    pulse_alignment = _round0_pulse_alignment(backend, target)
    translate_pm = None

    # Y is used only as a temporary logical marker.  Scheduling is based on
    # the calibrated X duration, and the marker is translated after insertion
    # to X followed by the zero-duration virtual frame change RZ(pi).
    if policy == "XY4" and "y" not in target.operation_names:
        if "x" not in target.operation_names:
            raise RuntimeError("Backend target has no calibrated X gate.")
        xy4_equivalences = EquivalenceLibrary(base=sel)
        y_native = QuantumCircuit(1, name="native_y_for_xy4")
        y_native.x(0)
        y_native.rz(np.pi, 0)
        xy4_equivalences.add_equivalence(YGate(), y_native)
        translate_pm = PassManager([
            BasisTranslator(xy4_equivalences, native_basis)
        ])

    schedule_pm = PassManager([
        ALAPScheduleAnalysis(target=target),
        PadDelay(
            target=target,
            fill_very_end=False,
            schedule_idle_qubits=False,
        ),
    ])
    return schedule_pm, translate_pm, pulse_alignment

def _round0_insert_dd(
        circuit, backend, allowed_qubits, pulse_alignment, policy):
    """Replace eligible top-level delays with an XX or XY4 identity cycle.

    Control-flow bodies are copied without modification.  DD is never inserted
    after ``measure_2`` on the measured wire, after an ordinary terminal
    measurement anywhere, or into end-of-circuit padding.  Every replacement
    has exactly the same duration as the delay it replaces.
    """
    if policy == "XX":
        sequence = ("x", "x")
        spacing = np.asarray([0.25, 0.50, 0.25], dtype=float)
    elif policy == "XY4":
        sequence = ("x", "y", "x", "y")
        spacing = np.asarray([0.125, 0.25, 0.25, 0.25, 0.125], dtype=float)
    else:
        raise ValueError(f"Manual DD insertion is unavailable for {policy!r}.")

    allowed = set(map(int, allowed_qubits))
    alignment = max(1, int(pulse_alignment))
    durations = backend.target.durations()
    rewritten = circuit.copy_empty_like()
    measured_qubits = set()
    terminal_started = False
    cycles = 0
    skipped_after_measure_2 = 0
    skipped_after_terminal_measure = 0

    for item in circuit.data:
        qindices = [circuit.find_bit(q).index for q in item.qubits]
        cindices = [circuit.find_bit(c).index for c in item.clbits]
        new_qargs = [rewritten.qubits[index] for index in qindices]
        new_cargs = [rewritten.clbits[index] for index in cindices]
        operation = item.operation
        name = operation.name

        if name == "measure":
            terminal_started = True

        eligible_delay = (
            name == "delay"
            and len(qindices) == 1
            and qindices[0] in allowed
            and operation.unit == "dt"
        )
        if eligible_delay and terminal_started:
            skipped_after_terminal_measure += 1
        if eligible_delay and qindices[0] in measured_qubits:
            skipped_after_measure_2 += 1

        if (
            not eligible_delay
            or terminal_started
            or qindices[0] in measured_qubits
        ):
            rewritten.append(operation, new_qargs, new_cargs)
        else:
            qubit = qindices[0]
            interval = int(operation.duration)
            try:
                x_duration = int(round(
                    durations.get("x", qubit, unit="dt")
                ))
            except Exception:
                rewritten.append(operation, new_qargs, new_cargs)
                x_duration = 0

            sequence_duration = len(sequence) * x_duration
            if (
                x_duration <= 0
                or x_duration % alignment != 0
                or interval / sequence_duration <= 2.0
            ):
                if x_duration > 0:
                    rewritten.append(operation, new_qargs, new_cargs)
            else:
                slack = interval - sequence_duration
                taus = (
                    alignment * np.floor(slack * spacing / alignment)
                ).astype(int)
                extra = int(slack - int(taus.sum()))
                middle = (len(taus) - 1) // 2
                to_middle = alignment * (extra // alignment)
                taus[middle] += to_middle
                taus[-1] += extra - to_middle
                if int(taus.sum()) + sequence_duration != interval:
                    raise RuntimeError(
                        f"Control-flow-safe {policy} changed delay "
                        f"duration {interval}."
                    )

                for gate_index, gate_name in enumerate(sequence):
                    if taus[gate_index] > 0:
                        rewritten.delay(
                            int(taus[gate_index]), new_qargs[0], unit="dt"
                        )
                    if gate_name == "x":
                        rewritten.x(new_qargs[0])
                    else:
                        rewritten.y(new_qargs[0])
                if taus[-1] > 0:
                    rewritten.delay(int(taus[-1]), new_qargs[0], unit="dt")
                cycles += 1

        if name in {"measure", "measure_2"}:
            measured_qubits.update(qindices)
        elif name == "reset":
            measured_qubits.difference_update(qindices)

    # XYXY = -I.  Restore exact circuit equivalence, including global phase.
    if policy == "XY4" and cycles % 2:
        rewritten.global_phase += np.pi

    return rewritten, {
        "dd_cycles": cycles,
        "skipped_delays_after_measure_2": skipped_after_measure_2,
        "skipped_delays_after_terminal_measure": skipped_after_terminal_measure,
    }

class Round0AuditedSampler:
    """Inspect and submit exactly the audited circuits; server DD stays off."""
    def __init__(self, backend, policy, physical_qubits, label):
        self._backend = backend
        self._sampler = make_hardware_session_sampler(backend)
        self._sampler.options.dynamical_decoupling.enable = False
        self.options = self._sampler.options
        self.policy = policy
        self.physical_qubits = list(physical_qubits)
        self.label = label
        self.audit_id = uuid4().hex[:12]
        self.rows = []
        self._calls = 0
        self._schedule, self._translate, self._alignment = _round0_scheduling_pipeline(backend, policy, physical_qubits)

    def run(self, pubs, **run_options):
        shots = run_options.pop('shots', None)
        pubs = [SamplerPub.coerce(pub, shots=shots) for pub in pubs]
        self._calls += 1
        scheduled_pubs = []
        OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        for index, pub in enumerate(pubs):
            before = pub.circuit
            delayed = self._schedule.run(before)
            padded, stats = (delayed, {'dd_cycles': 0}) if self.policy == 'none' else _round0_insert_dd(
                delayed, self._backend, self.physical_qubits, self._alignment, self.policy)
            before_ops, after_ops = before.count_ops(), padded.count_ops()
            x_added = int(after_ops.get('x', 0) - before_ops.get('x', 0))
            y_added = int(after_ops.get('y', 0) - before_ops.get('y', 0))
            cycles = int(stats['dd_cycles'])
            expected = {'none': (0, 0), 'XX': (2*cycles, 0), 'XY4': (2*cycles, 2*cycles)}[self.policy]
            if (x_added, y_added) != expected:
                raise RuntimeError(f'Round-0 DD pulse audit failed: {(x_added, y_added)} != {expected}')
            final = self._translate.run(padded) if self._translate is not None else padded
            _validate_physical_readouts(final, self._backend)
            assert_terminal_measurements_are_final(final, 'Round-0 audited schedule')
            stem = f'r0_{self.audit_id}_{self._calls}_{index}'
            qpy_path = OUTPUT_DIR / (stem + '.qpy')
            with qpy_path.open('wb') as stream:
                qpy.dump(final, stream)
            status = 'disabled' if self.policy == 'none' else ('inserted' if cycles else 'no_eligible_idle_window')
            row = dict(audit_id=self.audit_id, label=self.label, scheme=CASCADE_SCHEME,
                       round=0, dd_policy=self.policy, sampler_call=self._calls, pub_index=index,
                       shots=pub.shots, physical_qubits=json.dumps(self.physical_qubits),
                       dd_implementation='local_round0_safe_dd', server_builtin_dd_enabled=False,
                       dd_cycles=cycles, inserted_x_before_translation=x_added,
                       inserted_y_before_translation=y_added, dd_pulses_inserted=bool(cycles),
                       dd_status=status, scheduled_circuit_qpy=qpy_path.name)
            self.rows.append(row)
            print(f'[{self.label}] PUB {index}: DD={self.policy}, cycles={cycles}, X={x_added}, Y={y_added}, {status}', flush=True)
            scheduled_pubs.append(SamplerPub(final, parameter_values=pub.parameter_values, shots=pub.shots))
        pd.DataFrame(self.rows).to_csv(OUTPUT_DIR / f'r0_{self.audit_id}_dd_audit.csv', index=False)
        return self._sampler.run(scheduled_pubs, **run_options)

    def __getattr__(self, name):
        return getattr(self._sampler, name)


def _attach_round0_audit(record, source):
    record.update(source.get('dd_audit', {}))
    record['round0_acquisition'] = 'independent_per_policy_and_trial'


# ---------------- Authoritative round-0 / CHSH / plotting corrections ----------------

# Panel order: hardware left, matching simulation right; raw above M3.
TRAJECTORY_CONDITIONS = ["hw_raw", "sim_raw", "hw_m3", "sim_m3"]
TRAJECTORY_CONDITION_LABELS = {
    "hw_raw": "Real hardware — raw",
    "hw_m3": "Real hardware + M3",
    "sim_raw": "Noisy Aer — raw (no DD)",
    "sim_m3": "Noisy Aer + M3 (no DD)",
}
TRAJECTORY_POLICY_ORDER = ["none", "XY4", "XX"]
TRAJECTORY_POLICY_LABELS = {"none": "No DD", "XY4": "XY4", "XX": "XX"}
TRAJECTORY_POLICY_COLORS = {"none": "#9ECAE1", "XY4": "#4292C6", "XX": "#08519C"}


# Abstract logical pair followed by its indices in the 14-qubit cascade.


def trajectory_state_pair_spec(full_layout, channel, round_number):
    """Return an auditable identity for the state measured by both analyses."""
    channel = str(channel)
    round_number = int(round_number)
    if channel not in TRAJECTORY_STATE_PAIRS:
        raise ValueError(f"Unknown channel {channel!r}.")
    left, right, full_indices, label = TRAJECTORY_STATE_PAIRS[channel][round_number]
    physical = [int(full_layout[index]) for index in full_indices]
    analysis_indices = [0, 1] if round_number == 0 else list(full_indices)
    return {
        "state_pair_label": label,
        "state_qubit_left_role": left,
        "state_qubit_right_role": right,
        "full_circuit_logical_indices": list(full_indices),
        "analysis_circuit_logical_indices": analysis_indices,
        "state_physical_qubits": physical,
    }


def _state_pair_record_fields(state_pair):
    if state_pair is None:
        raise ValueError("Every fidelity/CHSH record requires state-pair metadata.")
    return {
        "state_pair_label": state_pair["state_pair_label"],
        "state_qubit_left_role": state_pair["state_qubit_left_role"],
        "state_qubit_right_role": state_pair["state_qubit_right_role"],
        "state_full_logical_indices": json.dumps(state_pair["full_circuit_logical_indices"]),
        "state_analysis_logical_indices": json.dumps(state_pair["analysis_circuit_logical_indices"]),
        "state_physical_qubits": json.dumps(state_pair["state_physical_qubits"]),
        "tomography_chsh_same_state_pair": True,
    }


def print_trajectory_state_pair_audit(full_layout):
    rows = []
    for channel in ("A", "B"):
        for round_number in (0, 1, 2, 3):
            spec = trajectory_state_pair_spec(full_layout, channel, round_number)
            rows.append({
                "channel": channel,
                "round": round_number,
                "two_qubit_state": spec["state_pair_label"],
                "cascade_logical_indices": spec["full_circuit_logical_indices"],
                "physical_qubits": spec["state_physical_qubits"],
            })
    audit = pd.DataFrame(rows)
    print("Tomography and direct CHSH use these identical two-qubit pairs:")
    print(audit.to_string(index=False))
    return audit





def _run_source_pair_measurements(
        backend, sampler, mitigator, full_layout, channel, label, policy="none"):
    """Measure Round 0 per policy and derive CHSH from the same rho as fidelity."""
    global SHOW_SCHEDULER_TIMING
    state_pair = trajectory_state_pair_spec(full_layout, channel, 0)
    source_layout = state_pair["state_physical_qubits"]
    source_circuit = build_source_epr_pair_circuit()
    audited_sampler = None
    if not _readout_is_simulator(backend):
        audited_sampler = Round0AuditedSampler(backend, policy, source_layout, label)
        sampler = audited_sampler
    show_timing_before = SHOW_SCHEDULER_TIMING
    SHOW_SCHEDULER_TIMING = False
    try:
        rhos = run_raw_and_m3_tomography(
            source_circuit, backend, sampler, source_layout, [0, 1],
            source_layout, mitigator, SHOTS, run_label=label + ", tomography",
        )
        # Round 0 is the input state before parallel teleportation.  Use
        # Tr[rho sqrt(2)(XX+ZZ)] from exactly the same reconstructed rho
        # as the fidelity instead of submitting four additional CHSH PUBs.
        # A physical two-qubit rho cannot exceed the Tsirelson bound.
        chsh = {
            method: metrics_from_density_matrix(rhos[method])[1]
            for method in ("raw", "m3")
        }
        audit = {'round0_dd_status': 'simulation_no_dd', 'round0_dd_pulses_inserted': False}
        if audited_sampler is not None:
            rows = audited_sampler.rows
            if not rows:
                raise RuntimeError('Round-0 tomography bypassed its scheduled-circuit audit.')
            inserted = sum(row['dd_cycles'] for row in rows)
            audit = dict(round0_dd_audit_csv=f'r0_{audited_sampler.audit_id}_dd_audit.csv',
                         round0_dd_cycles=inserted, round0_dd_pulses_inserted=bool(inserted),
                         round0_dd_status=('disabled' if policy == 'none' else
                                           'inserted' if inserted else 'no_eligible_idle_window'),
                         dd_implementation='local_round0_safe_dd', dd_scheduling_method='alap',
                         server_builtin_dd_enabled=False)
        return {"rhos": rhos, "chsh": chsh, "state_pair": state_pair, "dd_audit": audit}
    finally:
        SHOW_SCHEDULER_TIMING = show_timing_before



def _check_direct_chsh(value, label):
    value = float(value)
    if not np.isfinite(value) or abs(value) > 4.0 + 1e-9:
        raise RuntimeError(f"Invalid direct CHSH value {value!r} for {label}.")
    if abs(abs(value) - 4.0) <= 1e-10:
        raise RuntimeError(
            f"Direct CHSH reached the exact algebraic limit for {label}. "
            "This is treated as a register/M3 mapping failure, not as a "
            "physical result. Inspect the saved pair and bit-order audit."
        )
    return value



def append_mcm_cascade_record(
        records, trial, method, channel, round_number, rho, direct_s,
        policy, domain="Hardware", input_rho=None, input_state_source="",
        state_pair=None, input_state_available=True):
    physical_rho, _, _ = _sanitize_two_qubit_density_matrix(rho)
    fidelity, chsh_from_rho = metrics_from_density_matrix(physical_rho)
    direct_s = _check_direct_chsh(
        direct_s, f"{domain}, MCM, {policy}, {channel}, round {round_number}, {method}"
    )
    direct_abs = abs(direct_s)
    is_hardware = domain == "Hardware"
    record = {
        "measurement_variant": EXPERIMENT_VARIANT,
        "trial": int(trial), "domain": domain, "method": method,
        "measurement_mitigation_engine": MTHREE_IMPLEMENTATION if method == "m3" else "none",
        "series_key": f"{policy}_{method}" if is_hardware else f"sim_{method}",
        "dd_policy": policy,
        "dd_enabled": bool(is_hardware and policy != "none"),
        "dd_sequence": policy if is_hardware and policy != "none" else "none",
        "dd_implementation": "manual_runtime_dynamic_pass" if is_hardware else "no_dd_aer_reference",
        "dd_scheduling_method": "alap" if is_hardware else "none",
        "server_builtin_dd_enabled": False,
        "channel": channel, "round": int(round_number),
        "state_role": _state_role(round_number),
        "round_1_output_is_round_2_input": True,
        "round_2_output_is_round_3_input": True,
        "fidelity": fidelity, "chsh_S_direct": direct_s,
        "chsh_S_from_rho": chsh_from_rho,
        "chsh_exceeds_tsirelson": bool(direct_abs > 2 * np.sqrt(2)),
        "equation9_lower_from_direct_S": direct_abs / (2 * np.sqrt(2)),
        "equation9_upper_from_direct_S": direct_abs / (4 * np.sqrt(2)) + 0.5,
    }
    record["equation9_direct_point_pass"] = bool(
        record["equation9_lower_from_direct_S"] <= fidelity
        <= record["equation9_upper_from_direct_S"]
    )
    record.update(_state_pair_record_fields(state_pair))
    record.update(_rho_payload(physical_rho))
    record["input_state_available"] = bool(input_state_available)
    if input_state_available:
        if input_rho is None:
            input_rho = physical_rho
            input_state_source = input_state_source or "same_measured_round_0_state"
        input_payload = _rho_payload(input_rho)
        record.update({
            "input_rho_real": input_payload["rho_real"],
            "input_rho_imag": input_payload["rho_imag"],
            "input_rho_psd_projection_frobenius": input_payload["rho_psd_projection_frobenius"],
            "input_rho_preprojection_min_eigenvalue": input_payload["rho_preprojection_min_eigenvalue"],
        })
    else:
        record.update({
            "input_rho_real": np.nan,
            "input_rho_imag": np.nan,
            "input_rho_psd_projection_frobenius": np.nan,
            "input_rho_preprojection_min_eigenvalue": np.nan,
        })
        input_state_source = input_state_source or "unavailable_previous_round"
    record["input_state_source"] = input_state_source
    records.append(record)



def append_dcm_cascade_record(
        records, trial, method, channel, round_number, rho, direct_s,
        policy, dd_enabled, dd_sequence, domain="Hardware", input_rho=None,
        input_state_source="", state_pair=None):
    physical_rho, _, _ = _sanitize_two_qubit_density_matrix(rho)
    fidelity, chsh_from_rho = metrics_from_density_matrix(physical_rho)
    direct_s = _check_direct_chsh(
        direct_s, f"{domain}, DCM, {policy}, {channel}, round {round_number}, {method}"
    )
    direct_abs = abs(direct_s)
    is_hardware = domain == "Hardware"
    active_dd = bool(is_hardware and dd_enabled)
    record = {
        "measurement_variant": EXPERIMENT_VARIANT,
        "trial": int(trial), "domain": domain, "method": method,
        "measurement_mitigation_engine": MTHREE_IMPLEMENTATION if method == "m3" else "none",
        "series_key": f"{policy}_{method}" if is_hardware else f"sim_{method}",
        "dd_policy": policy, "dd_enabled": active_dd,
        "dd_sequence": dd_sequence if active_dd else "none",
        "dd_implementation": "runtime_sampler_option" if is_hardware else "no_dd_aer_reference",
        "dd_scheduling_method": DD_SCHEDULING_METHOD if active_dd else "none",
        "channel": channel, "round": int(round_number),
        "state_role": _state_role(round_number),
        "round_1_output_is_round_2_input": True,
        "round_2_output_is_round_3_input": True,
        "fidelity": fidelity, "chsh_S_direct": direct_s,
        "chsh_S_from_rho": chsh_from_rho,
        "chsh_exceeds_tsirelson": bool(direct_abs > 2 * np.sqrt(2)),
        "equation9_lower_from_direct_S": direct_abs / (2 * np.sqrt(2)),
        "equation9_upper_from_direct_S": direct_abs / (4 * np.sqrt(2)) + 0.5,
    }
    record["equation9_direct_point_pass"] = bool(
        record["equation9_lower_from_direct_S"] <= fidelity
        <= record["equation9_upper_from_direct_S"]
    )
    record.update(_state_pair_record_fields(state_pair))
    record.update(_rho_payload(physical_rho))
    if input_rho is None:
        input_rho = physical_rho
        input_state_source = input_state_source or "same_measured_round_0_state"
    input_payload = _rho_payload(input_rho)
    record.update({
        "input_rho_real": input_payload["rho_real"],
        "input_rho_imag": input_payload["rho_imag"],
        "input_rho_psd_projection_frobenius": input_payload["rho_psd_projection_frobenius"],
        "input_rho_preprojection_min_eigenvalue": input_payload["rho_preprojection_min_eigenvalue"],
        "input_state_source": input_state_source,
    })
    records.append(record)



def run_mcm_cascade_policy(
        records, backend, sampler, round1_mitigator, round2_mitigator,
        full_layout, policy, domain="Hardware", use_measure_2=True,
        source_mitigator=None, source_cache=None, acquire_round0=True):
    round_circuits = {
        1: build_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=use_measure_2),
        2: build_two_round_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=use_measure_2),
        **({3: build_three_round_epr_half_teleportation_circuit(with_corrections=True, use_measure_2=use_measure_2)} if domain == "Hardware" else {}),
    }
    mitigators = {1: round1_mitigator, 2: round2_mitigator, 3: round2_mitigator}
    source_cache = {} if source_cache is None else source_cache
    if acquire_round0 and source_mitigator is None:
        raise ValueError("Round-0 acquisition requires its terminal-measurement M3 mitigator.")
    specs = cascade_channel_specs(full_layout)
    for trial in range(1, N_TRIALS + 1):
        print(f"\n[{domain}; MCM cascade; {policy}] trial {trial}/{N_TRIALS}")
        for channel in ("A", "B"):
            cache_key = (domain, policy, int(trial), channel)
            if acquire_round0:
                source = _run_source_pair_measurements(
                    backend, sampler, source_mitigator, full_layout, channel,
                    f"{domain}, MCM input, trial={trial}, channel={channel}, policy={policy}, round=0", policy=policy,
                )
                source_cache[cache_key] = source
                for method in ("raw", "m3"):
                    append_mcm_cascade_record(
                        records, trial, method, channel, 0,
                        source["rhos"][method], source["chsh"][method],
                        policy if domain == "Hardware" else "reference",
                        domain=domain, input_rho=source["rhos"][method],
                        input_state_source="measured_source_EPR_pair",
                        state_pair=source["state_pair"],
                    )
                    _attach_round0_audit(records[-1], source)
            elif cache_key not in source_cache:
                raise RuntimeError("Round-0 cache missing; acquire this policy before reusing its input state.")
            source = source_cache[cache_key]
            previous_output_rhos = None
            for round_number in ((1, 2, 3) if domain == "Hardware" else (1, 2)):
                spec = specs[channel][round_number]
                state_pair = trajectory_state_pair_spec(
                    full_layout, channel, round_number
                )
                if list(spec["indices"]) != state_pair["analysis_circuit_logical_indices"]:
                    raise RuntimeError(
                        "Tomography/CHSH logical-pair audit failed: "
                        f"{spec['indices']} != {state_pair['analysis_circuit_logical_indices']}."
                    )
                measured_physical = [
                    int(spec["layout"][index]) for index in spec["indices"]
                ]
                if measured_physical != state_pair["state_physical_qubits"]:
                    raise RuntimeError(
                        "Tomography/CHSH physical-pair audit failed: "
                        f"{measured_physical} != {state_pair['state_physical_qubits']}."
                    )
                label = f"{domain}, MCM, policy={policy}, trial={trial}, channel={channel}, round={round_number}"
                rhos = run_raw_and_m3_tomography(
                    round_circuits[round_number], backend, sampler, spec["layout"],
                    spec["indices"], spec["m3_qubits"], mitigators[round_number],
                    SHOTS, run_label=label + ", tomography",
                )
                chsh = run_raw_and_m3_chsh(
                    round_circuits[round_number], backend, sampler, spec["layout"],
                    spec["indices"], spec["m3_qubits"], mitigators[round_number],
                    SHOTS, run_label=label + ", CHSH",
                )
                for method in ("raw", "m3"):
                    input_rho = source["rhos"][method] if round_number == 1 else previous_output_rhos[method]
                    append_mcm_cascade_record(
                        records, trial, method, channel, round_number,
                        rhos[method], chsh[method], policy, domain=domain,
                        input_rho=input_rho,
                        input_state_source="measured_source_EPR_pair" if round_number == 1 else f"measured_round_{round_number - 1}_output",
                        state_pair=state_pair,
                    )
                previous_output_rhos = rhos



def run_dcm_cascade_policy(
        records, backend, sampler, mitigator, full_layout, policy,
        dd_enabled, dd_sequence, domain="Hardware", source_cache=None,
        acquire_round0=True):
    round_circuits = {
        1: build_epr_half_teleportation_circuit(with_corrections=True),
        2: build_two_round_epr_half_teleportation_circuit(with_corrections=True),
        **({3: build_three_round_epr_half_teleportation_circuit(with_corrections=True)} if domain == "Hardware" else {}),
    }
    source_cache = {} if source_cache is None else source_cache
    specs = cascade_channel_specs(full_layout)
    for trial in range(1, N_TRIALS + 1):
        print(f"\n[{domain}; DCM cascade; {policy}] trial {trial}/{N_TRIALS}")
        for channel in ("A", "B"):
            cache_key = (domain, policy, int(trial), channel)
            if acquire_round0:
                source = _run_source_pair_measurements(
                    backend, sampler, mitigator, full_layout, channel,
                    f"{domain}, DCM input, trial={trial}, channel={channel}, policy={policy}, round=0", policy=policy,
                )
                source_cache[cache_key] = source
                for method in ("raw", "m3"):
                    append_dcm_cascade_record(
                        records, trial, method, channel, 0,
                        source["rhos"][method], source["chsh"][method],
                        policy if domain == "Hardware" else "reference",
                        dd_enabled, dd_sequence, domain=domain,
                        input_rho=source["rhos"][method],
                        input_state_source="measured_source_EPR_pair",
                        state_pair=source["state_pair"],
                    )
                    _attach_round0_audit(records[-1], source)
            elif cache_key not in source_cache:
                raise RuntimeError("Round-0 cache missing; acquire this policy before reusing its input state.")
            source = source_cache[cache_key]
            previous_output_rhos = None
            for round_number in ((1, 2, 3) if domain == "Hardware" else (1, 2)):
                spec = specs[channel][round_number]
                state_pair = trajectory_state_pair_spec(
                    full_layout, channel, round_number
                )
                if list(spec["indices"]) != state_pair["analysis_circuit_logical_indices"]:
                    raise RuntimeError(
                        "Tomography/CHSH logical-pair audit failed: "
                        f"{spec['indices']} != {state_pair['analysis_circuit_logical_indices']}."
                    )
                measured_physical = [
                    int(spec["layout"][index]) for index in spec["indices"]
                ]
                if measured_physical != state_pair["state_physical_qubits"]:
                    raise RuntimeError(
                        "Tomography/CHSH physical-pair audit failed: "
                        f"{measured_physical} != {state_pair['state_physical_qubits']}."
                    )
                label = f"{domain}, DCM, policy={policy}, trial={trial}, channel={channel}, round={round_number}"
                rhos = run_raw_and_m3_tomography(
                    round_circuits[round_number], backend, sampler, spec["layout"],
                    spec["indices"], spec["m3_qubits"], mitigator, SHOTS,
                    run_label=label + ", tomography",
                )
                chsh = run_raw_and_m3_chsh(
                    round_circuits[round_number], backend, sampler, spec["layout"],
                    spec["indices"], spec["m3_qubits"], mitigator, SHOTS,
                    run_label=label + ", CHSH",
                )
                for method in ("raw", "m3"):
                    input_rho = source["rhos"][method] if round_number == 1 else previous_output_rhos[method]
                    append_dcm_cascade_record(
                        records, trial, method, channel, round_number,
                        rhos[method], chsh[method], policy, dd_enabled,
                        dd_sequence, domain=domain, input_rho=input_rho,
                        input_state_source="measured_source_EPR_pair" if round_number == 1 else f"measured_round_{round_number - 1}_output",
                        state_pair=state_pair,
                    )
                previous_output_rhos = rhos



def _summarize_round_trajectory(records, channel, metric):
    frame = records if isinstance(records, pd.DataFrame) else pd.DataFrame(records)
    subset = frame[frame["channel"] == channel].copy()
    rows = []
    for (domain, method, policy, round_number), group in subset.groupby(
            ["domain", "method", "dd_policy", "round"], sort=False):
        mean, ci = mean_ci99(group[metric])
        condition_key = ("hw_" if domain == "Hardware" else "sim_") + method
        pair_labels = group["state_pair_label"].dropna().unique().tolist()
        if len(pair_labels) != 1:
            raise RuntimeError(
                f"Expected one state-pair label for {condition_key}, {policy}, "
                f"round {round_number}; got {pair_labels}."
            )
        rows.append({
            "condition_key": condition_key,
            "condition_label": TRAJECTORY_CONDITION_LABELS[condition_key],
            "domain": domain, "method": method, "dd_policy": str(policy),
            "channel": channel, "round": int(round_number),
            "state_pair_label": pair_labels[0],
            "metric_key": metric, "mean": mean, "ci99_halfwidth": ci,
            "n_trials": len(group),
            "round0_acquisition": "independent_per_policy_and_trial" if int(round_number) == 0 else "",
            "round0_dd_status": ";".join(sorted(group["round0_dd_status"].dropna().unique()))
                if int(round_number) == 0 and "round0_dd_status" in group else "",
            "round0_dd_audit_csvs": ";".join(group["round0_dd_audit_csv"].dropna().astype(str))
                if int(round_number) == 0 and "round0_dd_audit_csv" in group else "",
            "simulator_dd_note": "No DD: backend-derived Aer is a gate/readout noise-model reference, not a pulse-level DD model.",
        })
    return pd.DataFrame(rows).sort_values(
        ["condition_key", "round", "dd_policy"]
    ).reset_index(drop=True)






# Final plotting override: one raw axis and one M3 axis.
def _plot_round_trajectory_performance(summary, scheme, channel, metric, save_path):
    """Put hardware and its matching simulation together in each subplot."""
    fig, axes = plt.subplots(2, 1, figsize=(15.5, 10.8), sharey=True)
    round_colors = {0: "#BDBDBD", 1: "#6BAED6", 2: "#2171B5", 3: "#08306B"}
    round_labels = {
        0: "Round 0 input (measured per DD policy)",
        1: "Round 1 output",
        2: "Round 2 output",
        3: "Round 3 output",
    }
    row_specs = [
        ("Raw", "hw_raw", "sim_raw"),
        ("M3-mitigated", "hw_m3", "sim_m3"),
    ]
    group_x = np.arange(4, dtype=float)
    group_labels = ["No DD\nReal HW", "XY4", "XX", "No DD\nSimulation (rounds 0-2)"]
    group_policies = ["none", "XY4", "XX", "reference"]
    bar_width = 0.18
    round_offsets = {0: -0.30, 1: -0.10, 2: 0.10, 3: 0.30}
    all_tops = []
    legend_handles = {}

    for axis, (row_title, hardware_key, simulation_key) in zip(axes, row_specs):
        for round_number in (0, 1, 2, 3):
            means, errors, positions = [], [], []
            for group_index, policy in enumerate(group_policies):
                is_simulation = group_index == len(group_policies) - 1
                if is_simulation and round_number == 3:
                    continue  # Simulation round 3 is intentionally not acquired.
                positions.append(group_x[group_index] + round_offsets[round_number])
                condition_key = simulation_key if is_simulation else hardware_key
                source_policy = (
                    "reference" if is_simulation
                    else policy
                )
                row = summary[
                    (summary["condition_key"] == condition_key)
                    & (summary["round"] == round_number)
                    & (summary["dd_policy"] == source_policy)
                ]
                if len(row) != 1:
                    raise RuntimeError(
                        f"Expected one summary row for {condition_key}, "
                        f"policy={source_policy}, round={round_number}; got {len(row)}."
                    )
                means.append(float(row.iloc[0]["mean"]))
                errors.append(float(row.iloc[0]["ci99_halfwidth"]))

            bars = axis.bar(
                positions, means, bar_width,
                yerr=errors, capsize=4, color=round_colors[round_number],
                edgecolor="black", linewidth=0.7,
                label=round_labels[round_number],
            )
            legend_handles.setdefault(round_labels[round_number], bars)
            for bar, value, error in zip(bars, means, errors):
                finite_error = error if np.isfinite(error) else 0.0
                top = value + finite_error
                axis.annotate(
                    f"{value:.3f}",
                    (bar.get_x() + bar.get_width() / 2.0, top),
                    xytext=(0, 4 + 6 * round_number),
                    textcoords="offset points", ha="center", va="bottom",
                    fontsize=7, fontweight="bold",
                    bbox={"boxstyle": "round,pad=0.12", "fc": "white",
                          "ec": "none", "alpha": 0.88},
                    annotation_clip=False,
                )
                all_tops.append(top)

        axis.set_title(row_title)
        axis.set_xticks(group_x, group_labels)
        axis.set_xlabel(
            "Hardware DD policy, followed by the no-DD simulation reference"
        )
        axis.grid(axis="y", alpha=0.25)
        if metric == "fidelity":
            axis.axhline(1.0, color="gray", linestyle=":", linewidth=1.2)
        else:
            axis.axhline(2.0, color="#555555", linestyle="--", linewidth=1.2)
            axis.axhline(
                2 * np.sqrt(2), color="gray", linestyle=":", linewidth=1.2
            )

    title_metric = (
        "entanglement fidelity to |Phi+>"
        if metric == "fidelity" else "CHSH S (input: tomography; outputs: direct)"
    )
    base_upper = 1.12 if metric == "fidelity" else 3.08
    padding = 0.30 if metric == "fidelity" else 0.55
    upper = max(base_upper, max(all_tops or [0.0]) + padding)
    for axis in axes:
        axis.set_ylim(0, upper)
        axis.set_ylabel(title_metric)

    fig.legend(
        list(legend_handles.values()), list(legend_handles),
        loc="upper center", ncol=4, bbox_to_anchor=(0.5, 0.945),
    )
    pair_text = " / ".join(
        TRAJECTORY_STATE_PAIRS[channel][round_number][3]
        for round_number in (0, 1, 2, 3)
    )
    fig.suptitle(
        f"{scheme} parallel teleportation - channel {channel}: "
        f"{title_metric}: HW rounds 0-3; simulation rounds 0-2\n"
        f"State pairs: {pair_text}; {N_TRIALS} trials; "
        "error bars are 99% CI",
        fontsize=13,
    )
    fig.tight_layout(rect=(0, 0, 1, 0.89))
    fig.savefig(save_path, dpi=180, bbox_inches="tight")
    plt.show()
    plt.close(fig)


def _save_round_trajectory_outputs(records, scheme):
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    frame = records if isinstance(records, pd.DataFrame) else pd.DataFrame(records)
    outputs = {}
    prefix = f'{scheme.lower()}_rank{DISENTANGLER_RANK}'
    all_records_csv = OUTPUT_DIR / f"{prefix}_HW0123_SIM012_records.csv"
    frame.to_csv(all_records_csv, index=False)
    outputs["all_records_csv"] = all_records_csv
    for channel in ("A", "B"):
        for metric, short in (("fidelity", "fidelity"), ("chsh_S_direct", "chsh")):
            summary = _summarize_round_trajectory(frame, channel, metric)
            metric_tag = "fid" if short == "fidelity" else "chsh"
            csv_path = OUTPUT_DIR / f"{prefix}_HW0123_SIM012_{metric_tag}_{channel}.csv"
            png_path = OUTPUT_DIR / f"{prefix}_HW0123_SIM012_{metric_tag}_{channel}.png"
            summary.to_csv(csv_path, index=False)
            _plot_round_trajectory_performance(summary, scheme, channel, metric, png_path)
            outputs[(channel, short)] = {"csv": csv_path, "png": png_path}
    return outputs



def save_mcm_two_round_outputs(records):
    return _save_round_trajectory_outputs(records, "MCM")


def save_dcm_two_round_outputs(records):
    return _save_round_trajectory_outputs(records, "DCM")







def _trajectory_subset(frame, condition_key, round_number, channel, policy_by_round):
    policy = str(policy_by_round[int(round_number)])
    mask = _condition_mask(frame, policy, condition_key)
    return frame[mask & (frame["round"] == int(round_number)) & (frame["channel"] == channel)]










def _trajectory_fidelity_matrices(records, condition_key, policy_by_round_by_channel):
    frame = records if isinstance(records, pd.DataFrame) else pd.DataFrame(records)
    rounds = (0, 1, 2, 3) if condition_key.startswith('hw_') else (0, 1, 2)
    labels = [TRAJECTORY_STATE_PAIRS[channel][r][3] for channel in ('A', 'B') for r in rounds]
    trials = []
    for trial in range(1, N_TRIALS + 1):
        states = []
        for channel in ('A', 'B'):
            for r in rounds:
                subset = _trajectory_subset(frame, condition_key, r, channel, policy_by_round_by_channel[channel])
                row = subset[subset['trial'] == trial]
                if len(row) != 1:
                    raise RuntimeError(f'Expected one state for {condition_key}, {channel}, round {r}, trial {trial}; got {len(row)}.')
                states.append(_rho_from_record(row.iloc[0]))
        matrix = np.full((len(labels), len(labels)), np.nan)
        for i, left in enumerate(states):
            for j, right in enumerate(states):
                if i // len(rounds) == j // len(rounds):
                    matrix[i, j] = float(np.clip(state_fidelity(
                        DensityMatrix(left), DensityMatrix(right), validate=True), 0.0, 1.0))
        trials.append(matrix)
    return labels, np.asarray(trials)

def _state_fidelity_export_rows(records, conditions, policies):
    """Compute plot-ready means and CSV rows for each domain's available rounds."""
    rows, panels = [], {}
    for condition in conditions:
        labels, matrices = _trajectory_fidelity_matrices(records, condition, policies)
        means = np.full(matrices.shape[1:], np.nan)
        for i in range(len(labels)):
            for j in range(len(labels)):
                if i // (len(labels) // 2) != j // (len(labels) // 2):
                    continue
                values = matrices[:, i, j]
                mean, ci = mean_ci99(values)
                means[i, j] = mean
                row = {'condition_key': condition, 'condition_label': HEATMAP_CONDITION_LABELS[condition],
                       'reference_state': labels[i], 'comparison_state': labels[j],
                       'dd_policy': policies[('A', 'B')[i // (len(labels)//2)]][i % (len(labels)//2)],
                       'reference_dd_policy': policies[('A', 'B')[i // (len(labels)//2)]][i % (len(labels)//2)],
                       'comparison_dd_policy': policies[('A', 'B')[j // (len(labels)//2)]][j % (len(labels)//2)],
                       'mean_state_fidelity': mean, 'ci99_halfwidth': ci, 'n_trials': len(values),
                       'included_rounds': '0,1,2,3' if condition.startswith('hw_') else '0,1,2'}
                row.update({f'trial_{k + 1}': float(value) for k, value in enumerate(values)})
                rows.append(row)
        panels[condition] = (labels, means)
    return pd.DataFrame(rows), panels








```

```python
# Shared-preparation repair. Definitions only; hardware is submitted by the sweep cell.
import hashlib
import io
from itertools import product
from uuid import uuid4
from qiskit import qpy

_SHARED_BACKEND = None
_SHARED_LAYOUT = None
_SHARED_PREFIX_CACHE = {}
_SHARED_ROUTED_CACHE = {}
_SHARED_AUDIT_ROWS = []
# New results must not overwrite or be mixed with the earlier acquisition.
OUTPUT_DIR = Path.cwd() / 'mcm_shared_prep_rank1_hybrid_v7_v5_m3'
MCM_DD_AUDIT_CSV = OUTPUT_DIR / 'legacy_dd_audit.csv'
EXPERIMENT_VARIANT = 'MCM cumulative master v3 compact execution; nested R0-R3 physical prefixes; no-DD/XY4/XX'


def _shared_digest(circuit):
    """Stable instruction hash: QPY serialization itself contains varying identifiers."""
    def describe(block):
        rows = []
        for item in block.data:
            op = item.operation
            branches = tuple(getattr(op, 'blocks', ()))
            condition = getattr(op, 'condition', None)
            if isinstance(condition, tuple):
                variable, value = condition
                try:
                    variable = ['clbit', block.find_bit(variable).index]
                except Exception:
                    variable = repr(variable)
                condition = [variable, int(value)]
            rows.append(dict(name=op.name, qubits=[block.find_bit(q).index for q in item.qubits],
                             clbits=[block.find_bit(c).index for c in item.clbits],
                             parameters=[] if branches else [str(p) for p in op.params],
                             condition=condition, label=getattr(op, 'label', None),
                             delay_unit=getattr(op, 'unit', None),
                             branches=[describe(branch) for branch in branches]))
        return dict(num_qubits=block.num_qubits, num_clbits=block.num_clbits,
                    global_phase=str(block.global_phase), instructions=rows)
    payload = json.dumps(describe(circuit), sort_keys=True, default=str).encode('utf-8')
    return hashlib.sha256(payload).hexdigest()


def _shared_counts(circuit):
    counts = {}
    for item in circuit.data:
        name = item.operation.name
        counts[name] = counts.get(name, 0) + 1
        for block in getattr(item.operation, 'blocks', ()):
            for key, value in _shared_counts(block).items():
                counts[key] = counts.get(key, 0) + value
    return counts


_SHARED_CUMULATIVE_VERSION = 'cumulative_master_v2_recovery'
_SHARED_CUMULATIVE_CACHE = {}


def _shared_round_cuts(circuit):
    """Require one ordered, full-device boundary for every stopping point."""
    cuts = {}
    for index, item in enumerate(circuit.data):
        label = getattr(item.operation, 'label', None)
        if item.operation.name == 'barrier' and label in {
                f'CUMULATIVE_ROUND_{r}' for r in range(4)}:
            r = int(label.rsplit('_', 1)[1])
            if r in cuts or len(item.qubits) != circuit.num_qubits:
                raise RuntimeError(f'Invalid cumulative boundary {label}.')
            cuts[r] = index
    if list(cuts) != [0, 1, 2, 3]:
        raise RuntimeError('Missing or reordered cumulative round boundaries.')
    return cuts


def _shared_role_maps(circuit):
    """Recover state locations at each cut, rather than using only final_layout."""
    expected = {f'CUMULATIVE_ROLE_{r}_{channel}_{role}': (r, channel, role)
                for r in range(4) for channel in ('A', 'B')
                for role in ('ref', 'receiver')}
    locations, pending = {}, {}
    for item in circuit.data:
        op = item.operation
        label = getattr(op, 'label', None)
        physical = [circuit.find_bit(q).index for q in item.qubits]
        if op.name == 'barrier' and label in expected:
            if label in locations or len(physical) != 1:
                raise RuntimeError(f'Invalid state-location marker {label}.')
            locations[label] = physical[0]
            pending[label] = physical[0]
        elif op.name == 'barrier' and label in {
                f'CUMULATIVE_ROUND_{r}' for r in range(4)}:
            r = int(label.rsplit('_', 1)[1])
            required = {key for key, value in expected.items() if value[0] == r}
            if set(pending) != required:
                raise RuntimeError(f'Incomplete role markers at round {r}.')
            if len(set(pending.values())) != 4:
                raise RuntimeError(f'Overlapping A/B state locations at round {r}.')
            pending.clear()
        elif op.name not in ('barrier', 'delay') and set(physical) & set(pending.values()):
            raise RuntimeError('An operation moved past a state-location marker.')
    if set(locations) != set(expected) or pending:
        raise RuntimeError('Compiler or scheduler lost cumulative state-location markers.')
    return {r: {channel: [locations[f'CUMULATIVE_ROLE_{r}_{channel}_{role}']
                          for role in ('ref', 'receiver')]
                for channel in ('A', 'B')} for r in range(4)}


def _shared_syndrome_map(circuit, count):
    measured = {}
    for item in circuit.data:
        if item.operation.name in ('measure', 'measure_2'):
            bit = circuit.find_bit(item.clbits[0]).index
            if bit < count:
                if bit in measured:
                    raise RuntimeError('A syndrome bit was measured more than once.')
                measured[bit] = circuit.find_bit(item.qubits[0]).index
    if set(measured) != set(range(count)):
        raise RuntimeError('Incomplete cumulative syndrome measurement map.')
    return [measured[i] for i in range(count)]


def _shared_cut_prefix(master, round_number):
    """Copy an existing physical prefix verbatim. Never transpile or reschedule it."""
    label = f'CUMULATIVE_ROUND_{round_number}'
    cuts = [i for i, item in enumerate(master.data)
            if item.operation.name == 'barrier' and item.operation.label == label]
    if len(cuts) != 1:
        raise RuntimeError(f'Expected one preserved boundary {label}.')
    registers = master.cregs[:2 * round_number]
    expected_names = [f'r{r}_m{channel}' for r in range(1, round_number + 1)
                      for channel in ('A', 'B')]
    if [reg.name for reg in registers] != expected_names:
        raise RuntimeError('Unexpected cumulative syndrome-register order.')
    prefix = QuantumCircuit(*master.qregs, *registers,
                            name=f'cumulative_preparation_round_{round_number}')
    prefix.global_phase = master.global_phase
    prefix.metadata = copy.deepcopy(master.metadata or {})
    prefix.metadata.update(teleportation_rounds=round_number,
                           shared_compilation_scheme=_SHARED_CUMULATIVE_VERSION)
    for item in master.data[:cuts[0] + 1]:
        cindices = [master.find_bit(c).index for c in item.clbits]
        if any(c >= prefix.num_clbits for c in cindices):
            raise RuntimeError('A later-round operation crossed the prefix boundary.')
        prefix.append(item.operation,
                      [prefix.qubits[master.find_bit(q).index] for q in item.qubits],
                      [prefix.clbits[c] for c in cindices])
    if _shared_counts(prefix).get('measure_2', 0) != 4 * round_number:
        raise RuntimeError('Unexpected syndrome count in cumulative prefix.')
    return prefix


def _shared_assert_nested(shorter, longer, round_number):
    cut = _shared_cut_prefix(longer, round_number)
    if _shared_digest(shorter) != _shared_digest(cut):
        raise RuntimeError(
            f'Round {round_number} physical gates, conditions, or delays changed '
            'inside a longer experiment. No jobs submitted.')


def _shared_active_qubits(circuit, physical_indices=None):
    """Control-flow qargs can include idle device wires; inspect actual leaves."""
    physical_indices = (list(range(circuit.num_qubits)) if physical_indices is None
                        else physical_indices)
    active = set()
    for item in circuit.data:
        physical = [physical_indices[circuit.find_bit(q).index] for q in item.qubits]
        branches = getattr(item.operation, 'blocks', ())
        if branches:
            for branch in branches:
                active.update(_shared_active_qubits(branch, physical))
        elif item.operation.name not in ('barrier', 'delay'):
            active.update(physical)
    return sorted(active)


def _shared_compile_candidate(backend, full_layout, strategy):
    """Compile/schedule a dual-channel master once, DD once per policy, then cut."""
    if len(full_layout) != 19 or len(set(full_layout)) != 19:
        raise ValueError('Cumulative compilation requires one distinct 19-qubit layout.')
    template = build_cascade_epr_half_teleportation_circuit(3, round_boundaries=True)
    anchor = ClassicalRegister(4, 'shared_anchor')
    template.add_register(anchor)
    specs = cascade_channel_specs(full_layout)
    final_indices = specs['A'][3]['indices'] + specs['B'][3]['indices']
    _append_measure_2(template, [template.qubits[i] for i in final_indices], list(anchor))
    route_key = (id(backend), backend.name, strategy['seed_transpiler'], tuple(full_layout),
                 _SHARED_CUMULATIVE_VERSION, _shared_digest(template))
    if set(DD_POLICIES) != {'none', 'XX', 'XY4'}:
        raise ValueError('Cumulative audit expects none, XX and XY4 policies.')

    if route_key not in _SHARED_ROUTED_CACHE:
        routed = transpile(template, backend=backend, initial_layout=list(full_layout),
                           optimization_level=0, seed_transpiler=strategy['seed_transpiler'],
                           approximation_degree=1.0, routing_method='sabre')
        # Include route-added bridge qubits in synchronization boundaries too.
        expanded = routed.copy_empty_like()
        for item in routed.data:
            if item.operation.name == 'barrier' and getattr(item.operation, 'label', '') in {
                    f'CUMULATIVE_{kind}_{r}' for kind in ('ROUND', 'GUARD') for r in range(4)}:
                expanded.barrier(*expanded.qubits, label=item.operation.label)
            else:
                expanded.append(item.operation, item.qubits, item.clbits)
        expanded._layout = routed.layout
        _shared_round_cuts(expanded)
        _shared_role_maps(expanded)
        _SHARED_ROUTED_CACHE[route_key] = expanded
    routed = _SHARED_ROUTED_CACHE[route_key]
    routed_pairs = _shared_role_maps(routed)
    routed_syndromes = _shared_syndrome_map(routed, 12)
    isa = transpile(routed, backend=backend, initial_layout=list(range(routed.num_qubits)),
                    routing_method='none', optimization_level=OPTIMIZATION_LEVEL,
                    seed_transpiler=strategy['seed_transpiler'], approximation_degree=1.0)
    if (_shared_role_maps(isa) != routed_pairs
            or _shared_syndrome_map(isa, 12) != routed_syndromes):
        raise RuntimeError('Native optimization changed the frozen routed state locations.')
    _shared_round_cuts(isa)
    schedule, translator, alignment = build_mcm_scheduling_pipeline(backend, 'XY4', full_layout)
    if strategy['scheduling_method'] == 'asap':
        from qiskit_ibm_runtime.transpiler.passes.scheduling import ASAPScheduleAnalysis
        target = copy.deepcopy(backend.target)
        schedule = PassManager([ASAPScheduleAnalysis(target=target), PadDelay(
            target=target, fill_very_end=False, schedule_idle_qubits=False)])
    master = schedule.run(isa)
    _shared_round_cuts(master)
    if _shared_role_maps(master) != routed_pairs:
        raise RuntimeError('Scheduling changed a cumulative state-location marker.')
    active = _shared_active_qubits(master)
    family = {'version': _SHARED_CUMULATIVE_VERSION, 'prefixes': {}, 'stats': {},
              'masters': {}, 'mappings': {}, 'pairs': routed_pairs, 'ancestors': {},
              'recovery_strategy': dict(strategy), 'failures': [],
              'master_sha256': _shared_digest(master),
              'routing_master_sha256': _shared_digest(routed)}
    for policy in ['none'] + [p for p in DD_POLICIES if p != 'none']:
        try:
            if policy == 'none':
                prepared = master.copy()
                cycle_counts = {r: 0 for r in range(4)}
            else:
                prepared, stats = _insert_control_flow_safe_dd(master, backend, active, alignment, policy)
                cycle_counts = stats['dd_cycles_by_round']
                if set(cycle_counts) != set(range(4)):
                    raise RuntimeError('DD did not record every cumulative boundary.')
                if policy == 'XY4' and translator is not None:
                    prepared = translator.run(prepared)
            _shared_round_cuts(prepared)
            if _shared_role_maps(prepared) != routed_pairs:
                raise RuntimeError('DD changed cumulative state locations.')
            for operation in ('measure_2', 'if_else'):
                if _shared_counts(prepared).get(operation, 0) != _shared_counts(master).get(operation, 0):
                    raise RuntimeError(f'DD changed the {operation} count.')
            family['masters'][policy] = prepared
            family['prefixes'][policy] = {r: _shared_cut_prefix(prepared, r) for r in range(4)}
            family['stats'][policy] = {r: {'dd_cycles': cycle_counts[r]} for r in range(4)}
            family['ancestors'][policy] = {}
            for r, prefix in family['prefixes'][policy].items():
                for earlier in range(r):
                    _shared_assert_nested(family['prefixes'][policy][earlier], prefix, earlier)
                family['ancestors'][policy][r] = {
                    str(earlier): _shared_digest(family['prefixes'][policy][earlier])
                    for earlier in range(r + 1)}
        except Exception as exc:
            if policy == 'none':
                raise
            for section in ('masters', 'prefixes', 'stats', 'ancestors'):
                family[section].pop(policy, None)
            family['failures'].append(dict(stage='policy_preparation', policy=policy,
                error_type=type(exc).__name__, error_message=str(exc)))
    for r in range(4):
        syndrome_map = _shared_syndrome_map(family['prefixes']['none'][r], 4 * r)
        family['mappings'][r] = {channel: syndrome_map + routed_pairs[r][channel]
                                 for channel in ('A', 'B')}
    return family


def _shared_write_cumulative_audit(family):
    folder = OUTPUT_DIR / 'audit'
    folder.mkdir(parents=True, exist_ok=True)
    stem = f"cumulative_level_{OPTIMIZATION_LEVEL}_{family['master_sha256'][:12]}"
    for policy, master in family['masters'].items():
        with (folder / f'{stem}_{policy}_master.qpy').open('wb') as stream:
            qpy.dump(master, stream)
    payload = dict(version=family['version'], optimization_level=OPTIMIZATION_LEVEL,
                   master_sha256=family['master_sha256'],
                   routing_master_sha256=family['routing_master_sha256'],
                   recovery_strategy=family['recovery_strategy'],
                   preflight_complete=family['preflight_complete'],
                   unavailable_preparation_conditions=family['failures'],
                   physical_pairs_by_round=family['pairs'],
                   measurement_maps_by_round=family['mappings'],
                   prefix_sha256_by_policy_and_round={policy: {
                       r: _shared_digest(prefix) for r, prefix in rounds.items()}
                       for policy, rounds in family['prefixes'].items()},
                   cumulative_dd_cycles=family['stats'],
                   ancestor_prefix_sha256=family['ancestors'],
                   physical_prefixes_verified=True,
                   preparation_shared_across_channels=True,
                   boundaries_include_routing_ancillas=True)
    (folder / f'{stem}.json').write_text(json.dumps(payload, indent=2), encoding='utf-8')


def _shared_attach_analysis(prefix, suffix, backend, physical_pair, seed_transpiler=None):
    """Compatibility entry point for offline checks and individual analyzers."""
    analyzer = _shared_compact_analyzer(
        backend, suffix, physical_pair,
        SEED_TRANSPILER if seed_transpiler is None else seed_transpiler)
    return _shared_join_analysis(prefix, analyzer, physical_pair, backend)


def _shared_pair_distribution(distribution, width):
    """Marginalize syndrome outcomes without postselection; last two bits are analysis."""
    pair = {}
    for key, value in distribution.items():
        bits = normalize_bit_key(key, width)
        analysis = bits[:2]
        pair[analysis] = pair.get(analysis, 0.0) + float(value)
    return pair


def _shared_fit_tomography(qst, circuits, pair_distributions, shots):
    # Reuse Qiskit Experiments' analysis and the metadata generated by its basis circuits.
    # target=None below prevents automatic ideal-state simulation.
    data = ExperimentData(experiment=qst)
    for basis_circuit, distribution in zip(circuits, pair_distributions):
        counts = probabilities_to_integer_counts(distribution, shots)
        data.add_data({'counts': counts, 'shots': int(shots),
                       'metadata': copy.deepcopy(basis_circuit.metadata)})
    qst.analysis.run(data, replace_results=True).block_for_results()
    result = data.analysis_results(dataframe=True)
    states = result[result['name'] == 'state']
    if len(states) != 1:
        raise RuntimeError(f'Expected one unconditional tomography state, got {len(states)}.')
    return np.asarray(states.iloc[0]['value'].data, dtype=complex)


def _shared_measure_pair(backend, sampler, mitigator, family, channel, policy, label, round_number, domain='Hardware'):
    plan, execution_circuits = _shared_execution_circuits(
        backend, family, policy, round_number, channel, domain)
    prefix = family['prefixes']['none'][round_number]
    mapping = family['mappings'][round_number][channel]
    stats = family['stats'][policy][round_number]
    common_digest = family['prefix_sha256']['none'][round_number]
    prepared_digest = plan['prepared_sha256']
    qst, tomo_suffixes = plan['qst'], plan['tomo_suffixes']
    n_circuits = len(execution_circuits)
    audit_id = uuid4().hex[:12]
    folder = OUTPUT_DIR / 'audit'
    folder.mkdir(parents=True, exist_ok=True)
    with (folder / f'{audit_id}_circuits.qpy').open('wb') as stream:
        qpy.dump(execution_circuits, stream)
    with (folder / f'{audit_id}_common_prefix.qpy').open('wb') as stream:
        qpy.dump(prefix, stream)
    row = dict(audit_id=audit_id, label=label, domain=domain, optimization_level=OPTIMIZATION_LEVEL,
               seed_transpiler=family['recovery_strategy']['seed_transpiler'], dd_policy=policy, round=round_number,
               common_prefix_sha256=common_digest, prepared_prefix_sha256=prepared_digest,
               identical_preparation_across_all_bases=True,
               identical_preparation_across_channels=True,
               shared_compilation_scheme=family['version'],
               cumulative_master_sha256=family['master_sha256'],
               cumulative_prepared_master_sha256=family['prepared_master_sha256'][policy],
               physical_prefixes_verified=True,
               round_boundary=f'CUMULATIVE_ROUND_{round_number}',
               ancestor_prefix_sha256=json.dumps(family['ancestors'][policy][round_number]),
               measurement_physical_map=json.dumps(mapping), dd_cycles=stats['dd_cycles'],
               n_circuits=n_circuits, analysis_optimization_level=0,
               routing_optimization_level=0, routing_frozen_across_levels=None, routing_comparison_status='not_yet_compared',
               routing_master_sha256=family['routing_master_sha256'],
               preparation_attempt=family['recovery_strategy']['attempt'],
               dd_scheduling_method=family['recovery_strategy']['scheduling_method'],
               analysis_is_outside_DD=True, scheduled_no_DD_baseline=True)
    _SHARED_AUDIT_ROWS.append(row)
    pd.DataFrame(_SHARED_AUDIT_ROWS).to_csv(folder / 'audit.csv', index=False)
    print(f"[{label}] {len(execution_circuits)} bases share preparation {prepared_digest[:12]}; DD cycles={stats['dd_cycles']}")
    # Plain Runtime sampler: prepared circuits must NOT enter the old DD wrapper again.
    result = run_sampler_job_with_retries(sampler, execution_circuits, int(SHOTS), label)
    names = [reg.name for reg in execution_circuits[0].cregs]
    counts = sampler_joint_dynamic_counts(result, names)
    # Circuits and per-shot Runtime results are no longer needed for fitting.
    del execution_circuits, result
    (folder / f'{audit_id}_counts.json').write_text(json.dumps({
        'audit': row, 'joint_counts': counts,
        'tomography_metadata': [c.metadata for c in tomo_suffixes],
        'CHSH_settings': CHSH_SETTINGS,
    }, default=str), encoding='utf-8')
    if len(counts) != n_circuits:
        raise RuntimeError('Unexpected number of measurement results; received counts saved.')
    width = len(mapping)
    rhos, chsh, checks = {}, {}, {}
    for method in ('raw', 'm3'):
        if method == 'm3' and mitigator is None:
            # Calibration failed; retain real raw data without inventing mitigated data.
            continue
        try:
            pair_distributions = (
                [_shared_pair_distribution(c, width) for c in counts] if method == 'raw' else
                _m3_recover_pair_distributions(counts, mapping, mitigator,
                                               label + f' [audit {audit_id}]'))
            rho = _shared_fit_tomography(qst, tomo_suffixes, pair_distributions[:9], int(SHOTS))
            fidelity, from_rho = metrics_from_density_matrix(rho)
            direct = compute_chsh(dict(zip(CHSH_SETTINGS, pair_distributions[9:])), 0, 1, 2)
            chsh[method] = direct
            checks[method] = dict(shared_audit_id=audit_id, shared_prefix_sha256=prepared_digest,
                                  preparation_optimization_level=OPTIMIZATION_LEVEL,
                                  shared_compilation_scheme=family['version'],
                                  cumulative_master_sha256=family['master_sha256'],
                                  physical_prefixes_verified=True,
                                  identical_preparation_across_channels=True,
                                  ancestor_prefix_sha256=json.dumps(family['ancestors'][policy][round_number]),
                                  analysis_physical_qubits=json.dumps(mapping[-2:]),
                                  routing_optimization_level=0, routing_frozen_across_levels=None, routing_comparison_status='not_yet_compared',
                                  analysis_optimization_level=0, shared_dd_cycles=stats['dd_cycles'],
                                  dd_implementation='cumulative_master_local_dd', dd_scheduling_method=family['recovery_strategy']['scheduling_method'],
                                  seed_transpiler=family['recovery_strategy']['seed_transpiler'],
                                  preparation_attempt=family['recovery_strategy']['attempt'],
                                  routing_master_sha256=family['routing_master_sha256'],
                                  acquisition_domain=domain,
                                   chsh_estimator='direct_four_settings',
                                  chsh_direct_minus_tomography=float(direct - from_rho),
                                  same_state_lower_bound_residual=float(fidelity - direct / (2*np.sqrt(2))))
            print(f'[{method}] F={fidelity:.4f}, S_direct={direct:.4f}, S_tomography={from_rho:.4f}, '
                  f'difference={direct-from_rho:+.4f} (diagnostic, not a significance test)')
            rhos[method] = rho
        except Exception as exc:
            chsh.pop(method, None)
            checks.pop(method, None)
            _shared_failure('reconstruction', exc, channel=channel, round=round_number,
                            policy=policy, domain=domain, method=method, label=label,
                            counts_file=str(folder / f'{audit_id}_counts.json'))
    if not rhos:
        raise RuntimeError('Both reconstructions failed; raw counts saved; no job resubmitted.')
    return rhos, chsh, checks



def run_three_round_dd_sweep(include_simulator_reference=True):
    """Collect each valid condition once; retain partial domains and methods."""
    global _SHARED_BACKEND, _SHARED_LAYOUT
    records = []
    _SHARED_RECORDS_BY_LEVEL[OPTIMIZATION_LEVEL] = records
    _SHARED_PLANNED_CONDITIONS_BY_LEVEL[OPTIMIZATION_LEVEL] = (
        int(N_TRIALS) * 2 * len(DD_POLICIES) * (4 + (3 if include_simulator_reference else 0)))
    if CASCADE_SCHEME != 'MCM':
        raise ValueError('This repair is for the MCM notebook.')
    if _SHARED_BACKEND is None:
        _SHARED_BACKEND = make_runtime_service().backend(BACKEND_NAME)
        validate_runtime_backend(_SHARED_BACKEND, require_measure_2=True, wait_for_active=True)
    if _SHARED_LAYOUT is None:
        _SHARED_LAYOUT = select_balanced_physical_layout(
            build_three_round_epr_half_teleportation_circuit(), _SHARED_BACKEND)
    backend, layout = _SHARED_BACKEND, _SHARED_LAYOUT
    if backend.name != BACKEND_NAME:
        raise ValueError('Backend changed mid-sweep; restart the kernel to choose a new layout.')
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    try:
        append_hardware_error_history(backend, HARDWARE_CALIBRATION_HISTORY_PATH)
    except Exception as exc:
        _shared_failure('calibration_history_export', exc)
    print('Initial layout (actual routed measurement qubits are recorded in the shared audit):', layout)

    # The hardware master and separate simulation preparations are independent.
    family = _shared_cumulative_family(backend, layout)
    if family is not None:
        _shared_write_cumulative_audit(family)
    calibrated_qubits = {'Hardware': set(), 'Noisy simulation': set()}
    if family is not None:
        for policy, r, channel in family['conditions']:
            calibrated_qubits['Hardware'].update(family['mappings'][r][channel])

    simulation_plans = {}
    # Keep only the current optimization level's scheduled simulation prefixes.
    # The small routed cache is retained to freeze routes across levels, as in NEW.
    _SHARED_SIM_PREFIX_CACHE.clear()
    if include_simulator_reference:
        specs = cascade_channel_specs(layout)
        source_base = build_source_epr_pair_circuit()
        bases = {r: build_cascade_epr_half_teleportation_circuit(r) for r in (1, 2)}
        for channel in ('A', 'B'):
            for r in (0, 1, 2):
                state_pair = trajectory_state_pair_spec(layout, channel, r)
                base = source_base if r == 0 else bases[r]
                physical = state_pair['state_physical_qubits'] if r == 0 else specs[channel][r]['layout']
                indices = [0, 1] if r == 0 else specs[channel][r]['indices']
                try:
                    mapping = _shared_separate_prefix(backend, base, physical, indices)[1]
                    simulation_plans[(r, channel)] = (base, physical, indices)
                    calibrated_qubits['Noisy simulation'].update(mapping)
                except Exception as exc:
                    _shared_failure('simulation_preparation', exc, domain='Noisy simulation',
                                    channel=channel, round=r)

    domains = []
    requested_domains = ['Hardware'] + (['Noisy simulation'] if include_simulator_reference else [])
    for domain in requested_domains:
        if not calibrated_qubits[domain]:
            _shared_failure('domain_unavailable', RuntimeError('No valid preflight conditions.'), domain=domain)
            continue
        try:
            if domain == 'Hardware':
                domain_backend = backend
                sampler = make_hardware_session_sampler(backend)
                sampler.options.dynamical_decoupling.enable = False
            else:
                domain_backend = AerSimulator.from_backend(backend)
                sampler = Sampler(mode=domain_backend)
        except Exception as exc:
            _shared_failure('domain_setup', exc, domain=domain)
            continue
        try:
            options = ({'label_suffix': 'backend-derived noisy Aer'}
                       if domain == 'Noisy simulation' else {})
            mitigator = calibrate_m3(
                domain_backend, sampler, sorted(calibrated_qubits[domain]),
                shots=M3_CALIBRATION_SHOTS, **options)
            if mitigator is None:
                raise RuntimeError('Calibration returned no M3 mitigator.')
        except Exception as exc:
            _shared_failure('m3_calibration', exc, domain=domain, raw_acquisition_available=True)
            mitigator = None
        mitigator = _m3_recovery_state(domain_backend, sampler, mitigator, domain)
        domains.append((domain, sampler, mitigator))

    previous = {}
    for trial in range(1, N_TRIALS + 1):
        for channel in ('A', 'B'):
            for r in (0, 1, 2, 3):
                policies = list(DD_POLICIES)
                offset = (trial - 1 + r + (channel == 'B')) % len(policies)
                policies = policies[offset:] + policies[:offset]
                for policy in policies:
                    # Unavailable compilation conditions were recorded during
                    # preflight; never substitute another DD policy or family.
                    for domain, sampler, mitigator in domains:
                        if domain == 'Hardware':
                            if family is None or (policy, r, channel) not in family['conditions']:
                                continue
                            strategy = family['recovery_strategy']
                        else:
                            if (r, channel) not in simulation_plans:
                                continue
                            strategy = dict(scheduling_method='alap', seed_transpiler=SEED_TRANSPILER, attempt=1)
                        label = (f'{domain}, shared-preparation, level={OPTIMIZATION_LEVEL}, '
                                 f'trial={trial}, channel={channel}, round={r}, policy={policy}')
                        context = dict(domain=domain, trial=trial, channel=channel,
                                       round=r, policy=policy, label=label)
                        try:
                            if domain == 'Hardware':
                                rhos, chsh, checks = _shared_measure_pair(
                                    backend, sampler, mitigator, family, channel,
                                    policy, label, r, domain=domain)
                            else:
                                base, physical, indices = simulation_plans[(r, channel)]
                                rhos, chsh, checks = _shared_separate_measure_pair(
                                    backend, sampler, mitigator, base, physical, indices,
                                    policy, label, r, domain=domain, channel=channel)
                        except Exception as exc:
                            # The measurement helper owns job handling. This
                            # catch never recompiles or resubmits a condition.
                            _shared_failure('condition_acquisition', exc, **context)
                            _shared_save_progress(records)
                            continue
                        for method, rho in rhos.items():
                            try:
                                key = (domain, trial, channel, policy, method, r)
                                input_rho = (rho if r == 0 else previous.get(
                                    (domain, trial, channel, policy, method, r - 1)))
                                available = input_rho is not None
                                source = ('measured_source_EPR_pair' if r <= 1
                                          else f'measured_round_{r - 1}_output')
                                if not available:
                                    source = f'unavailable_measured_round_{r - 1}_output'
                                state_pair = copy.deepcopy(trajectory_state_pair_spec(layout, channel, r))
                                if domain == 'Hardware':
                                    state_pair['analysis_circuit_logical_indices'] = list(
                                        state_pair['full_circuit_logical_indices'])
                                state_pair['state_physical_qubits'] = json.loads(
                                    checks[method]['analysis_physical_qubits'])
                                pending = []
                                append_mcm_cascade_record(
                                    pending, trial, method, channel, r, rho, chsh[method],
                                    policy, domain=domain, input_rho=input_rho,
                                    input_state_source=source, state_pair=state_pair,
                                    input_state_available=available)
                                record = pending[0]
                                record.update(checks[method])
                                record.update(
                                    optimization_level=OPTIMIZATION_LEVEL,
                                    series_key=f'{domain.lower().replace(" ", "_")}_{policy}_{method}',
                                    dd_enabled=bool(policy != 'none'),
                                    dd_sequence=policy if policy != 'none' else 'none',
                                    dd_implementation=checks[method]['dd_implementation'],
                                    dd_scheduling_method=strategy['scheduling_method'],
                                    requested_seed_transpiler=int(SEED_TRANSPILER),
                                    effective_seed_transpiler=int(strategy['seed_transpiler']),
                                    seed_transpiler=int(strategy['seed_transpiler']),
                                    recovery_attempt=int(strategy['attempt']),
                                    routing_master_sha256=checks[method]['routing_master_sha256'],
                                    routing_frozen_across_levels=pd.NA,
                                    routing_comparison_status='pending_acquired_level_comparison',
                                    round0_dd_cycles=np.nan, round0_dd_status=None,
                                    server_builtin_dd_enabled=False,
                                    simulation_model=('AerSimulator.from_backend hardware noise model'
                                                      if domain == 'Noisy simulation' else 'QPU'))
                                if r == 0:
                                    cycles = checks[method]['shared_dd_cycles']
                                    record['round0_dd_cycles'] = cycles
                                    record['round0_dd_status'] = (
                                        'disabled' if policy == 'none' else
                                        'inserted' if cycles else 'no_eligible_idle_window')
                                records.append(record)
                                previous[key] = rho
                            except Exception as exc:
                                _shared_failure('record_processing', exc, method=method, **context)
                        _shared_save_progress(records)
    _shared_save_progress(records)
    return _shared_records_frame(records)


print('Cumulative preparation loaded: one routed/scheduled master, shared A/B and R0-R3 prefixes, verified gates/timing/DD, joint tomography/CHSH.')


_SHARED_RECOVERY_EVENTS = []
_SHARED_FAILURES = []
_SHARED_RECORDS_BY_LEVEL = {}


def _shared_recovery_event(stage, exc=None, **context):
    """Keep recovered attempts separate from conditions that could not be acquired."""
    from datetime import datetime, timezone
    event = dict(timestamp=datetime.now(timezone.utc).isoformat(), stage=stage,
                 level=OPTIMIZATION_LEVEL)
    event.update(context)
    if exc is not None:
        event.update(error_type=type(exc).__name__, error_message=str(exc))
    _SHARED_RECOVERY_EVENTS.append(event)
    try:
        OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        with (OUTPUT_DIR / 'recovery_events.jsonl').open('a', encoding='utf-8') as stream:
            stream.write(json.dumps(event, default=str) + '\n')
    except OSError as log_error:
        print(f'Could not persist recovery log: {log_error}', flush=True)
    print(f'[{stage}] {context} {str(exc) if exc is not None else ""}', flush=True)
    return event


def _shared_failure(stage, exc=None, **context):
    event = _shared_recovery_event(stage, exc, **context)
    _SHARED_FAILURES.append(event)
    return event


def _shared_recovery_strategies():
    # The requested level, physical layout and DD policy never change silently.
    return [dict(attempt=i + 1, seed_transpiler=int(SEED_TRANSPILER) + offset,
                 scheduling_method=schedule)
            for i, (offset, schedule) in enumerate(
                [(0, 'alap'), (101, 'alap'), (202, 'asap'), (303, 'asap')])]


def _shared_build_condition(backend, family, policy, round_number, channel):
    """Validate full circuits one at a time; retain only compact analyzers."""
    prepared = family['prefixes'][policy][round_number]
    mapping = tuple(family['mappings'][round_number][channel])
    physical_pair = mapping[-2:]
    bundle = _shared_analysis_bundle(family)
    suffixes = bundle['suffixes']  # 9 tomography + 4 direct CHSH at every round
    cache = family['_analysis_cache']
    analyzers = []
    seed = family['recovery_strategy']['seed_transpiler']
    for basis_index, suffix in enumerate(suffixes):
        key = (physical_pair, basis_index, seed)
        if key not in cache:
            cache[key] = _shared_compact_analyzer(backend, suffix, physical_pair, seed)
        analyzer = cache[key]
        # Complete preflight still occurs before any calibration or experiment.
        # The temporary full circuit is released before assembling the next one.
        checked = _shared_join_analysis(prepared, analyzer, physical_pair, backend)
        del checked
        analyzers.append(analyzer)
    return dict(
        qst=bundle['qst'], tomo_suffixes=bundle['tomo_suffixes'],
        analyzers=tuple(analyzers), measurement_map=mapping,
        prepared_sha256=family['prefix_sha256'][policy][round_number],
        analyzer_sha256=tuple(_shared_digest(c) for c in analyzers))


def _shared_preflight_candidate(backend, family):
    """Snapshot the verified prefixes and check all conditions with bounded RAM."""
    family['conditions'] = {}
    family['prefix_sha256'] = {
        policy: {r: _shared_digest(circuit) for r, circuit in rounds.items()}
        for policy, rounds in family['prefixes'].items()}
    family['prepared_master_sha256'] = {
        policy: _shared_digest(master) for policy, master in family['masters'].items()}
    for policy in DD_POLICIES:
        if policy not in family['prefixes']:
            continue
        for round_number in range(4):
            for channel in ('A', 'B'):
                try:
                    plan = _shared_build_condition(backend, family, policy, round_number, channel)
                    family['conditions'][(policy, round_number, channel)] = plan
                except Exception as exc:
                    family['failures'].append(dict(
                        stage='analysis_preflight', policy=policy, round=round_number,
                        channel=channel, error_type=type(exc).__name__, error_message=str(exc)))
    return family


def _shared_verify_plan(family, policy, round_number, channel):
    """Check stored prefixes and analyzers; full circuits do not remain in RAM."""
    plan = family['conditions'][(policy, round_number, channel)]
    expected_prefixes = family['prefix_sha256'][policy]
    # Nested identity was checked by _shared_compile_candidate. Checking these
    # snapshots detects later mutation without copying each ancestor again.
    for earlier in range(round_number + 1):
        prefix = family['prefixes'][policy][earlier]
        if _shared_digest(prefix) != expected_prefixes[earlier]:
            raise RuntimeError('Selected preparation changed after preflight.')
    if plan['prepared_sha256'] != expected_prefixes[round_number]:
        raise RuntimeError('Analysis plan refers to a different preparation.')
    if tuple(family['mappings'][round_number][channel]) != plan['measurement_map']:
        raise RuntimeError('Selected measurement map changed after preflight.')
    if len(plan['analyzers']) != len(plan['analyzer_sha256']):
        raise RuntimeError('An analysis basis was removed after preflight.')
    for analyzer, digest in zip(plan['analyzers'], plan['analyzer_sha256']):
        if _shared_digest(analyzer) != digest:
            raise RuntimeError('An analysis basis changed after preflight.')
    return plan


def _shared_cumulative_family(backend, full_layout):
    """Initial compilation + three changed strategies; select one family only."""
    if (len(full_layout) != 19 or len(set(full_layout)) != 19
            or set(DD_POLICIES) != {'none', 'XX', 'XY4'}):
        _shared_failure('preparation_configuration', ValueError(
            'Need 19 distinct physical qubits and policies none/XX/XY4. '
            'Routing retries cannot repair this configuration.'))
        return None
    best = None
    expected = len(DD_POLICIES) * 4 * 2
    for strategy in _shared_recovery_strategies():
        if strategy['attempt'] > 1:
            # A failed cache entry must never contaminate a changed attempt.
            _SHARED_CUMULATIVE_CACHE.clear()
            _SHARED_ROUTED_CACHE.clear()
        try:
            candidate = _shared_compile_candidate(backend, full_layout, strategy)
            candidate = _shared_preflight_candidate(backend, candidate)
            available = len(candidate['conditions'])
            _shared_recovery_event('preparation_attempt', strategy=strategy,
                                   valid_conditions=available, expected_conditions=expected,
                                   preparation_failures=candidate['failures'])
            if available and (best is None or available > len(best['conditions'])):
                best = candidate
            if available == expected:
                break
        except Exception as exc:
            _shared_recovery_event('preparation_attempt_failed', exc, strategy=strategy)
    # Persist the selected family before any acquisition; do not return invalid candidates.
    if best is None:
        _shared_failure('preparation_exhausted', RuntimeError(
            'No valid preparation after initial attempt plus three recovery attempts.'))
        return None
    best['preflight_complete'] = len(best['conditions']) == expected
    for failure in best['failures']:
        details = dict(failure)
        stage = details.pop('stage')
        _shared_failure(stage, selected_attempt=best['recovery_strategy']['attempt'], **details)
    _shared_recovery_event('preparation_selected', strategy=best['recovery_strategy'],
        valid_conditions=len(best['conditions']), expected_conditions=expected,
        master_sha256=best['master_sha256'], complete=best['preflight_complete'])
    return best


_SHARED_PLANNED_CONDITIONS_BY_LEVEL = {}


def _shared_empty_records():
    """Keep downstream exports and filters valid when a level has no usable rows."""
    return pd.DataFrame(columns=[
        'optimization_level', 'domain', 'trial', 'channel', 'round', 'dd_policy',
        'method', 'series_key', 'fidelity', 'chsh_S_direct', 'chsh_S_from_rho',
        'equation9_lower_from_direct_S', 'equation9_upper_from_direct_S',
        'chsh_direct_minus_tomography', 'same_state_lower_bound_residual',
        'rho_real', 'rho_imag', 'input_rho_real', 'input_rho_imag',
        'input_state_available', 'input_state_source', 'shared_audit_id',
        'shared_prefix_sha256', 'routing_master_sha256',
        'requested_seed_transpiler', 'effective_seed_transpiler',
        'dd_scheduling_method', 'recovery_attempt', 'round0_dd_cycles', 'round0_dd_status',
        'routing_frozen_across_levels',
        'routing_comparison_status'])


def _shared_records_frame(records):
    frame = records.copy() if isinstance(records, pd.DataFrame) else pd.DataFrame(records)
    return frame if not frame.empty else _shared_empty_records()


def _shared_update_status(records, planned_conditions):
    """An acquired condition can still be partial if one reconstruction is absent."""
    from datetime import datetime, timezone
    frame = _shared_records_frame(records)
    planned = int(planned_conditions)
    keys = ['optimization_level', 'domain', 'trial', 'channel', 'round', 'dd_policy']
    completed = len(frame[keys].drop_duplicates()) if len(frame) else 0
    method_rows = len(frame[keys + ['method']].drop_duplicates()) if len(frame) else 0
    payload = dict(
        updated_at=datetime.now(timezone.utc).isoformat(),
        status=('no_valid_data' if frame.empty else
                'complete' if completed == planned and method_rows == 2 * planned
                and not _SHARED_FAILURES else 'partial'),
        planned_conditions=planned, completed_conditions=completed,
        missing_conditions=max(0, planned - completed),
        expected_method_rows=2 * planned, saved_method_rows=method_rows,
        missing_method_rows=max(0, 2 * planned - method_rows),
        saved_rows=len(frame), valid_records=len(frame), failure_count=len(_SHARED_FAILURES),
        failures=list(_SHARED_FAILURES),
        recovery_event_count=len(_SHARED_RECOVERY_EVENTS))
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    temporary = OUTPUT_DIR / 'acquisition_status.json.tmp'
    temporary.write_text(json.dumps(payload, indent=2, default=str), encoding='utf-8')
    temporary.replace(OUTPUT_DIR / 'acquisition_status.json')
    return payload


def _shared_save_progress(records):
    """Export errors do not discard the live rows held for final level export."""
    try:
        OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        _shared_records_frame(records).to_csv(
            OUTPUT_DIR / f'shared_level_{OPTIMIZATION_LEVEL}_checkpoint.csv', index=False)
    except Exception as exc:
        _shared_failure('checkpoint_export', exc)
    try:
        live = [row for rows in _SHARED_RECORDS_BY_LEVEL.values() for row in rows]
        _shared_update_status(live, sum(_SHARED_PLANNED_CONDITIONS_BY_LEVEL.values()))
    except Exception as exc:
        _shared_failure('status_export', exc)


def _shared_route_comparison_group(records):
    """Verify routing identity only from the routes used in acquired levels."""
    frame = _shared_records_frame(records)
    levels = frame['optimization_level'].dropna().unique()
    hashes = frame.get('routing_master_sha256', pd.Series(index=frame.index, dtype=object))
    if len(levels) < 2:
        verified, status = pd.NA, 'unverified_fewer_than_two_acquired_levels'
    elif hashes.isna().any() or hashes.astype(str).str.len().eq(0).any():
        verified, status = pd.NA, 'unverified_missing_route_hash'
    else:
        verified = bool(hashes.nunique() == 1)
        status = 'identical_across_acquired_levels' if verified else 'different_across_acquired_levels'
    frame['routing_frozen_across_levels'] = verified
    frame['routing_comparison_status'] = status
    return frame

def _shared_route_comparison(records):
    """Compare matching preparations across levels, never different domains/rounds."""
    frame = _shared_records_frame(records)
    if frame.empty:
        return _shared_route_comparison_group(records)
    groups = []
    hardware = frame[frame['domain'] == 'Hardware']
    if not hardware.empty:
        groups.append(_shared_route_comparison_group(hardware.to_dict('records')))
    simulation = frame[frame['domain'] != 'Hardware']
    for _, group in simulation.groupby(['domain', 'channel', 'round'], sort=False, dropna=False):
        groups.append(_shared_route_comparison_group(group.to_dict('records')))
    return pd.concat(groups, ignore_index=True)



def _shared_analysis_bundle(family):
    """Keep one small basis library per compilation attempt."""
    if '_analysis_bundle' not in family:
        qst = StateTomography(QuantumCircuit(2), target=None)
        qst.set_transpile_options(
            optimization_level=0,
            seed_transpiler=family['recovery_strategy']['seed_transpiler'])
        tomography = tuple(qst.circuits())
        if len(tomography) != 9:
            raise RuntimeError('Expected nine two-qubit Pauli tomography bases.')
        suffixes = list(tomography)
        for setting_a, setting_b in CHSH_SETTINGS:
            suffix = QuantumCircuit(2, 2)
            suffix.ry(-CHSH_ANGLES[setting_a], 0)
            suffix.ry(-CHSH_ANGLES[setting_b], 1)
            _append_measure_2(suffix, [0, 1], [0, 1])
            suffixes.append(suffix)
        family['_analysis_bundle'] = dict(
            qst=qst, tomo_suffixes=tomography, suffixes=tuple(suffixes))
        family['_analysis_cache'] = {}
    return family['_analysis_bundle']


def _shared_compact_analyzer(backend, suffix, physical_pair, seed_transpiler):
    """Compile the same analyzer as v2, retaining only its two used wires."""
    compiled = transpile(
        suffix, backend=backend, initial_layout=list(physical_pair),
        optimization_level=0, seed_transpiler=seed_transpiler,
        approximation_degree=1.0)
    physical_to_local = {int(q): i for i, q in enumerate(physical_pair)}
    if len(physical_to_local) != 2 or compiled.num_clbits != 2:
        raise RuntimeError('Analyzer must measure two distinct physical qubits.')
    compact = QuantumCircuit(2, 2, name='compact_native_analyzer')
    compact.global_phase = compiled.global_phase
    for item in compiled.data:
        # Basis rotations and readouts must not route onto another qubit.
        physical = [compiled.find_bit(q).index for q in item.qubits]
        if (any(q not in physical_to_local for q in physical)
                or getattr(item.operation, 'blocks', ())):
            raise RuntimeError('Analyzer compilation used an unexpected wire or control flow.')
        compact.append(item.operation,
                       [compact.qubits[physical_to_local[q]] for q in physical],
                       [compact.clbits[compiled.find_bit(c).index] for c in item.clbits])
    return compact


def _shared_join_analysis(prefix, analyzer, physical_pair, backend, *, validate=True):
    """Append native operations without recompiling or rescheduling preparation."""
    out = prefix.copy()
    analysis = ClassicalRegister(2, 'analysis')
    out.add_register(analysis)
    out.global_phase += analyzer.global_phase
    for item in analyzer.data:
        out.append(item.operation,
                   [out.qubits[physical_pair[analyzer.find_bit(q).index]] for q in item.qubits],
                   [analysis[analyzer.find_bit(c).index] for c in item.clbits])
    if list(out.data[:len(prefix.data)]) != list(prefix.data):
        raise RuntimeError('Appending the analyzer changed the state preparation.')
    if validate:
        _validate_physical_readouts(out, backend)
        assert_terminal_measurements_are_final(out, 'shared-preparation final circuit')
    return out


def _shared_execution_circuits(backend, family, policy, round_number, channel, domain):
    """Materialize one condition only, avoiding a second batch for noisy Aer."""
    plan = _shared_verify_plan(family, policy, round_number, channel)
    prepared = family['prefixes'][policy][round_number]
    circuits = []
    for analyzer in plan['analyzers']:
        # Every full circuit passed readout/terminal checks during preflight.
        # Prefix and analyzer snapshots were just checked again above.
        circuit = _shared_join_analysis(
            prepared, analyzer, plan['measurement_map'][-2:], backend, validate=False)
        if domain != 'Hardware':
            circuit = _rewrite_readouts(circuit, simulator=True)
            if _shared_counts(circuit).get('measure_2', 0):
                raise RuntimeError('Noisy-Aer circuit retained measure_2 after readout conversion.')
        circuits.append(circuit)
    return plan, circuits


_SHARED_SIM_PREFIX_CACHE = {}
_SHARED_SIM_ROUTED_CACHE = {}

def _shared_separate_prefix(backend, base, layout, indices):
    """One transpilation and schedule, ending at a marked analysis boundary."""
    key = (backend.name, OPTIMIZATION_LEVEL, SEED_TRANSPILER,
           tuple(layout), tuple(indices), _shared_digest(base))
    if key in _SHARED_SIM_PREFIX_CACHE:
        return _SHARED_SIM_PREFIX_CACHE[key]
    template = base.copy()
    anchor = ClassicalRegister(2, 'shared_anchor')
    template.add_register(anchor)
    template.barrier(*template.qubits, label='SHARED_ANALYSIS_BOUNDARY')
    _append_measure_2(template, [template.qubits[i] for i in indices], list(anchor))
    # Route ONCE at level 0, then compare native optimization levels on fixed physical wires.
    # Otherwise each level can move the state to different physical qubits.
    route_key = (backend.name, SEED_TRANSPILER, tuple(layout), tuple(indices), _shared_digest(base))
    if route_key not in _SHARED_SIM_ROUTED_CACHE:
        _SHARED_SIM_ROUTED_CACHE[route_key] = transpile(
            template, backend=backend, initial_layout=list(layout),
            optimization_level=0, seed_transpiler=SEED_TRANSPILER, approximation_degree=1.0)
    routed = _SHARED_SIM_ROUTED_CACHE[route_key]
    isa = transpile(routed, backend=backend, initial_layout=list(range(routed.num_qubits)),
                    routing_method='none', optimization_level=OPTIMIZATION_LEVEL,
                    seed_transpiler=SEED_TRANSPILER, approximation_degree=1.0)
    measured = {}
    for item in isa.data:
        if item.operation.name in ('measure', 'measure_2'):
            measured[isa.find_bit(item.clbits[0]).index] = isa.find_bit(item.qubits[0]).index
    if set(measured) != set(range(base.num_clbits + 2)):
        raise RuntimeError('Incomplete classical-to-physical measurement mapping in shared template.')
    mapping = [measured[i] for i in range(base.num_clbits + 2)]
    routed_map = {routed.find_bit(item.clbits[0]).index: routed.find_bit(item.qubits[0]).index
                  for item in routed.data if item.operation.name in ('measure', 'measure_2')}
    if measured != routed_map:
        raise RuntimeError('Native optimization moved measurements despite frozen routing.')
    schedule, translator, alignment = build_mcm_scheduling_pipeline(backend, 'XY4', layout)
    scheduled = schedule.run(isa)
    boundaries = [i for i, item in enumerate(scheduled.data)
                  if item.operation.name == 'barrier'
                  and getattr(item.operation, 'label', None) == 'SHARED_ANALYSIS_BOUNDARY']
    if len(boundaries) != 1:
        raise RuntimeError('Scheduler did not preserve the unique analysis boundary; no jobs submitted for this case.')
    boundary = boundaries[0]
    # Retain syndrome registers, but discard the temporary terminal measurement register.
    prefix = QuantumCircuit(*scheduled.qregs, *base.cregs, name='shared_state_preparation')
    prefix.global_phase = scheduled.global_phase
    for item in scheduled.data[:boundary + 1]:
        cindices = [scheduled.find_bit(c).index for c in item.clbits]
        if any(i >= base.num_clbits for i in cindices):
            raise RuntimeError('Analysis measurement moved before the shared boundary.')
        prefix.append(item.operation,
                      [prefix.qubits[scheduled.find_bit(q).index] for q in item.qubits],
                      [prefix.clbits[i] for i in cindices])
    active = sorted({scheduled.find_bit(q).index for item in scheduled.data
                     if item.operation.name not in ('barrier', 'delay') for q in item.qubits})
    answer = (prefix, mapping, translator, alignment, active)
    _SHARED_SIM_PREFIX_CACHE[key] = answer
    return answer

def _shared_separate_attach_analysis(prefix, suffix, backend, physical_pair):
    """Compile only a two-qubit analyzer; never recompile or DD-pad preparation."""
    suffix = transpile(suffix, backend=backend, initial_layout=physical_pair,
                       optimization_level=0, seed_transpiler=SEED_TRANSPILER,
                       approximation_degree=1.0)
    out = prefix.copy()
    analysis = ClassicalRegister(2, 'analysis')
    out.add_register(analysis)
    out.global_phase += suffix.global_phase
    for item in suffix.data:
        out.append(item.operation,
                   [out.qubits[suffix.find_bit(q).index] for q in item.qubits],
                   [analysis[suffix.find_bit(c).index] for c in item.clbits])
    if list(out.data[:len(prefix.data)]) != list(prefix.data):
        raise RuntimeError('Appending the analyzer changed the state preparation.')
    _validate_physical_readouts(out, backend)
    assert_terminal_measurements_are_final(out, 'shared-preparation final circuit')
    return out

def _shared_separate_measure_pair(backend, sampler, mitigator, base, layout, indices, policy, label, round_number, domain='Noisy simulation', channel=None):
    prefix, mapping, translator, alignment, active = _shared_separate_prefix(backend, base, layout, indices)
    common_digest = _shared_digest(prefix)
    if policy == 'none':
        prepared, stats = prefix.copy(), {'dd_cycles': 0}
    else:
        prepared, stats = _insert_control_flow_safe_dd(prefix, backend, active, alignment, policy)
        if policy == 'XY4' and translator is not None:
            prepared = translator.run(prepared)
    for operation in ('measure_2', 'if_else'):
        if _shared_counts(prepared).get(operation, 0) != _shared_counts(prefix).get(operation, 0):
            raise RuntimeError(f'DD changed the {operation} count.')
    prepared_digest = _shared_digest(prepared)
    # The nine basis circuits carry Qiskit's own m_idx/clbits metadata for reconstruction.
    qst = StateTomography(QuantumCircuit(2), target=None)
    qst.set_transpile_options(optimization_level=0, seed_transpiler=SEED_TRANSPILER)
    tomo_suffixes = qst.circuits()
    if len(tomo_suffixes) != 9:
        raise RuntimeError('Expected nine two-qubit Pauli tomography bases.')
    suffixes = list(tomo_suffixes)
    for setting_a, setting_b in CHSH_SETTINGS:
        suffix = QuantumCircuit(2, 2)
        suffix.ry(-CHSH_ANGLES[setting_a], 0)
        suffix.ry(-CHSH_ANGLES[setting_b], 1)
        _append_measure_2(suffix, [0, 1], [0, 1])
        suffixes.append(suffix)
    circuits = [_shared_separate_attach_analysis(prepared, suffix, backend, mapping[-2:]) for suffix in suffixes]
    execution_circuits = (circuits if domain == 'Hardware' else
                          [_rewrite_readouts(circuit, simulator=True) for circuit in circuits])
    if domain != 'Hardware':
        for circuit in execution_circuits:
            if _shared_counts(circuit).get('measure_2', 0):
                raise RuntimeError('Noisy-Aer circuit retained measure_2 after readout conversion.')
    n_circuits = len(execution_circuits)
    del circuits
    audit_id = uuid4().hex[:12]
    folder = OUTPUT_DIR / 'audit'
    folder.mkdir(parents=True, exist_ok=True)
    with (folder / f'{audit_id}_circuits.qpy').open('wb') as stream:
        qpy.dump(execution_circuits, stream)
    with (folder / f'{audit_id}_common_prefix.qpy').open('wb') as stream:
        qpy.dump(prefix, stream)
    row = dict(audit_id=audit_id, label=label, domain=domain, optimization_level=OPTIMIZATION_LEVEL,
               seed_transpiler=SEED_TRANSPILER, dd_policy=policy, round=round_number,
               common_prefix_sha256=common_digest, prepared_prefix_sha256=prepared_digest,
               identical_preparation_across_all_bases=True,
               measurement_physical_map=json.dumps(mapping), dd_cycles=stats['dd_cycles'],
               n_circuits=len(execution_circuits), analysis_optimization_level=0,
               routing_optimization_level=0, routing_frozen_across_levels=True,
               analysis_is_outside_DD=True, scheduled_no_DD_baseline=True)
    route_key = (backend.name, SEED_TRANSPILER, tuple(layout), tuple(indices), _shared_digest(base))
    route_hash = _shared_digest(_SHARED_SIM_ROUTED_CACHE[route_key])
    row.update(shared_compilation_scheme='separate_rounds_NEW',
               identical_preparation_across_channels=False,
               physical_prefixes_verified=False,
               cumulative_master_sha256=None,
               routing_master_sha256=route_hash,
               routing_frozen_across_levels=None,
               routing_comparison_status='not_yet_compared')
    _SHARED_AUDIT_ROWS.append(row)
    pd.DataFrame(_SHARED_AUDIT_ROWS).to_csv(folder / 'audit.csv', index=False)
    print(f"[{label}] {len(execution_circuits)} bases share preparation {prepared_digest[:12]}; DD cycles={stats['dd_cycles']}")
    # Plain Runtime sampler: prepared circuits must NOT enter the old DD wrapper again.
    result = run_sampler_job_with_retries(sampler, execution_circuits, int(SHOTS), label)
    names = [reg.name for reg in execution_circuits[0].cregs]
    counts = sampler_joint_dynamic_counts(result, names)
    # Circuits and per-shot Runtime results are no longer needed for fitting.
    del execution_circuits, result
    (folder / f'{audit_id}_counts.json').write_text(json.dumps({
        'audit': row, 'joint_counts': counts,
        'tomography_metadata': [c.metadata for c in tomo_suffixes],
        'CHSH_settings': CHSH_SETTINGS,
    }, default=str), encoding='utf-8')
    if len(counts) != n_circuits:
        raise RuntimeError('Unexpected number of measurement results; received counts saved.')
    width = len(mapping)
    rhos, chsh, checks = {}, {}, {}
    for method in ('raw', 'm3'):
        if method == 'm3' and mitigator is None:
            # Calibration failed; retain real raw data without inventing mitigated data.
            continue
        try:
            pair_distributions = (
                [_shared_pair_distribution(c, width) for c in counts] if method == 'raw' else
                _m3_recover_pair_distributions(counts, mapping, mitigator,
                                               label + f' [audit {audit_id}]'))
            rho = _shared_fit_tomography(qst, tomo_suffixes, pair_distributions[:9], int(SHOTS))
            fidelity, from_rho = metrics_from_density_matrix(rho)
            direct = compute_chsh(dict(zip(CHSH_SETTINGS, pair_distributions[9:])), 0, 1, 2)
            chsh[method] = direct
            checks[method] = dict(shared_audit_id=audit_id, shared_prefix_sha256=prepared_digest,
                                  preparation_optimization_level=OPTIMIZATION_LEVEL,
                                  analysis_physical_qubits=json.dumps(mapping[-2:]),
                                  routing_optimization_level=0, routing_frozen_across_levels=True,
                                  analysis_optimization_level=0, shared_dd_cycles=stats['dd_cycles'],
                                  dd_implementation='shared_prefix_local_dd', dd_scheduling_method='alap',
                                  acquisition_domain=domain,
                                   chsh_estimator='direct_four_settings',
                                  chsh_direct_minus_tomography=float(direct - from_rho),
                                  same_state_lower_bound_residual=float(fidelity - direct / (2*np.sqrt(2))))
            print(f'[{method}] F={fidelity:.4f}, S_direct={direct:.4f}, S_tomography={from_rho:.4f}, '
                  f'difference={direct-from_rho:+.4f} (diagnostic, not a significance test)')
            checks[method].update(
                shared_compilation_scheme='separate_rounds_NEW',
                identical_preparation_across_channels=False,
                physical_prefixes_verified=False,
                cumulative_master_sha256=None,
                routing_master_sha256=route_hash,
                routing_frozen_across_levels=None,
                routing_comparison_status='not_yet_compared')
            rhos[method] = rho
        except Exception as exc:
            chsh.pop(method, None)
            checks.pop(method, None)
            _shared_failure('reconstruction', exc, channel=channel, round=round_number,
                            policy=policy, domain=domain, method=method, label=label,
                            counts_file=str(folder / f'{audit_id}_counts.json'))
    if not rhos:
        raise RuntimeError('Both reconstructions failed; raw counts saved; no job resubmitted.')
    return rhos, chsh, checks

print('NEWv7: v5 M3 calibration/correction/recovery; direct CHSH at every round; cumulative hardware and separate-round noisy Aer.')


def _m3_recovery_state(backend, sampler, mitigator, domain):
    """One recovery cache per domain and calibration cycle."""
    return dict(backend=backend, sampler=sampler, original=mitigator,
                domain=domain, recovered={})


def _m3_recovery_note(state, label, **details):
    from datetime import datetime, timezone
    payload = dict(time=datetime.now(timezone.utc).isoformat(),
                   domain=state['domain'], level=OPTIMIZATION_LEVEL,
                   label=label, **details)
    path = OUTPUT_DIR / 'M3_recovery.txt'
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open('a', encoding='utf-8') as stream:
        stream.write(json.dumps(payload, default=str) + '\n')
    print('[M3 recovery] ' + json.dumps(details, default=str), flush=True)


def _m3_unusable_qubits(mitigator, qubits):
    """Reject missing/nonfinite/singular calibrations and M3's faulty flags."""
    matrices = getattr(mitigator, 'single_qubit_cals', None)
    faulty = set(getattr(mitigator, 'faulty_qubits', []) or [])
    bad = []
    for q in dict.fromkeys(qubits):
        try:
            a = np.asarray(matrices[q], dtype=float)
            valid = (a.shape == (2, 2) and np.isfinite(a).all()
                     and (a >= 0).all() and np.allclose(a.sum(axis=0), 1.0)
                     and np.linalg.cond(a) < 1e8 and q not in faulty)
        except (TypeError, IndexError, ValueError, np.linalg.LinAlgError):
            valid = False
        if not valid:
            bad.append(int(q))
    return bad


def _m3_marginal_counts(counts, width, positions):
    """Sum over excluded bits; positions are in classical-bit order (LSB first)."""
    result = {}
    for key, count in counts.items():
        bits = normalize_bit_key(key, width)[::-1]
        selected = ''.join(bits[i] for i in positions)[::-1]
        result[selected] = result.get(selected, 0) + int(count)
    return result


def _m3_correct_pairs(counts, mapping, mitigator, positions):
    selected = [mapping[i] for i in positions]
    bad = _m3_unusable_qubits(mitigator, selected)
    if bad:
        raise ValueError(f'Unusable M3 calibration for physical qubits {bad}')
    pairs = []
    for distribution in counts:
        reduced = _m3_marginal_counts(distribution, len(mapping), positions)
        corrected = m3_probability_distribution(reduced, selected, mitigator)
        values = np.asarray(list(corrected.values()), dtype=float)
        if (not values.size or not np.isfinite(values).all()
                or np.any(values < -1e-10) or values.sum() <= 0):
            raise ValueError('M3 returned an empty or nonfinite probability distribution')
        pair = _shared_pair_distribution(corrected, len(selected))
        total = sum(pair.values())
        pairs.append({key: float(value / total) for key, value in pair.items()})
    return pairs


def _m3_recover_pair_distributions(counts, mapping, state, label):
    """Original correction plus three distinct recovery attempts, no data resubmission."""
    mapping = list(map(int, mapping))
    key = tuple(mapping)
    pair = mapping[-2:]
    if len(pair) != 2 or pair[0] == pair[1]:
        raise ValueError(f'Invalid physical analysis pair: {pair}')
    pair_positions = [len(mapping)-2, len(mapping)-1]
    cached = state['recovered'].get(key)
    initial = cached or (state['original'], list(range(len(mapping))))
    try:
        result = _m3_correct_pairs(counts, mapping, *initial)
        if cached:
            _m3_recovery_note(state, label, action='reused_recovered_calibration',
                              retained_physical=[mapping[i] for i in cached[1]],
                              excluded_classical_bits=[i for i in range(len(mapping)) if i not in cached[1]])
        return result
    except Exception as exc:
        state['recovered'].pop(key, None)
        _m3_recovery_note(state, label, attempt=0, action='initial_correction_failed',
                          error=str(exc), analysis_pair=pair,
                          unusable_physical=_m3_unusable_qubits(initial[0], mapping))

    actions = ('exclude_unusable_syndrome_bits', 'recalibrate_analysis_pair',
               'calibrate_pair_qubits_individually_with_more_shots')
    for attempt, action in enumerate(actions, 1):
        shots = int(M3_CALIBRATION_SHOTS) * (4 if attempt == 3 else 1)
        _m3_recovery_note(state, label, attempt=attempt, action=action,
                          analysis_pair=pair, calibration_shots_per_prepared_state=shots if attempt > 1 else 0)
        try:
            if attempt == 1:
                mitigator = state['original']
                bad = set(_m3_unusable_qubits(mitigator, mapping))
                if bad.intersection(pair):
                    raise ValueError('Analysis-pair calibration is unusable; pair must be recalibrated')
                positions = [i for i, q in enumerate(mapping) if q not in bad]
                # If no identifiable bad matrix exists, remove the joint syndrome
                # solve that failed numerically and correct the unconditional pair.
                if not bad:
                    positions = pair_positions
            elif attempt == 2:
                mitigator = calibrate_m3(state['backend'], state['sampler'], pair,
                    shots=shots, label_suffix=f'M3 recovery 2: {label}')
                positions = pair_positions
            else:
                matrices = [None] * int(state['backend'].num_qubits)
                errors = []
                for q in pair:
                    try:
                        single = calibrate_m3(state['backend'], state['sampler'], [q],
                            shots=shots, label_suffix=f'M3 recovery 3, qubit {q}: {label}')
                        if _m3_unusable_qubits(single, [q]):
                            raise ValueError(f'Qubit {q} still has unusable calibration')
                        matrices[q] = np.asarray(single.single_qubit_cals[q], dtype=float).copy()
                    except Exception as exc:
                        errors.append(f'q{q}: {exc}')
                if errors:
                    raise RuntimeError('; '.join(errors))
                mitigator = mthree.M3Mitigation(state['backend'])
                mitigator.cals_from_matrices(matrices)
                positions = pair_positions
            recovered = _m3_correct_pairs(counts, mapping, mitigator, positions)
            state['recovered'][key] = (mitigator, positions)
            _m3_recovery_note(state, label, attempt=attempt, action='recovery_succeeded',
                              retained_physical=[mapping[i] for i in positions],
                              excluded_classical_bits=[i for i in range(len(mapping)) if i not in positions],
                              excluded_physical=[mapping[i] for i in range(len(mapping)) if i not in positions],
                              retained_matrices={str(q): np.asarray(mitigator.single_qubit_cals[q]).tolist()
                                                 for q in dict.fromkeys(mapping[i] for i in positions)},
                              shots_discarded=0, plot_method='m3')
            return recovered
        except Exception as exc:
            _m3_recovery_note(state, label, attempt=attempt, action='recovery_failed', error=str(exc))
    raise RuntimeError('M3 failed after three recovery attempts; raw counts remain saved. See M3_recovery.txt.')

```

```python
"""Plot EXISTING sweep records only. No backend connection or hardware jobs."""
from pathlib import Path
from datetime import datetime
import ast
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t


def _plot_ci(values):
    # Two-sided 99% Student-t CI of the mean across independent trials:
    # mean +/- t(0.995, n-1) * sample_std / sqrt(n).
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    n = len(values)
    return (float(values.mean()) if n else np.nan,
            float(t.ppf(.995, n - 1) * values.std(ddof=1) / np.sqrt(n)) if n > 1 else np.nan, n)


def _plot_rho(row):
    def array(value):
        return np.asarray(ast.literal_eval(value) if isinstance(value, str) else value, dtype=float)
    rho = array(row['rho_real']) + 1j * array(row['rho_imag'])
    if rho.shape != (4, 4) or not np.isfinite(rho).all():
        raise ValueError('Expected a finite 4x4 saved density matrix.')
    if not np.allclose(rho, rho.conj().T, atol=1e-7) or not np.isclose(np.trace(rho), 1, atol=1e-7):
        raise ValueError('Saved density matrix is not Hermitian and normalized.')
    eigenvalues, vectors = np.linalg.eigh((rho + rho.conj().T) / 2)
    if eigenvalues.min() < -1e-7:
        raise ValueError('Saved density matrix is not positive semidefinite.')
    eigenvalues = np.maximum(eigenvalues, 0)
    return (vectors * (eigenvalues / eigenvalues.sum())) @ vectors.conj().T


def _plot_state_fidelity(left, right):
    eigenvalues, vectors = np.linalg.eigh(left)
    root = (vectors * np.sqrt(np.maximum(eigenvalues, 0))) @ vectors.conj().T
    product = root @ right @ root
    eigenvalues = np.linalg.eigvalsh((product + product.conj().T) / 2)
    return float(np.clip(np.sqrt(np.maximum(eigenvalues, 0)).sum() ** 2, 0, 1))


TSIRELSON_BOUND = 2 * np.sqrt(2)  # 2.8284...; no quantum state can exceed this CHSH S.


def _label_sweep_bar_values(axis, entries):
    """Label each mean above its upper CI; resolve text collisions in display space.

    entries may be (bar, mean, error) or (bar, mean, error, flagged). When a
    metric has a known physical bound (currently chsh_S_direct vs the
    Tsirelson bound) and `flagged` is True, the bar is hatched/outlined and
    its label gets a dagger + red color so an unphysical value cannot be
    mistaken for a normal result while skimming the figure.
    """
    labels = []
    for entry in entries:
        bar, mean, error = entry[0], entry[1], entry[2]
        flagged = entry[3] if len(entry) > 3 else False
        if not np.isfinite(mean):
            continue
        if flagged:
            bar.set_hatch('///')
            bar.set_edgecolor('crimson')
            bar.set_linewidth(1.4)
        top = mean + (abs(error) if np.isfinite(error) else 0.0)
        text = f'{mean:.3f}' + (' †' if flagged else '')
        label = axis.annotate(
            text, (bar.get_x() + bar.get_width() / 2, top),
            xytext=(0, 5), textcoords='offset points', ha='center', va='bottom',
            fontsize=9, annotation_clip=False, clip_on=False,
            color='crimson' if flagged else 'black',
            fontweight='bold' if flagged else 'normal',
        )
        labels.append(label)
    axis.margins(y=0.18)
    return labels


def _space_sweep_bar_labels(fig, labels_by_axis):
    """Stagger touching labels and reserve enough space above them before saving."""
    for _ in range(12):
        fig.canvas.draw()
        renderer = fig.canvas.get_renderer()
        expanded = False
        for axis, labels in labels_by_axis:
            placed = []
            # Low labels first: higher labels can move upward without covering them.
            ordered = sorted(labels, key=lambda label: label.get_window_extent(renderer).y0)
            for label in ordered:
                box = label.get_window_extent(renderer).expanded(1.08, 1.15)
                for _ in range(100):
                    if not any(box.overlaps(other) for other in placed):
                        break
                    xoff, yoff = label.get_position()
                    label.set_position((xoff, yoff + 4))
                    box = label.get_window_extent(renderer).expanded(1.08, 1.15)
                placed.append(box)
            if placed:
                highest = max(box.y1 for box in placed) + 8
                if highest > axis.bbox.y1:
                    bottom, top = axis.get_ylim()
                    extra = (highest - axis.bbox.y1) / axis.bbox.height
                    axis.set_ylim(bottom, top + (top - bottom) * (extra + 0.04))
                    expanded = True
        if not expanded:
            return


def replot_existing_sweep(records, output_dir=None, show=True):
    frame = pd.DataFrame(records).copy()
    required = {'domain', 'optimization_level', 'dd_policy', 'method', 'channel',
                'round', 'trial', 'fidelity', 'chsh_S_direct'}
    if missing := required - set(frame.columns):
        raise ValueError(f'Missing columns: {sorted(missing)}. Use the combined optimization-level records.')
    if frame.empty:
        raise ValueError('No hardware or noisy-simulation records were supplied.')
    keys = ['domain', 'optimization_level', 'dd_policy', 'method', 'channel', 'round', 'trial']
    if frame[keys].isna().any().any():
        raise ValueError('Missing trial, level, policy, channel, method or round labels.')
    if frame.duplicated(keys).any():
        raise ValueError('Duplicate records for the same level/policy/method/channel/round/trial; do not merge repeated runs with reused trial numbers.')
    out = Path(output_dir) if output_dir is not None else (
        Path.cwd() / 'epr_replots' / datetime.now().strftime('%Y%m%d_%H%M%S_%f'))
    out.mkdir(parents=True, exist_ok=True)
    frame.to_csv(out / 'all_domains_records_used.csv', index=False)
    summaries = []
    group_keys = keys[:-1]
    for key, group in frame.groupby(group_keys, sort=True):
        for metric in ['fidelity', 'chsh_S_direct']:
            mean, ci, n = _plot_ci(group[metric])
            exceeds_tsirelson = bool(
                metric == 'chsh_S_direct' and np.isfinite(mean) and mean > TSIRELSON_BOUND
            )
            summaries.append(dict(zip(group_keys, key), metric=metric, mean=mean,
                                  ci99_halfwidth=ci, n_trials=n,
                                  exceeds_tsirelson=exceeds_tsirelson))
    summary = pd.DataFrame(summaries)
    summary.to_csv(out / 'all_domains_metric_summary.csv', index=False)
    slug = lambda value: re.sub(r'[^A-Za-z0-9_.-]+', '_', str(value))
    files = []
    for (domain, level, channel), part in summary.groupby(['domain', 'optimization_level', 'channel'], sort=True):
        methods = list(part['method'].drop_duplicates())
        fig, axes = plt.subplots(len(methods), 2, figsize=(16, 4.5 * len(methods)), squeeze=False)
        labels_by_axis = []
        figure_has_tsirelson_violation = False
        for i, method in enumerate(methods):
            for j, metric in enumerate(['fidelity', 'chsh_S_direct']):
                axis = axes[i, j]
                subset = part.loc[part['method'].eq(method) & part['metric'].eq(metric)]
                rounds = sorted(subset['round'].unique())
                policies = sorted(subset['dd_policy'].unique())
                x = np.arange(len(rounds), dtype=float)
                width = 0.8 / max(len(policies), 1)
                label_entries = []
                for policy_index, policy in enumerate(policies):
                    data = (subset.loc[subset['dd_policy'].eq(policy)]
                            .set_index('round').reindex(rounds))
                    offset = (policy_index - (len(policies) - 1) / 2) * width
                    means = data['mean'].to_numpy(float)
                    errors = data['ci99_halfwidth'].to_numpy(float)
                    bars = axis.bar(x + offset, means, width, label=policy)
                    if metric == 'chsh_S_direct':
                        flags = np.where(np.isfinite(means), means > TSIRELSON_BOUND, False)
                        if flags.any():
                            figure_has_tsirelson_violation = True
                        label_entries.extend(zip(bars, means, errors, flags))
                    else:
                        label_entries.extend(zip(bars, means, errors))
                    available = np.isfinite(errors)
                    axis.errorbar(
                        (x + offset)[available], means[available],
                        yerr=errors[available], fmt='none',
                        color='black', capsize=4, capthick=1.2, elinewidth=1.2, zorder=3)
                labels_by_axis.append((axis, _label_sweep_bar_values(axis, label_entries)))
                axis.set_xticks(x, [str(r) for r in rounds])
                axis.set_xlabel('Teleportation round')
                axis.set_ylabel('Bell fidelity to Phi+' if metric == 'fidelity' else 'CHSH S')
                axis.set_title(f'{method.upper()} | {metric}')
                axis.grid(axis='y', alpha=.25)
                if metric == 'chsh_S_direct':
                    axis.axhline(
                        TSIRELSON_BOUND, color='crimson', linestyle='--',
                        linewidth=1.1, zorder=1,
                    )
                    axis.annotate(
                        'Tsirelson bound 2\u221a2', (axis.get_xlim()[0], TSIRELSON_BOUND),
                        xytext=(2, 3), textcoords='offset points', fontsize=7.5,
                        color='crimson', ha='left', va='bottom', annotation_clip=False,
                    )
                axis.legend(loc='upper left', bbox_to_anchor=(1.01, 1), borderaxespad=0)
        suptitle = (f'{domain} | optimization level {level} | channel {channel}\n'
                    '99% CI across trials where available; one trial has no estimated CI.\n'
                    'CHSH uses four direct measurement settings at every round.')
        if figure_has_tsirelson_violation:
            suptitle += (
                '\n\u2020 chsh_S_direct exceeds the Tsirelson bound (2\u221a2 \u2248 2.828): '
                'per-setting M3 readout mitigation does not guarantee a physical combined '
                'value. Treat \u2020 bars as a mitigation artifact, not evidence of '
                'post-quantum correlations, and refer to chsh_S_from_rho for a physically '
                'bounded estimate.'
            )
        fig.suptitle(suptitle)
        fig.tight_layout(rect=(0, 0, 1, .89))
        _space_sweep_bar_labels(fig, labels_by_axis)
        path = out / f'{slug(domain)}_level_{slug(level)}_channel_{slug(channel)}_metrics.png'
        fig.savefig(path, dpi=160, bbox_inches='tight')
        files.append(path)
        if show:
            plt.show()
        plt.close(fig)
    heatmap_rows = []
    if {'rho_real', 'rho_imag'} <= set(frame.columns):
        heat_keys = ['domain', 'optimization_level', 'dd_policy', 'method', 'channel']
        for key, part in frame.groupby(heat_keys, sort=True):
            rounds = sorted(part['round'].unique())
            labels = []
            for r in rounds:
                subset = part.loc[part['round'].eq(r)]
                labels.append(str(subset.iloc[0]['state_pair_label']) if 'state_pair_label' in part else f'Round {r}')
            by_trial = {trial: {row['round']: _plot_rho(row) for _, row in group.iterrows()}
                        for trial, group in part.groupby('trial')}
            matrix = np.full((len(rounds), len(rounds)), np.nan)
            for i, r1 in enumerate(rounds):
                for j, r2 in enumerate(rounds):
                    paired = [(trial, _plot_state_fidelity(states[r1], states[r2]))
                              for trial, states in by_trial.items() if r1 in states and r2 in states]
                    mean, ci, n = _plot_ci([value for _, value in paired])
                    matrix[i, j] = mean
                    row = dict(zip(heat_keys, key), reference_round=r1, comparison_round=r2,
                               reference_state=labels[i], comparison_state=labels[j],
                               mean_state_fidelity=mean, ci99_halfwidth=ci, n_paired_trials=n)
                    row.update({f'trial_{trial}': value for trial, value in paired})
                    heatmap_rows.append(row)
            fig, axis = plt.subplots(figsize=(8, 7))
            cmap = plt.get_cmap('viridis').copy()
            cmap.set_bad('lightgray')
            image = axis.imshow(np.ma.masked_invalid(matrix), vmin=0, vmax=1, cmap=cmap)
            for i in range(len(rounds)):
                for j in range(len(rounds)):
                    if np.isfinite(matrix[i, j]):
                        axis.text(j, i, f'{matrix[i,j]:.3f}', ha='center', va='center',
                                  color='white' if matrix[i,j] < .55 else 'black')
            axis.set_xticks(range(len(rounds)), labels, rotation=40, ha='right')
            axis.set_yticks(range(len(rounds)), labels)
            domain, level, policy, method, channel = key
            axis.set_title(f'{domain} | Level {level} | {policy} | {method.upper()} | channel {channel}\n'
                           'State fidelity between acquired rounds (paired within trial)')
            fig.colorbar(image, ax=axis, label='Mean state fidelity')
            fig.tight_layout()
            path = out / ('heatmap_' + '_'.join(map(slug, key)) + '.png')
            fig.savefig(path, dpi=160, bbox_inches='tight')
            files.append(path)
            if show:
                plt.show()
            plt.close(fig)
        pd.DataFrame(heatmap_rows).to_csv(out / 'all_domains_state_fidelity.csv', index=False)
    else:
        print('State matrices are absent: metric plots saved; state-fidelity heatmaps skipped.')
    print(f'Saved {len(files)} figures and their data to: {out.resolve()}')
    print('Used existing records only. Plotting submitted no hardware or simulation jobs.')
    return out

replot_folder = None
if 'all_round_records' in globals() and not pd.DataFrame(all_round_records).empty:
    try:
        replot_folder = replot_existing_sweep(all_round_records)
    except Exception as exc:
        _shared_failure('plotting', exc)
        print('Plot generation failed. Acquired CSV records remain available in:', OUTPUT_DIR)
        try:
            _shared_update_status(all_round_records, sum(_SHARED_PLANNED_CONDITIONS_BY_LEVEL.values()))
        except Exception as status_exc:
            _shared_failure('status_export', status_exc)
else:
    print('No acquired records to plot. See acquisition_status.json and recovery_events.jsonl for skipped work.')
```
